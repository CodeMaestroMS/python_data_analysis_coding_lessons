Method ```__new__``` is used to control the creation of new instances. It can be used, for example, to create classes that have only one instance.

While ```__init__``` sets up an already-created instance, ```__new__``` is the (rarer, more advanced) hook that controls the actual creation step itself. Example use — the "Singleton" pattern (a class that only ever allows one instance to exist, no matter how many times you "create" it). This is niche — you likely won't write ```__new__``` often as a beginner, but good to know it exists.
