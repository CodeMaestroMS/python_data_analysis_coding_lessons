Objects and classes

Python is an object-oriented programming language like Java and C++. But unlike Java, Python doesn’t force you to use classes, inheritance and methods. If you like, you can also choose the structural programming paradigm with functions and modules.

Every value in Python is an object. Objects are a way to combine data and the functions that handle that data. This combination is called encapsulation. The data items and functions of objects are called attributes, and in particular the function attributes are called methods. For example, the operator ```+``` on integers calls a method of integers, and the operator ```+``` on strings calls a method of strings.

## Objects and Classes — Simply

This introduces a **new way of organizing code** — one you've actually been using this whole time without knowing its name!

---

### Python Lets You Choose Your Style

> *"Python doesn't force you to use classes... you can also choose the structural programming paradigm with functions and modules."*

Two different **styles** of writing programs:

```
Structural style (what you've mostly done so far):
    functions + modules
    def detect_ranges(L): ...
    def summary(filename): ...

Object-oriented style (what this section introduces):
    classes + objects + methods
    "5".upper()
    my_list.append(3)
```

Some languages (like Java) **force** you to always use classes. Python is flexible — write whichever style fits the problem.

---

### "Every Value in Python Is an Object" — You Already Knew This!

Remember way back — when we discussed namespaces and how `gzip.open` and `os.open` are functions living inside module "boxes"? An **object** is the same box idea, but applied to *every single value*, not just modules.

```python
x = 5
s = "hello"
L = [1, 2, 3]
```

`5`, `"hello"`, and `[1, 2, 3]` are all **objects** — each one is a little bundle that carries **both**:
- **data** (the actual value it holds)
- **functions that work on that data** (attached to it)

---

### "Encapsulation" — Bundling Data + Its Functions Together

```
An object = data + the functions that operate on that data, bundled as ONE thing
```

Think of a string object like a **sealed package**:

```
"hello"  (the object)
    │
    ├── data:    the characters h-e-l-l-o
    └── methods: .upper(), .strip(), .split(), .find(), ...
```

The string doesn't just **hold** the text — it **carries its own toolkit** for working with that text, glued to it. You never have to hunt in some far-away module for a function to uppercase a string — it's **built into the string itself**.

---

### Attributes and Methods — Vocabulary You Half-Know Already

You met "attributes" back in the module lesson (`mod.a`, `mod.f`)! Same word, same dot-notation — just now applied to **any object**, not only modules.

| Term | Meaning |
|---|---|
| **attribute** | anything accessible via the dot — data OR function |
| **method** | an attribute that is specifically a **function** |

```python
s = "hello"

s.upper       # this IS an attribute (a method, before calling it)
s.upper()     # CALLING that method
```

```python
L = [1, 2, 3]
L.append       # attribute (method)
L.append(4)    # calling it
```

So really: **every `.something` you've been writing this whole time — `.split()`, `.strip()`, `.append()`, `.keys()`, `math.cos` — was already "attribute access" on an object.** This section is just giving that pattern its formal name.

---

### The Surprising Reveal — Even `+` Is a Method Call!

> *"the operator `+` on integers calls a method of integers, and the operator `+` on strings calls a method of strings"*

This is the genuinely new insight: operators like `+` **aren't magic symbols** — they're secretly **shortcuts for calling a method**.

```python
5 + 3            # what you write
(5).__add__(3)   # what ACTUALLY happens underneath — calling a METHOD!
```

```python
"hi" + "there"           # what you write
"hi".__add__("there")    # underneath — a DIFFERENT __add__ method, belonging to strings!
```

**Why does `+` behave differently for ints vs strings** (`5+3=8` but `"a"+"b"="ab"`, no error mixing types like lists+strings failed earlier)? Because **each object type carries its own version** of the `+` method — integers have an addition-flavored `+`, strings have a concatenation-flavored `+`. The symbol `+` is just a friendly face; which actual method runs depends entirely on **which object's method it belongs to**.

---

### Why This Matters — It Explains Things You Already Noticed

Remember our very early conversation about `list + list` vs `list + tuple` failing?

```python
[1, 2] + [3, 4]     # ✓ works — list's own + method knows what to do
[1, 2] + (3, 4)      # ✗ TypeError — list's + method doesn't know how to handle a tuple
```

Now you know **why**: `+` is really `list.__add__`, and that particular method was written to only handle "another list" — not "any tuple." The error wasn't Python being arbitrary; it was **that specific object's method** rejecting an input it wasn't designed for.

---

### The Big Picture

```
OBJECT  =  DATA  +  METHODS (functions bundled with the data)

"hello"        →  characters  +  .upper(), .split(), .strip(), __add__, ...
[1,2,3]        →  elements    +  .append(), .sort(), __add__ (list version), ...
5              →  the number  +  __add__ (int version), __sub__, ...
```

**Encapsulation** = the fact that each object carries its own relevant toolset, rather than functions floating around loosely, disconnected from the data they work on.

---

### The One-Sentence Summary

> An **object** bundles data together with the functions (**methods**) that know how to work on that data — this bundling is called **encapsulation**. Everything you access with a dot (`.upper()`, `.append()`, `math.cos`) is an **attribute**; when that attribute is a function, it's specifically called a **method**. Even operators like `+` are secretly shortcuts for calling a method — which is why `+` behaves completely differently depending on which type of object it's attached to. 🎯

Functions, modules, methods, classes, etc are all first class objects. This means that these objects can be

1. stored in a container
2. passed to a function as a parameter
3. returned by a function
4. bound to a variable

One can access an attribute of an object using the dot operator: ```object.attribute```. For example: if ```L``` is a list, we can refer to the method append with ```L.append```. The method call can look, for instance, like this: ```L.append(4)```. Because modules are also objects in Python, we can interpret the expression ```math.pi``` as accessing the data attribute pi of module object math.

Numbers like ```2``` and ```100``` are instances of type ```int```. Similarly, "hello" is an instance of type str. When we write ```s=set()```, we are actually creating a new instance of type set, and bind the resulting instance object to s.

## "First-Class Objects" and "Instances" — Simply

Two important ideas here — let's connect each to things you've already been doing.

---

### Part 1 — "First-Class Objects"

This is a fancy-sounding phrase for something surprisingly simple:

> **A function (or module, class, method...) can be treated exactly like any other value — a number, a string, a list.**

The text lists four "powers" a first-class object has. You've actually used **every single one** already, without the label!

---

### Power 1 — Stored in a Container

```python
functions = [str, int, len]     # a LIST containing FUNCTIONS as its elements!
print(functions[0]("hello"))     # → 'hello'   (calling str() via the list)
```

Remember your `SHAPES` dictionary from way back?
```python
SHAPES = {
    'triangle': (area_of_triangle, ['base', 'height']),
    ...
}
```
`area_of_triangle` — a **function** — was stored as a **value inside a dictionary**. That's exactly this power.

---

### Power 2 — Passed to a Function as a Parameter

```python
def apply_twice(f, x):
    return f(f(x))

print(apply_twice(str.upper, "hi"))   # passing the FUNCTION str.upper as an argument
```

You did this constantly! `map(str, L)`, `sorted(L, key=len)`, `defaultdict(list)` — every one of these hands a **function itself** (not its result) into another function.

---

### Power 3 — Returned by a Function

```python
def get_operation(name):
    if name == "add":
        return lambda a, b: a + b
    else:
        return lambda a, b: a - b

op = get_operation("add")
print(op(3, 4))    # → 7
```

A function can **hand back another function** as its result — same as returning a number or a list.

---

### Power 4 — Bound to a Variable

```python
my_function = print       # 'my_function' now points to the SAME object as 'print'
my_function("hello")       # → hello
```

Remember your earlier discovery with `gzip.open`? *"You can even re-point it: `my_opener = gzip.open`"* — that's this power exactly. A variable name is just a **label**, and it can point at a function object just as easily as at a number.

---

### Why Call It "First-Class"?

Because in **some** languages, functions are treated as **second-class citizens** — special, restricted, unable to be stored in a list or passed around freely. Python (and many modern languages) removes those restrictions: **functions are just objects**, full stop, with all the same rights as any other value.

### Part 2 — The Dot Operator, Reframed

> *"we can interpret the expression `math.pi` as accessing the data attribute `pi` of module object `math`"*

This connects **directly** back to your namespace lesson! Recall the labeled-box diagram — `math` was a box, `pi` and `cos` were names inside it. Now you're getting the **formal vocabulary** for that exact same picture:

```python
math.pi        # accessing the ATTRIBUTE 'pi' of the OBJECT 'math'
L.append        # accessing the ATTRIBUTE 'append' (a METHOD) of the OBJECT 'L'
L.append(4)      # accessing it, then CALLING it
```

**The insight:** `math` isn't special syntax — it's an **object**, exactly like a list or a string is an object. Modules, lists, strings — all objects, all accessed the same dot way.

### Part 3 — "Instances" — The Vocabulary for "One Specific Example Of a Type"

> *"Numbers like 2 and 100 are instances of type `int`"*

**Instance** = *"one particular object, built from a certain type/class."*

Think of it like a cookie cutter analogy:

```
type/class  =  the COOKIE CUTTER shape (the mold, the blueprint)
instance    =  one ACTUAL COOKIE made from that cutter
```

```python
2      # an instance of int
100    # ANOTHER instance of int  (different value, same type)
"hi"   # an instance of str
```

Both `2` and `100` are **different objects**, but they're both **built from the same mold** — the `int` type. That's what "instance of" means.

### `s = set()` — Creating a New Instance

> *"we are actually creating a new instance of type `set`"*

```python
s = set()
```

Break this down using what you now know:

```
set        →  the TYPE (the cookie cutter — the blueprint for "a set")
set()      →  CALLING the type, like calling a function → produces a brand-new INSTANCE
s          →  a variable NAME, bound to (pointing at) that new instance
```

You've secretly been creating instances this whole course!

```python
L = list()      # instance of type list
d = dict()       # instance of type dict
n = int("5")      # instance of type int, built from a string
```

Even `L = [1, 2, 3]` is shorthand for creating a **list instance** — just using the friendlier bracket syntax instead of calling `list()` directly.

---

### Verifying With `type()` and `isinstance()`

```python
print(type(2))            # → <class 'int'>
print(type("hi"))          # → <class 'str'>
print(type(set()))          # → <class 'set'>

print(isinstance(2, int))       # → True    "is 2 AN INSTANCE OF int?"
print(isinstance("hi", int))     # → False
```

`isinstance(obj, Type)` literally asks: *"is `obj` an instance of `Type`?"* — using the exact vocabulary from this section.

### The Big Picture, Connected

```
class/type   =  the blueprint          (int, str, list, set, math's module type...)
instance     =  one object made from that blueprint    (2, "hi", [1,2,3], s)
attribute    =  something accessible via the dot on an instance   (.pi, .append, .upper)
method       =  an attribute that happens to be a function        (.append, .upper)

FIRST-CLASS OBJECT = ANY of these (functions, modules, instances...) can be
                      stored, passed around, returned, and assigned —
                      just like a plain number or string can.
```

---

### The One-Sentence Summary

> "First-class object" means functions (and modules, classes, methods) get treated with the **same freedom** as any ordinary value — storable in lists/dicts, passable as arguments, returnable from functions, and assignable to variables — which you've already been doing (`map(str, L)`, `defaultdict(list)`) without the formal name. Separately, **"instance"** just means "one specific object built from a type" — `2` is an instance of `int`, and `s = set()` literally means *"call the `set` blueprint to manufacture a new set instance, and point the name `s` at it."* 🎯

A user can define their own data types. These are called ```classes```. A user can call these classes like they were functions, and they return a new instance object of that type. Classes can be thought as recipes for creating objects.

````python
class Myclass(object):
    """Documentation string of the class"""

    def __init__(self,param1,param2):
        """This initializes an instance of type ClassName"""
        self.b = param1
        c=param2

    def f(self, param1):
        """This is a method of the class"""
        #some statements

    a=1 # This creates a class attribute
````

The class definition starts with the class statement. With this statement you give a name for your new type, and also in parentheses list the base classes of your class. The next indented block is the class body. After the whole class body is read, a new type is created. Note that no instances are created yet. All the attributes and methods of the class are defined in the class body.

The example class has two methods: ```__init__``` and ```f```. Note that their first parameter is special: self refers to the instance of the class containing the method being called . For instance, if we create an instance ```a = MyClass(..)```, then self represents a the call ```a.f(param1)```. The concept of self is equivalent to that of the this variable of ```C++``` or ```Java```. ```__init__``` does the initialisation when an instance is created. At instantiation with i=MyClass(2,3) the parameters ```param1``` and ```param2``` are bound to values ```2``` and ```3```, respectively. Now that we have an instance ```i```, we can call its method ```f``` with the dot operator: ```i.f(1)```. The parameters of ```f``` are bound in the following way: ```self=i``` and ```param1=1```.

## Defining Your Own Classes — Simply

You've been **using** classes and instances this whole course (`list`, `str`, `set` are all classes!). Now you're learning to **build your own**.

---

### The Recipe Analogy — Take It Literally

> *"Classes can be thought as recipes for creating objects."*

```
class MyClass:          →  the RECIPE (written once)
MyClass(2, 3)            →  FOLLOWING the recipe → produces one COOKIE (an instance)
i = MyClass(2, 3)         →  the cookie gets a NAME: i
```

Same relationship as `list` (recipe) → `list()` or `[1,2,3]` (a cookie/instance) → `L = [1,2,3]` (naming it).

---

### Breaking Down the Class Definition

```python
class MyClass(object):
    """Documentation string of the class"""

    def __init__(self, param1, param2):
        """This initializes an instance of type ClassName"""
        self.b = param1
        c = param2

    def f(self, param1):
        """This is a method of the class"""
        # some statements

    a = 1  # This creates a class attribute
```

Let's label every piece:

| Piece | What it is |
|---|---|
| `class MyClass(object):` | starts the recipe, names it `MyClass` |
| `"""Documentation..."""` | the class's **docstring** — exactly like a function's! |
| `def __init__(self, ...):` | a special method that runs **automatically** when a new instance is created |
| `def f(self, ...):` | an ordinary **method** — a function attached to instances |
| `a = 1` (at class level, not inside a method) | a **class attribute** — shared data, same as `math.pi` |

---

### The Strange First Parameter — `self`

Every method starts with `self`. This is the **biggest new idea** here, so let's slow down.

**`self` = "whichever instance is calling this method right now."**

When you write:

```python
i = MyClass(2, 3)
i.f(1)
```

Python secretly does this translation for you:

```python
MyClass.f(i, 1)
#          ↑  ↑
#        self param1
```

**The instance you called the method ON automatically becomes `self`.** You never type it yourself — Python fills it in for you, based on what's sitting to the left of the dot.

---

### Visualizing It — `self` Is Just a Reference to "This Particular Object"

```
i = MyClass(2, 3)          i  is one instance (one "cookie")
j = MyClass(9, 8)          j  is a DIFFERENT instance (a different cookie)

i.f(1)   →  inside f, self = i     (this call is about i's data)
j.f(5)   →  inside f, self = j     (this call is about j's data)
```

Same method `f`, but `self` changes depending on **which object** you called it through — exactly like `self` = your earlier `variable_name` label pointing at a specific object in memory.

---

### `__init__` — The Setup Method

> *"`__init__` does the initialisation when an instance is created."*

This method runs **automatically**, the instant you "call the recipe":

```python
i = MyClass(2, 3)
```

Behind the scenes:

```
1. Python creates a brand-new, empty instance object
2. Python IMMEDIATELY calls __init__ on it:
       __init__(self=<the new instance>, param1=2, param2=3)
3. Inside __init__:
       self.b = param1     → stores 2 as an ATTRIBUTE of this instance, named 'b'
       c = param2            → just a normal local variable — thrown away after __init__ ends!
4. i now points to the fully set-up instance
```

```python
print(i.b)    # → 2      ✓ accessible — it's an ATTRIBUTE
print(i.c)    # ✗ AttributeError — 'c' was never saved to the instance, just a throwaway local
```

**Critical distinction:** `self.b = param1` **saves** the value onto the object (permanent, accessible later via the dot). `c = param2` (no `self.`) is just an **ordinary local variable** inside the function — it vanishes the moment `__init__` finishes. This is exactly the same local-vs-attribute distinction from your namespace/scope lesson — `c` lives in `__init__`'s temporary local namespace, `self.b` lives in the instance's own permanent namespace.

---

### Calling a Regular Method — `i.f(1)`

```python
i.f(1)
```

Python translates this to:

```python
MyClass.f(i, 1)
#          ↑  ↑
#        self  param1
```

So **inside** `f`, you have access to `self` (which is `i`) — meaning you can reach `self.b` (which is `2`, set earlier in `__init__`):

```python
def f(self, param1):
    """This is a method of the class"""
    print(self.b + param1)    # → 2 + 1 = 3
```

---

### Class Attribute vs Instance Attribute — The `a = 1` Line

```python
class MyClass(object):
    a = 1              # CLASS attribute — defined OUTSIDE any method

    def __init__(self, param1, param2):
        self.b = param1    # INSTANCE attribute — defined INSIDE a method, via self.
```

| | Where defined | Shared or unique? |
|---|---|---|
| **Class attribute** (`a`) | directly inside `class ...:`, no `self.` | **shared** by ALL instances |
| **Instance attribute** (`self.b`) | inside a method, via `self.` | **unique** to each instance |

```python
i = MyClass(2, 3)
j = MyClass(9, 8)

print(i.a, j.a)    # → 1 1     ← SAME class attribute, shared
print(i.b, j.b)    # → 2 9     ← DIFFERENT instance attributes, one per object
```

Think of `a` like a label printed on the **cookie cutter itself** (every cookie made from it shares that same stamp) — while `self.b` is something written **onto each individual cookie** after it's made.

---

### The Full Picture, Traced Through One Example

```python
i = MyClass(2, 3)
```
```
1. Create new, empty instance
2. Call __init__(self=new_instance, param1=2, param2=3)
       self.b = 2        → new_instance now HAS an attribute 'b' = 2
       c = 3              → thrown away, not saved anywhere permanent
3. i now refers to this fully-built instance
```

```python
i.f(1)
```
```
1. Python sees i.f(1) → translates to MyClass.f(self=i, param1=1)
2. Inside f: self is i, so self.b is 2
3. f runs its body (whatever statements are there), using self.b as needed
```

### `self` vs `this` — For Anyone Coming From Other Languages

> *"The concept of `self` is equivalent to that of the `this` variable of C++ or Java."*

Same idea, different name. Python's twist: **`self` isn't automatic magic** — it's an **ordinary parameter**, just conventionally named `self` and conventionally filled in for you by the dot-call syntax. Nothing hidden; it's genuinely just the first argument, always.

---

### The One-Sentence Summary

> A **class** is a recipe; calling it like a function (`MyClass(2, 3)`) produces a new **instance**. Every method automatically receives the calling instance as its first parameter, conventionally named **`self`** — `i.f(1)` is really shorthand for `MyClass.f(i, 1)`. Inside `__init__`, writing `self.b = param1` **permanently attaches** `b` to that specific instance (an *instance attribute*); a variable defined at the class body's top level, outside any method, becomes a *class attribute* **shared by every instance** made from that class. 🎯

There are differences in how an assignment inside a class body creates variables. The attribute a is at class level and is common for all instances of the class ```MyClass```. The variable ```c``` is a local variable of the function ```__init__```, and cannot therefore be used outside the function. The attribute ```b``` is specific to each instance of ```MyClass```. Note that self refers to the current instance. An example: for objects ```x=MyClass(1,0)``` and ```y=MyClass(2,0)``` we have ```x.b != y.b```, but ```x.a == y.a```.

All methods of a class have a mandatory first parameter which refers to the instance on which you called the method. This parameter is usually named ```self```. If you want to access the class attribute a from a method of the class, use the fully qualified form ```MyClass.a```. The methods whose names both begin and end with two underscores are called special methods. For example, ```__init__``` is a special method. These methods will be discussed in detail later.

## Accessing a Class Attribute From Inside a Method — Simply

This is a small but important rule about **which name to use** when a method needs to reach the shared class attribute.

---

### The Setup

```python
class MyClass:
    a = 1                       # class attribute — shared by everyone

    def __init__(self, param1, param2):
        self.b = param1

    def f(self, param1):
        print(MyClass.a)        # ← accessing the class attribute FROM INSIDE a method
```

---

### Why Not Just Write `self.a`?

You might expect `self.a` to work — and actually, it **does** work, sort of, but for a subtle and slightly dangerous reason. Let's see why the text recommends `MyClass.a` instead.

```python
i = MyClass(2, 3)
print(i.a)    # → 1     ✓ this actually works!
```

**Why does this work even though `a` was never set via `self.a = ...`?** Because when Python looks up `i.a`, it follows a search order:

```
1. Does the INSTANCE i have its own attribute 'a'?     → No
2. Does i's CLASS (MyClass) have an attribute 'a'?      → Yes! → use that
```

So `self.a` (or `i.a`) **falls back** to the class attribute when the instance doesn't have its own. It looks like it worked — but this creates a trap.

---

### The Trap — `self.a = ...` Doesn't Update the Shared Value

```python
i = MyClass(2, 3)
i.a = 99          # looks like it changes 'a'... but does it REALLY change the SHARED one?

print(i.a)          # → 99
print(MyClass.a)     # → 1     ← still the ORIGINAL! Unchanged!
```

**What actually happened:** `i.a = 99` didn't modify the shared class attribute at all. It **created a brand-new instance attribute** named `a`, belonging **only to `i`**, which now **shadows** the class attribute whenever you look it up through `i`.

```
Before:              After i.a = 99:
MyClass.a = 1          MyClass.a = 1        ← unchanged, shared
i (no own 'a')          i.a = 99             ← i now has its OWN private 'a'!

i.a  → found on i itself → 99
j.a  → j has no own 'a' → falls back to MyClass.a → 1   (other instances unaffected!)
```

So using `self.a` to try to **change** the shared value silently fails to do what you probably intended — it just creates a shadow copy on that one instance instead.

---

### Why `MyClass.a` Is the Safe, Explicit Way

Writing `MyClass.a` inside a method **removes all ambiguity** — you're saying, in plain terms: *"I specifically mean the class's shared attribute, not whatever an instance might have."*

```python
class MyClass:
    a = 1

    def increment(self):
        MyClass.a += 1     # explicitly modifies the SHARED class attribute
```

```python
i = MyClass(2, 3)
j = MyClass(9, 8)

i.increment()
print(MyClass.a)    # → 2
print(j.a)             # → 2      ← EVERY instance sees the change!
```

Compare with the trap version:

```python
def increment_wrong(self):
    self.a += 1          # this READS MyClass.a (via fallback), then creates i's OWN 'a'!
```

```python
i.increment_wrong()
print(MyClass.a)    # → 1      ← unchanged!
print(i.a)              # → 2      ← only i changed
print(j.a)              # → 1      ← j is untouched — surprising if you expected shared behavior!
```

---

### The Rule of Thumb

| You want to... | Write |
|---|---|
| **Read** a class attribute (rarely goes wrong either way) | `self.a` or `MyClass.a` both usually work |
| **Modify** a class attribute (must be explicit) | `MyClass.a = ...` — the ONLY reliable way |
| **Set a per-instance value** (deliberately) | `self.a = ...` — correct if you actually WANT a private copy |

---

### Simple Analogy

> Think of the class attribute like a **shared office whiteboard** 📋, and each instance like an **employee**.
>
> - Reading the board (`self.a`) — any employee can glance at the shared whiteboard if they don't have their own sticky note. Works fine.
> - Writing on it via `self.a = ...` — this is like the employee **sticking their own private note over the whiteboard**, instead of writing on the real board. They now see *their* note, but the actual whiteboard — and everyone else looking at it — never changed.
> - Writing via `MyClass.a = ...` — this is walking up and **actually editing the shared whiteboard** — visible to everyone.

---

### The One-Sentence Summary

> Reading a class attribute through `self.a` often *appears* to work (Python falls back to the class if the instance has none of its own), but **assigning** through `self.a = ...` doesn't update the shared value — it silently creates a separate, private attribute on that one instance instead. To reliably read *or* modify the actual shared class attribute, refer to it explicitly by the class's own name: `MyClass.a`. 🎯
