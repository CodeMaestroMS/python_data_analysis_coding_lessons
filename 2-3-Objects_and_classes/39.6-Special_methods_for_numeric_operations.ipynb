### Numeric Special Methods — The Full `+` Story, Completed

Remember: *"`+` on integers calls a method of integers, `+` on strings calls a method of strings."* Now you can **write your own version**:

```python
class Vector:
    def __init__(self, x, y):
        self.x, self.y = x, y

    def __add__(self, other):
        return Vector(self.x + other.x, self.y + other.y)

    def __repr__(self):
        return f"Vector({self.x}, {self.y})"

v1 = Vector(1, 2)
v2 = Vector(3, 4)
print(v1 + v2)     # → Vector(4, 6)
```

`v1 + v2` runs `v1.__add__(v2)` — exactly the same mechanism as `int`'s own `+`, `str`'s own `+`, and `list`'s own `+` — you're just now **supplying your own version** for a brand-new type you designed.

## The `complex` Type — Simply

`complex` is a **built-in number type** in Python, representing **complex numbers** — numbers with a "real" part and an "imaginary" part, from mathematics.

---

### Quick Math Refresher — What IS a Complex Number?

A complex number has two components:

```
z = a + bi
```

- `a` = the **real** part (an ordinary number)
- `b` = the **imaginary** part
- `i` = the special value where `i² = -1` (something no ordinary/"real" number can satisfy — hence "imaginary")

Complex numbers exist specifically to let you take the **square root of negative numbers**, which ordinary real numbers can't do:

```
√(-1) = i        (undefined for real numbers, but defined for complex numbers)
```

---

### Complex Numbers in Python — Using `j` Instead of `i`

Python (following an engineering convention rather than a math one) writes the imaginary unit as **`j`**, not `i`:

```python
z = 3 + 4j
print(z)             # → (3+4j)
print(type(z))         # → <class 'complex'>
```

`3` is the real part, `4` is the imaginary part (`4j` means "4 times the imaginary unit").

---

### The `complex()` Function — Building One Explicitly

```python
complex(3, 4)       # → (3+4j)      real=3, imaginary=4
complex(5)            # → (5+0j)      just a real number, imaginary defaults to 0
complex("3+4j")        # → (3+4j)      also parses from a string!
```

Same idea as `int()`, `float()`, `str()` — a **type constructor**: call it, get a new instance of that type.

---

### Accessing the Parts

```python
z = 3 + 4j

z.real        # → 3.0     (always returned as a float, even though we typed int-looking values)
z.imag        # → 4.0
```

---

### Basic Arithmetic — Works Naturally

```python
a = 1 + 2j
b = 3 + 4j

print(a + b)     # → (4+6j)      real parts add, imaginary parts add
print(a * b)     # → (-5+10j)    follows complex multiplication rules
```

These arithmetic operations are powered by the **exact same special methods** you just learned — `__add__`, `__mul__` — except `complex`'s own built-in versions know the actual math rules for combining real and imaginary parts correctly.

---

### `abs()` on a Complex Number — Gives the "Magnitude"

```python
z = 3 + 4j
print(abs(z))     # → 5.0
```

This should look **very familiar**! `abs()` on a complex number computes exactly your earlier `hypotenuse`/vector-length formula:

```
|z| = √(real² + imag²) = √(3² + 4²) = √25 = 5
```

Complex numbers are essentially **2D vectors in disguise** — `real` and `imag` behave like the `x` and `y` components from your earlier vector discussion, and `abs()` gives the same "length" you computed with `math.hypot()`.

---

### Connecting to `__complex__` — Your Original Question

Recall from the special methods lesson:

```python
complex(x)     # secretly runs   x.__complex__()
```

If you built your own class and wanted `complex(my_object)` to work sensibly, you'd define `__complex__` on it — telling Python **how** to turn your custom object into a complex number, the same way `__int__` and `__float__` let `int()`/`float()` work on custom types.

```python
class Signal:
    def __init__(self, amplitude, phase):
        self.amplitude = amplitude
        self.phase = phase

    def __complex__(self):
        import math
        real = self.amplitude * math.cos(self.phase)
        imag = self.amplitude * math.sin(self.phase)
        return complex(real, imag)

s = Signal(5, 0)
print(complex(s))    # → (5+0j)     ← calls s.__complex__() automatically!
```

---

### When Would You Actually Use `complex`?

Not everyday coding — but genuinely important in specific domains:
- **Electrical engineering** — AC circuit analysis uses complex numbers extensively (this is why Python uses `j` — matching engineering notation)
- **Signal processing** — Fourier transforms, audio/frequency analysis
- **Certain physics/math simulations** — quantum mechanics, wave equations
- **Some algorithms** — like the Mandelbrot set (fractal generation) rely on complex number iteration

For everyday programming (file processing, web apps, data wrangling), you'll rarely touch `complex` — but it's there, built-in, whenever a domain genuinely calls for it.

---

### The One-Sentence Summary

> `complex` is a built-in numeric type representing numbers with a real and imaginary part (`3+4j`), Python's version of the mathematical concept needed to handle things like square roots of negative numbers; `complex(x)` constructs one, `.real`/`.imag` access its parts, and `abs()` computes its magnitude using the exact same `√(a²+b²)` formula you already know from vector length — and just like `int`/`float`, you can make your own classes convert into it by defining `__complex__`. 🎯

In addition to the normal methods of containers, like the ```append``` method of the ```list```, there are several operations that are handled by calls to special methods of the container class.

The test whether ```x``` is a member of container ```c``` is done by the operation ```x in c```.

The corresponding special method call is ```x.__contains__(y)```. Deletion of an element of container c can be done with the operation ```del c[key]```. This will result in the method call ```x.__delitem__```.

## What "Container" Means Here — Simply

A **container** is just the formal name for a category you already know deeply — you've been using containers this entire course, since your very first lessons on lists!

---

### The Definition — Anything That "Holds" Multiple Values

> **Container** = an object whose whole purpose is to **hold a collection of other values** inside it.

```python
[1, 2, 3]         # a container — HOLDS three numbers
(1, 2, 3)          # a container — HOLDS three numbers
{1, 2, 3}          # a container — HOLDS three numbers
{"a": 1, "b": 2}    # a container — HOLDS key-value pairs
"hello"              # a container — HOLDS five characters!
```

**List, tuple, set, dict, and even string** are all containers. This is actually a **broader** umbrella than "sequence" (which you learned earlier) — sequences (list, tuple, str) are containers that are ALSO ordered/indexable, while sets and dicts are containers **without** that ordering guarantee.

---

### Why the Text Uses "Container" Instead of "List" Specifically

Because this section is describing behavior that applies to **all** of them at once — not just lists. The rules being described (`in`, `del`) are things every container-like type supports, each in its own way:

```python
5 in [1, 2, 3, 4, 5]         # ✓ works — list is a container
5 in (1, 2, 3, 4, 5)          # ✓ works — tuple is a container
5 in {1, 2, 3, 4, 5}           # ✓ works — set is a container
5 in {5: "five", 6: "six"}      # ✓ works — dict is a container (checks KEYS)
"e" in "hello"                    # ✓ works — string is a container
```

By saying "container" rather than "list," the text is telling you: *"this concept applies to the whole family, not just one member of it."*

---

### Connecting to Your Special Methods Lesson

This is the **same pattern** you just learned for `+`, `==`, `int()` — a piece of everyday syntax secretly triggers a method call on the object:

```python
x in c          →   secretly calls   c.__contains__(x)
del c[key]        →   secretly calls   c.__delitem__(key)
```

**`in` and `del` aren't special-cased for lists.** They work through the **same universal mechanism** as everything else — Python calls a specifically-named method (`__contains__`, `__delitem__`) on whatever container object you're using, and each container type provides its own version.

---

### Proving It — Call the Method Directly

```python
L = [1, 2, 3, 4, 5]

print(5 in L)                # → True
print(L.__contains__(5))       # → True    ← SAME thing, called directly!
```

```python
d = {"a": 1, "b": 2}
del d["a"]
print(d)                        # → {'b': 2}

d2 = {"a": 1, "b": 2}
d2.__delitem__("a")               # ← calling it manually — same effect!
print(d2)                          # → {'b': 2}
```

---

### Why This Design Matters — Consistency Across ALL Containers

Remember from your `zip()` and `sum()` lessons — Python's philosophy is: functions and operators **don't check what type they're given**, they just call the appropriate special method, and **each type defines its own correct behavior** for that method:

```
x in c    always calls c.__contains__(x)   — but WHAT that check means differs per type:
    list       → "is x one of my elements?"     (scans through)
    dict         → "is x one of my KEYS?"          (fast hash lookup)
    string        → "is x a SUBSTRING of me?"        (character search)
    set             → "is x one of my elements?"       (fast hash lookup)
```

Same syntax (`in`), same underlying mechanism (`__contains__`), but each container type implements the check in whatever way makes sense **for that specific type of data**.

---

### If You Build Your OWN Container Class

Because of this design, **you** can define `__contains__` and `__delitem__` on a class **you** write, and suddenly `in` and `del` will work naturally on your custom objects too — same idea as defining `__add__` to make `+` work on your `Vector` class:

```python
class NameList:
    def __init__(self, names):
        self.names = names

    def __contains__(self, name):
        return name.lower() in [n.lower() for n in self.names]   # case-insensitive!

    def __delitem__(self, name):
        self.names = [n for n in self.names if n.lower() != name.lower()]


people = NameList(["Alice", "Bob", "Carol"])

print("alice" in people)     # → True    ← custom __contains__ runs, case-insensitive!

del people["Bob"]
print(people.names)             # → ['Alice', 'Carol']    ← custom __delitem__ ran!
```

### The One-Sentence Summary

> **"Container"** is the general category name for **any object that holds a collection of values** — list, tuple, set, dict, and string all qualify. The text uses this broader term (rather than "list" specifically) because the rules it's describing — `in` secretly calling `__contains__`, `del` secretly calling `__delitem__` — apply universally across the **whole** container family, with each specific type providing its own sensible version of that check. 🎯

Reading an item of a container ```c``` is done with the operation ```c[key]```. The corresponding method call is ```c.__getitem__(key)```.

Similarly, setting an item with ```c[key]=value``` results in the call ```c.__setitem__(key,value)```.

The number of elements in a container ```c``` can be queried with the function call ```len(c)```. This function call actually calls the special method ```c.__len__```.

The call ```iter(c)``` will call the special method ```__iter__```.
