```python
class Rational(object):

    def __init__(self, numerator, denominator):
        self.numerator = int(numerator)
        self.denominator = int(denominator)

    def __add__(self, ra2):
        d = self.denominator * ra2.denominator
        n = ((d/self.denominator) * self.numerator) + ((d / ra2.denominator) * ra2.numerator)
        return Rational(n, d)

    def __sub__(self, ra2):
        d = self.denominator * ra2.denominator
        n = ((d/self.denominator) * self.numerator) - ((d / ra2.denominator) * ra2.numerator)
        return Rational(n, d)

    def __mul__(self, ra2):
        return Rational(self.numerator * ra2.numerator, self.denominator * ra2.denominator)

    def __truediv__(self, other):
        return Rational(self.numerator * other.denominator, self.denominator * other.numerator)

    def __eq__(self, other):
        return self.numerator == other.numerator and self.denominator == other.denominator

    def __lt__(self, other):
        com_det = self.denominator * other.denominator
        self_r =  ((com_det/self.denominator) * self.numerator)/com_det
        other_r = ((com_det/other.denominator) * other.numerator)/com_det
        return self_r < other_r

    def __gt__(self, other):
        com_det = self.denominator * other.denominator
        self_r =  ((com_det/self.denominator) * self.numerator)/com_det
        other_r = ((com_det/other.denominator) * other.numerator)/com_det
        return self_r > other_r


    def __repr__(self):
        return str(self.numerator) + '/' + str(self.denominator)
        
```

```python
r1=Rational(2,3)
r2=Rational(1,2)
```

```python
r1+r2
```

```python
r1-r2
```

```python
r1*r2
```
