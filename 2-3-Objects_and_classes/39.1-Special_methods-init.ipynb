Special Methods — Simply

You already know the most famous one: __init__! This section reveals that Python has a whole family of these — and remember __add__? You met it earlier when we discussed how 5 + 3 secretly calls (5).__add__(3). This section is where that idea gets fully explained.

The Big Pattern — "Dunder" Methods Are Hooks

Special methods (also called "dunder" methods, for double underscore) are Python's way of letting your own classes plug into built-in syntax. Each one is a hook: "when the user does X, call this method automatically."
```
User writes:          Python secretly calls:
print(obj)         →   obj.__str__()
x + y               →   x.__add__(y)
x == y               →   x.__eq__(y)
len(obj)              →   obj.__len__()
```
You're not calling these directly — you write everyday syntax (+, print, ==), and Python translates it into a method call behind the scenes.

### ```__init__```

We have already encountered one special method, namely the ```__init__``` method.

This method sets the instance attributes to some initial value. Its first parameter is self, and the subsequent parameters are the ones that were passed to the call of the class.

The ```__init__``` method should return no value. Next the main general purpose special methods are introduced. They are executed when certain operations on objects are performed.
