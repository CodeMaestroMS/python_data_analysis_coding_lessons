**Indentation**

Repetition is possible with the for loop. Note that the body of for loop is indented with a tabulator or four spaces. Unlike in some other languages, braces are not needed to denote the body of the loop. When the indentation stops, the body of the loop ends.

````python
for i in range(3):
    print('Hello')
print('Bye!')
````

Indentation applies to other compound statements as well, such as bodies of functions, different branches of an if statement, and while loops. We shall see examples of these later.

## Compound Statements in Python

A **compound statement** is any statement that contains a **header** and a **body** (a block of code indented beneath it).

The general structure always looks like this:
```python
header:        # ends with a colon
    body       # indented block
    body
```

---

### The three mentioned in your text:

**1. Functions**
```python
def greet(name):
    message = "Hello " + name   # body
    print(message)              # body
```

**2. if statements**
```python
if temperature > 30:
    print("It's hot")    # branch 1 body
else:
    print("It's cool")   # branch 2 body
```

**3. while loops**
```python
while count < 5:
    print(count)    # body
    count += 1      # body
```

---

### What makes them "compound"?

They are called compound because they **group multiple statements together** under one controlling header:

```
header →  if temperature > 30:
body   →      print("It's hot")
body   →      print("Stay hydrated")
body   →      fan_speed = "high"
```

The indentation is what tells Python **where the body starts and ends** — unlike languages like Java or C that use `{}` curly braces for this.

The range(3) expression above actually results with the sequence of integers 0, 1, and 2.

So, the range is a ```half-open interval``` with the end point excluded from the range.

In general, expression range(n) gives integers 0, 1, 2, ..., n-1 . ``` [0, n)```

```range(n) = X =  {x | x ∈ 0<x<(n-1) , n ∈ ℝ}```

## Half-Open Interval

An interval is just a **range of numbers between two points**. The "openness" describes **whether the endpoints are included or excluded**.

There are three kinds:

### Closed Interval `[a, b]`
**Both** endpoints are **included**:
```
[0, 3] → 0, 1, 2, 3
         ↑           ↑
      included    included
```

### Open Interval `(a, b)`
**Both** endpoints are **excluded**:
```
(0, 3) → 1, 2
          ↑     ↑
       excluded excluded
```

### Half-Open Interval `[a, b)`
**One included, one excluded** — this is what `range()` uses:
```
[0, 3) → 0, 1, 2
          ↑      ↑
       included excluded
```

---

### How it maps to `range(3)`

```python
range(3)  →  [0, 3)  →  0, 1, 2
```
- Start `0` is **included** (closed side `[`)
- End `3` is **excluded** (open side `)`)

And with explicit start: `range(2, 5)` → `[2, 5)` → `2, 3, 4`

---

### Why does Python do this?

It's a practical design choice — it makes many things cleaner, for example the **length is simply `end - start`**:
```
range(2, 5) → 5 - 2 = 3 items ✓
```
