A string is a sequence of characters commonly used to store input or output data in a program. The characters of a string are specified either between single``` (')``` or double ```(")``` quotes. This optionality is useful if, for example, a string needs to contain a quotation mark: ```"I don't want to go!"```. You can also achieve this by escaping the quotation mark with the backslash: ```'I don\'t want to go'```.

There are multiple ways to do string interpolation:

1. Python format strings
2. the format method
3. f-strings

````python
print('%i plus %i is equal to %i' % (1,3,4))  # Python format strings
````

````python
print('{} plus {} is equal to {}'.format(1,3,4)) # Format method
````

````python
print(f'{1} plus {3} is equal to {4}')  # f-string
````

The ```%``` format syntax uses ```%i``` for integers and ```%f``` for floats (```%d``` also works for integers — it's a synonym in this system only).

The ```.format()``` method and ```f-strings``` share the same rules as each other: they use ```d``` for ```integers``` and ```f``` for ```floats```, written inside the ```{...}``` after a colon, like ```{:d}``` or ```{:f}```.

In practice, specifiers are usually left out — ```f"{4}"``` displays fine without one. You add ```d``` or ```f``` mainly when applying extra formatting like width or decimal places, e.g. ```f"{4:3d}"``` (pad to 3 characters) or ```f"{4.5:.2f}"``` (2 decimal places).

It is often useful to specify the number of decimals when printing floats:

````python
print('%.1f %.2f %.3f' %(1.6,1.7,1.6))

# First use print('%f %f %f' %(1.6,1.7,1.6)) --> Normal
# Then use print('%.f %.f %.f' %(1.6,1.7,1.6))
# Finally use print('%.1f %.2f %.3f' %(1.6,1.7,1.6))

print('{:.1f},{:.2f},{:.3f}'.format(1.6,1.7,1.8))
# First use print('{:f},{:f},{:f}'.format(1.6,1.7,1.8))
# Then use print('{:.f},{:.f},{:.f}'.format(1.6,1.7,1.8))
# Finally use print('{:.1f},{:.2f},{:.3f}'.format(1.6,1.7,1.8))

print(f'{1.60000:.1f},{1.70000:.2f},{1.80000:.3f}')
# First use print(f'{1.60000:f},{1.70000:f},{1.80000:f}')
# Then use print(f'{1.60000:.f},{1.70000:.f},{1.80000:.f}')
# Finally use print(f'{1.60000:.1f},{1.70000:.2f},{1.80000:.3f}')
````

The specifier ```s``` is used for strings. An example:

````python
print('%s concatenated with %s produces %s' % ('water','melon','water'+'melon'))
print('{0} concatenated with {1} produces {0}{1}'.format('water','melon'))
print(f'{'water'} concatenated with {'melon'} produces {'water+melon'}')
````

## Left Padding with f-strings

Here's the general format syntax:

```
f'{value : [fill] [align] [width] [.precision] [type]}'
```

---

### Your current format:
```python
print(f'{1.6:.1f}')  # → 1.6
```
This only controls **decimal precision** (1 decimal place).

---

### Adding left padding (right align is default for numbers):
```python
print(f'{1.6:10.1f}')   # → "       1.6"  (10 chars wide)
```
The number `10` means **total width of 10 characters**, and the value gets pushed to the right by default.

---

### You can also control alignment explicitly:

```python
print(f'{1.6:>10.1f}')  # →        1.6   right aligned (default)
print(f'{1.6:<10.1f}')  # → 1.6          left aligned
print(f'{1.6:^10.1f}')  # →     1.6      center aligned
```

---

### And fill the padding with a custom character:
```python
print(f'{1.6:0>10.1f}')  # → 000000001.6  (fill with zeros)
print(f'{1.6:*>10.1f}')  # → *******1.6   (fill with stars)
```

---

### Reading the format `10.1f` simply:
```
f'{1.6 : 10 . 1 f}'
          ↑    ↑  ↑
       width  1  float
          10 decimal
         chars place
```

> The **width comes before the dot**, the **precision comes after the dot**.

## General Syntax for All 3 Formatting Styles

---

### 1. Format Syntax `%`
```
"[text] %[flags][width][.precision]type" % (values)
```
| Symbol | Meaning |
|---|---|
| `%d` or `%i` | integer |
| `%f` | float |
| `%s` | string |
| `%e` | scientific notation |

```python
"%d"       % (42)          # → 42
"%10d"     % (42)          # → "        42"  (width 10)
"%.2f"     % (3.14159)     # → "3.14"        (2 decimals)
"%10.2f"   % (3.14159)     # → "      3.14"  (width 10, 2 decimals)
"%s"       % ("hello")     # → "hello"
```

---

### 2. Format Method `.format()`
```
"[text] {[index]:[fill][align][width][.precision][type]}".format(values)
```
| Symbol | Meaning |
|---|---|
| `{}` | auto index |
| `{0}` | positional index |
| `{name}` | named argument |

```python
"{}"          .format(42)            # → 42
"{:10d}"      .format(42)            # → "        42"  (width 10)
"{:.2f}"      .format(3.14159)       # → "3.14"
"{:10.2f}"    .format(3.14159)       # → "      3.14"
"{:>10.2f}"   .format(3.14159)       # → "      3.14"  (right align)
"{:<10.2f}"   .format(3.14159)       # → "3.14      "  (left align)
"{:*^10.2f}"  .format(3.14159)       # → "***3.14***"  (center, fill *)
"{0} {1}"     .format("hi","world")  # → "hi world"    (positional)
"{x}"         .format(x=42)          # → 42            (named)
```

---

### 3. f-string `f""`
```
f"[text] {[expression]:[fill][align][width][.precision][type]}"
```
```python
f'{42}'            # → 42
f'{42:10d}'        # → "        42"  (width 10)
f'{3.14159:.2f}'   # → "3.14"
f'{3.14159:10.2f}' # → "      3.14"
f'{3.14159:>10.2f}'# → "      3.14"  (right align)
f'{3.14159:<10.2f}'# → "3.14      "  (left align)
f'{3.14159:*^10.2f}'# → "***3.14***" (center, fill *)
```

---

### Quick Comparison of All 3:
```python
# integer
"%d"        % 42           # % style
"{:d}".format(42)          # .format()
f'{42:d}'                  # f-string

# float with 2 decimals
"%.2f"      % 3.14         # % style
"{:.2f}".format(3.14)      # .format()
f'{3.14:.2f}'              # f-string

# width 10, 2 decimals
"%10.2f"    % 3.14         # % style
"{:10.2f}".format(3.14)    # .format()
f'{3.14:10.2f}'            # f-string
```

---

### The format specifier is the same for all 3:
```
[fill][align][width][.precision][type]
  *      >      10       .2        f
```
> Once you learn the format specifier pattern, it works the **same way across all 3 styles** — the only difference is the syntax wrapping around it.

## Checking If a String Is Empty — The Best Way

```python
if not s:
    print("empty!")
```

This is the **standard, Pythonic** way — and it's what you've actually already used in this very conversation (remember `if not line:` in your `file_extensions` exercise, and `if line.strip():` filtering blank lines earlier).

---

### Why `not s` Works — Truthiness

In Python, every value has a **truthy or falsy** nature when used in a boolean context. An empty string is specifically defined as **falsy**:

```python
bool("")        # → False    empty string is falsy
bool("hello")   # → True     non-empty string is truthy
```

So:
```python
if not s:        # "if s is falsy" → true when s == ""
    ...
```

---

### Comparison of Approaches

```python
s = ""

# ✓ Best — Pythonic
if not s:
    print("empty")

# ✓ Also correct, but more verbose
if s == "":
    print("empty")

# ✗ Works, but unnecessarily indirect
if len(s) == 0:
    print("empty")
```

All three give the same result **for a real string** — but `not s` is preferred because it's shorter, reads naturally ("if not s" = "if s is nothing"), and matches Python's general idiom of using truthiness rather than explicit comparisons (same reasoning behind preferring `if L:` over `if len(L) > 0:` for lists).

---

### The Important Distinction — "Empty" vs "Whitespace-Only"

Watch out — `not s` only catches a **truly empty** string (`""`), **not** a string full of spaces:

```python
s = "   "        # spaces only — NOT technically empty!

print(not s)       # → False!   "   " is actually TRUTHY (it has characters)
print(s == "")      # → False
print(len(s))         # → 3       has length, just no visible content
```

If you want to treat "just whitespace" as effectively empty too — which is very common when checking user input or file lines — combine with `.strip()` first:

```python
if not s.strip():
    print("empty or just whitespace")
```

```python
s = "   "
print(not s.strip())     # → True   ✓ now correctly caught
```

This is **exactly** the pattern you used in your `file_listing` and `red_green_blue` exercises: `if not line.strip():` — catching both genuinely empty lines and whitespace-padded "blank" lines.

---

### The `None`-Safety Bonus

`not s` has a nice extra benefit: it also correctly handles the case where `s` might be `None` instead of a string:

```python
s = None
if not s:          # → True — catches None too!
    print("nothing to process")
```

```python
if s == "":         # ✗ this would be False for None — misses the case!
    ...
```

If a variable might sometimes be `None` (e.g., a function that returns `None` when nothing was found) and sometimes an empty string, `not s` handles **both** gracefully in one check — a real advantage over strict equality.

---

### Comparison Table

| Check | Catches `""` | Catches `"   "` (whitespace) | Catches `None` |
|---|---|---|---|
| `not s` | ✓ | ✗ | ✓ |
| `s == ""` | ✓ | ✗ | ✗ |
| `len(s) == 0` | ✓ | ✗ | ✗ (crashes on None!) |
| `not s.strip()` | ✓ | ✓ | ✗ (crashes on None!) |

Note: `len(None)` and `None.strip()` both raise errors — so if `None` is a real possibility, check `not s` **first** (short-circuit), or guard explicitly.

---

### The One-Sentence Summary

> Use **`if not s:`** — it's the idiomatic Python way, relies on the fact that an empty string is falsy, reads naturally, and as a bonus also catches `None`. If you need to treat whitespace-only strings as "empty" too (very common for file/user-input processing), combine it with `.strip()` first: **`if not s.strip():`**. 🎯
