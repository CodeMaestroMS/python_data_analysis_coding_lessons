Statements

Statements are commands that have some effect. For example, a function call (that is not part of another expression) is a statement. Also, the variable assignment is a statement:

````python
i=5
i=i+1
i+=1
````

Note that in Python there are no operators ```++``` or ```--``` unlike in some other languages.

It turns out that the operators ```+ - * / // % & | ^ >> << **``` have the corresponding augmented assignment operators ```+= -= *= /= //= %= &= |= ^= >>= <<= **=```

Another large set of statements is the flow-control statements such as if-else, for and while loops. We will look into these in the next sections.

## Augmented Assignment Operators — Simply

You already met this pattern with sets (`|= &= -= ^=`) — this is the **general rule** those were following. Every binary operator in Python has a shortcut that combines "do the operation" with "save the result back."

---

### The Core Idea

Two ways to say the same thing:

```python
x = x + 5     # long way: compute x+5, then reassign to x
x += 5        # short way: same result, one step
```

`x += 5` reads as: *"increase x by 5, in place."*

---

### The Full List — Same Symbol, Just Add `=`

| Operator | Augmented | Means |
|---|---|---|
| `+` | `+=` | add |
| `-` | `-=` | subtract |
| `*` | `*=` | multiply |
| `/` | `/=` | divide |
| `//` | `//=` | floor divide |
| `%` | `%=` | modulo (remainder) |
| `**` | `**=` | exponent (power) |
| `&` | `&=` | bitwise AND |
| `\|` | `\|=` | bitwise OR |
| `^` | `^=` | bitwise XOR |
| `>>` | `>>=` | right shift |
| `<<` | `<<=` | left shift |

**One rule to remember them all:** take any operator, glue an `=` right after it — that's the augmented version.

---

### Examples for the Familiar Ones

```python
x = 10

x += 3      # x = x + 3   → 13
x -= 3      # x = x - 3   → 10
x *= 2      # x = x * 2   → 20
x /= 4      # x = x / 4   → 5.0
x //= 2     # x = x // 2  → 2.0
x %= 2      # x = x % 2   → 0.0
x **= 3     # x = x ** 3  → 0.0
```

Same pattern you already know from `s |= t` for sets — just applied to plain numbers instead.

---

### The Less Familiar Ones — Bitwise Operators

These act on the **binary representation** of integers, bit by bit (remember your bytes/bits lesson — 0s and 1s):

```python
x = 5      # binary: 0101
x &= 3     # AND with 3 (binary 0011) → 0001 = 1
x |= 8     # OR with 8  (binary 1000) → 1001 = 9
x ^= 1     # XOR with 1              → 1000 = 8
x <<= 2    # shift left 2 places      → multiplies by 4
x >>= 1    # shift right 1 place      → divides by 2 (integer)
```

You don't need these often in everyday code — mostly they show up in low-level programming, flags/permissions, or performance tricks. But now you know the naming pattern applies to them too.

---

### Why Use the Short Form?

**1. Less typing, less repetition** — you don't write the variable name twice:
```python
total = total + item_price     # variable name appears twice
total += item_price             # appears once
```

**2. Same efficiency benefit as `s |= t` for sets** — for mutable types, the augmented form can modify the object **in place** instead of building a brand new one:
```python
L = [1, 2, 3]
L += [4]        # extends the SAME list object (like list.extend)
L = L + [4]     # creates a NEW list, then reassigns
```
For numbers (immutable), there's no such memory benefit — but the shorter syntax habit stays consistent everywhere.

---

### The One-Sentence Summary

> **Take any binary operator, put `=` right after it — that's the augmented assignment form.** It's shorthand for "recompute this variable using itself and something else, then save the result back to the same name" — exactly the same idea you already learned for sets (`|=`, `&=`, `-=`, `^=`), just extended to every operator Python has, including the bitwise ones. 🎯
