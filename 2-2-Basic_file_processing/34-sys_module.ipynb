sys module

We saw above that the sys module contains the three file objects.

1. ```sys.stdin```
2. ```sys.stdout```
3. ```sys.stderr```

It has also few other useful attributes.

- The attribute ```sys.path``` is the list of folders that Python uses to look for imported modules.
- The list ```sys.argv``` contains the so called command line parameters.

  For example in Linux if you are using the terminal, then you can run your program with the command ```python3 programname.py param1 param2 ....``` After Python has started your program, the command line parameters are visible as follows.

  The name of the program is in ```sys.argv[0]```. The rest of the command line parameters are after the program name in this list: ```sys.argv[1]=="param1", sys.argv[2]=="param2"```, and so on. The command line parameters can be useful in adjusting the behaviour of your program. A few examples of these will be in the following exercises. (The terminal window is a textual interface to your computer instead of the usual graphical interface.)

- The function ```sys.exit``` can be used to exit immediately your program.

  The integer parameter given to this function is the return value of the program. Usually the return value 0 means that the program ran successfully, and non-zero integer means that an error occurred. This return value is accessible from the terminal window from where you started the program.

## `sys` Module Attributes — Simply

Three tools from the `sys` module. You already know `sys.path` (module lookup!) — let's cover it plus the two new ones.

---

### 1. `sys.path` — Where Python Looks for Modules (Recap)

You met this already: the list of folders Python searches through when you `import`. Just a reminder that it lives on `sys`, alongside these next two tools.

---

### 2. `sys.argv` — Command Line Arguments

**The setup:** normally you run a Python script with no extra info:

```bash
python3 myprogram.py
```

But you can **pass extra words** after the filename — these are called **command line parameters**:

```bash
python3 myprogram.py hello 42
```

`sys.argv` is how your **program** receives those extra words — as a list of strings.

---

### How the Pieces Map to the List

```bash
python3 myprogram.py hello 42
```

```python
import sys
print(sys.argv)
# → ['myprogram.py', 'hello', '42']
```

```
sys.argv[0]  →  'myprogram.py'    ← the SCRIPT NAME (always first, always there)
sys.argv[1]  →  'hello'          ← the first param you typed
sys.argv[2]  →  '42'             ← the second param you typed
```

**Important gotcha:** everything is a **string**, even `'42'`. If you need the number 42, you must convert it:

```python
n = int(sys.argv[2])   # '42' → 42
```

---

### Why This Is Useful — Customizing Behavior Without Editing Code

Think of it as letting the **user** control the program from outside, instead of you hardcoding values:

```python
# greet.py
import sys

name = sys.argv[1]      # whatever the user typed after the script name
print(f"Hello, {name}!")
```

```bash
python3 greet.py Alice
# → Hello, Alice!

python3 greet.py Bob
# → Hello, Bob!
```

Same script, different behavior each run — no editing the file. This is exactly how real command-line tools work: `git commit -m "message"`, `python3 -m pip install requests` — the words after the program name are all `sys.argv`.

---

### 3. `sys.exit()` — Stop the Program Immediately

A hard stop, right where you call it:

```python
import sys

print("Starting...")
sys.exit()
print("This never runs!")   # unreachable — program already ended
```

---

### The Integer Argument — "Exit Code"

```python
sys.exit(0)     # success
sys.exit(1)     # something went wrong (any non-zero = error)
```

This number is called the **return value** or **exit code**. It's not printed to your screen — it's a small signal sent back to **whoever launched your program** (the terminal, or another program).

**Universal convention:**

| Exit code | Meaning |
|---|---|
| `0` | success ✓ |
| non-zero (1, 2, ...) | something failed ✗ |

---

### Where Would You Even See This Return Value?

In the terminal, right after your program finishes:

```bash
python3 myprogram.py
echo $?
```

`echo $?` (Linux/Mac) prints the **exit code of the last command** — so it shows whatever number you passed to `sys.exit()`.

**A practical use** — validating input before doing real work:

```python
import sys

if len(sys.argv) < 2:
    print("Error: please provide a name!")
    sys.exit(1)          # signal failure to whoever ran this

name = sys.argv[1]
print(f"Hello, {name}!")
```

```bash
python3 greet.py
# → Error: please provide a name!
echo $?
# → 1                    ← the terminal can SEE that it failed
```

This matters a lot in **automated scripts** (e.g., a build pipeline running many programs in sequence) — the exit code tells the automation *"did this step succeed or fail?"* without a human reading the output.

---

### The Connection Between All Three

| Tool | Answers | Direction |
|---|---|---|
| `sys.path` | "where do I look for modules?" | Python → disk |
| `sys.argv` | "what did the user type when starting me?" | user → your program |
| `sys.exit(n)` | "how do I report success/failure when I finish?" | your program → whoever launched it |

All three are just ways your program **talks to its environment** — reading configuration, reading startup input, and reporting an outcome.

---

### The One-Sentence Summary

> `sys.argv` is a list of strings holding whatever you typed after `python3 yourscript.py` on the command line (with `sys.argv[0]` always being the script's own name); `sys.exit(n)` stops your program immediately and hands back a small integer signal — `0` for success, anything else for an error — that the terminal or another program can check afterward. 🎯
