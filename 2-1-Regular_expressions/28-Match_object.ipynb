Match object

Functions match, search, and finditer use match objects to describe the found occurence. The method groups() of the match object returns the tuple of all the substrings matched by the groups of the pattern. Each pair of parentheses in the pattern creates a new group. These groups are referred to by indices 1, 2, ... The group 0 is a special one: it refers to the match created by the whole pattern

## Match Objects and `groups()` — Simply

You already know most of this! The new pieces: multiple groups, the `.groups()` method, and group 0. Let's build it up with one example.

---

### The Setup — A Pattern With TWO Groups

Remember: each `(...)` creates a numbered group, counted by **opening parentheses, left to right**:

```python
import re

m = re.search(r'(\w+) is (\d+) years old', 'Alice is 30 years old')
#               └─┬─┘    └─┬─┘
#              group 1   group 2
```

---

### Asking for Each Group

You've used `.group(1)` before — now with two groups:

```python
m.group(1)    # → 'Alice'                   what the 1st (...) captured
m.group(2)    # → '30'                      what the 2nd (...) captured
```

---

### The NEW Method — `.groups()` Gets Them ALL at Once

```python
m.groups()    # → ('Alice', '30')           tuple of every capture
```

A **tuple** — so all your tuple skills apply, especially **unpacking**:

```python
name, age = m.groups()
print(name)   # → Alice
print(age)    # → 30
```

One line, both captures into named variables. This is the everyday payoff of `groups()`.

---

### Group 0 — The Special One

Group 0 = **the entire match** — everything the whole pattern touched, scaffolding included:

```python
m.group(0)    # → 'Alice is 30 years old'   the WHOLE match
m.group()     # → 'Alice is 30 years old'   same! 0 is the default
```

So when you've been writing `m.group()` all along — you were calling group 0 without knowing its name!

**The map:**

```
'Alice is 30 years old'
 └────────┬──────────┘
       group 0          ← whole pattern's match (scaffolding + captures)

 Alice        30
 └─┬─┘       └┬┘
group 1    group 2      ← just the parenthesized parts
```

Note the asymmetry: `.groups()` (plural) returns groups **1, 2, ...** only — group 0 is *not* in the tuple. Group 0 is the frame; the tuple holds the pictures.

---

### Connecting to What You Know

This should click with the **fishing analogy** 🎣 from the `findall` discussion:

- **Group 0** = the whole fishing spot — everything the pattern located
- **Groups 1, 2, ...** = the nets — the parts you wanted to keep
- `.groups()` = *"empty all nets into one tuple"*

And it explains `findall`'s behavior precisely: with one group, `findall` returns group-1 strings; with **multiple** groups, it returns **tuples** — literally each match's `.groups()`:

```python
s = 'Alice is 30 years old. Bob is 25 years old.'

re.findall(r'(\w+) is (\d+) years old', s)
# → [('Alice', '30'), ('Bob', '25')]        ← a list of .groups() tuples!

for name, age in re.findall(r'(\w+) is (\d+) years old', s):
    print(f'{name} → {age}')
# Alice → 30
# Bob → 25
```

That loop — `findall` + tuple unpacking — is one of the most common text-extraction idioms in Python.

---

### Cheat Sheet

| Call | Returns | Content |
|---|---|---|
| `m.group()` or `m.group(0)` | string | the **whole** match |
| `m.group(1)` | string | first `(...)` capture |
| `m.group(2)` | string | second `(...)` capture |
| `m.groups()` | **tuple** | all captures — groups 1, 2, ... (no 0!) |

> **One sentence:** the match object stores the whole match as group 0 and each parenthesized capture as groups 1, 2, ...; `.group(n)` fetches one, `.groups()` hands you all the captures as a tuple ready for unpacking. 🎯

````python
import re

str='first 123 45 67 890 last'
mo = re.search(r'\d+ (\d+) \d+ (\d+)', str)
#r'\d+ (\d+)<-G1 \d+ (\d+) <-G2'
print(mo)
````

The call mo.groups() returns a tuple (’45’, ’890’). We can access just some individual groups by using the method group(gid, ...). For example, the call mo.group(1) will return ’45’. The zeroth group will represent the whole match: ’123 45 67 890’

In addition to accessing the strings matched by the pattern and its groups, the corresponding indices of the original string can be accessed:

The start(gid=0) and end(gid=0) methods return the start and end indices of the matched group gid, correspondingly
The method span(gid) just returns the pair of these start and end indices

````python
print(mo.start())
print(mo.end())
print(mo.span())
````

The match object mo can also be used like a boolean value:




````python
mo = re.search(...)
if mo:
    # do something
````

will do something if a match was found. Alternatively, the match object can be converted to a boolean value by the call ```found = bool(mo).```
