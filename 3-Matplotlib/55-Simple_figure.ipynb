### Simple figure

We will start with an example. The standard way to import matplotlib is as the abbreviation `plt`.

```python
import numpy as np
import matplotlib.pyplot as plt
```

Let's first have some data to visualize:

```python
a=np.array([2,5,7,4,7,0,3,1,9,2])
```

Below the plot function does the actual drawing of the graph, the rest of the function calls adjust some details of the figure. Make sure you understand how the values in the array a correspond to the features in the figure!

```python
plt.plot(a)                   # plot the points (ලක්ෂ්‍ය සටහන් කිරීම) in the array
plt.title('My first figure')  # Add title to the figure
plt.xlabel('My x-axis')       # Give a label to the x-axis
plt.ylabel('My y-axis')       # Give a label to the y-axis
plt.show()                    # Tell matplotlib to output the figure.
                              # Not strictly required in notebooks (but a bit neater).
```

Here's the diagram:The diagram labels every part of a matplotlib plot. Here's what each term means, grouped by level:

**The containers (biggest to smallest):**
- **figure** — the entire outer image (the red border). One figure holds everything, and can contain multiple plots.
- **axes** — one individual plot (the box where data is drawn). Note: despite the name, "axes" means the *whole plotting area*, not the x/y lines. This figure has **two** axes side by side ("First degree" and "Second degree").

**The text labels:**
- **suptitle** — the "**sup**er title" for the *whole figure* ("Polynomials") — sits above everything.
- **title** — the title of *one individual axes* ("First degree", "Second degree").
- **xlabel / ylabel** — the names of the x-axis and y-axis ("x" and "y").

**The axis markings (the easily-confused pair):**
- **xtick / ytick** — the little **marks** along each axis (the small lines).
- **xticklabel / yticklabel** — the **numbers** printed next to those marks (0, 1, 2... or 0, 20, 40...).

The key distinction people mix up: a **tick** is the little mark, the **ticklabel** is the number beside it.

**The data:**
- **line** — the actual plotted data (the blue curve itself).

The useful mental model is the nesting: a **figure** contains one or more **axes**, and each **axes** contains its own title, labels, ticks, ticklabels, and plotted lines — while the **suptitle** sits at the figure level, above all the individual axes. This is exactly why, in code, `fig.suptitle(...)` is called on the *figure* but `ax.set_title(...)` is called on each *axes*.
