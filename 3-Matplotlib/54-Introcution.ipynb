### Matplotlib

To understand the data better it helps to be able to visualize it in various ways (දත්ත වඩා හොඳින් අවබෝධ කර ගැනීම සඳහා, ඒවා විවිධ ආකාරවලින් දෘශ්‍යමාන කිරීමට හැකිවීම ප්‍රයෝජනවත් වේ.) Matplotlib https://matplotlib.org/ is the most common low-level visualization library for Python. It can create line graphs, scatter plots, density plots, histograms, heatmaps, and so on. During this course we will not go deep into details of matplotlib, instead we have just some examples spread throughout the rest of this material of its use.
