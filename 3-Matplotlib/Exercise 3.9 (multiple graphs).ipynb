```python
import matplotlib.pyplot as plt
import numpy as np
```

```python
x1= np.array([2,4,6,7])
y1= np.array([4,3,5,1])

x2= np.array([1,2,3,4])
y2= np.array([4,2,3,1])
```

```python
plt.plot(x1,y1,label="Line 1")
plt.plot(x2,y2,label="Line 2")
plt.title("Multiple graphs")
plt.xlabel('x-axis')
plt.ylabel('y-axis')
plt.legend()       # ← shows which line is which
plt.show()
```
