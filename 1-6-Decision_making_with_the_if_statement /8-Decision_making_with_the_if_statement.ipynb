Decision making with the if statement

The ```if-else``` statement works as can be expected. Try running the below cell by pressing control+enter.

````python
x = input('Give me an integer:')
x=int(x)

if x>=0:
    a=x
else:
    a=-x
print(f'The absolute value of {x} is {a}')
````

The general form of an if-else statement is
```
if condition1:
    statement1_1
    statement1_2
    ...
elif condition2:
    statement2_1
    statement2_2
    ...
...
else:
    statementn_1
    statementn_2
    ...
```

````python
c=float(input('Give a number:'))

if c > 0 :
    print('c is positive')
elif c < 0 :
    print('c is negative')
else:
    print('c is zero')
````
