It would be nice that the number of arguments could be arbitrary, not just two. We could pass a list to the function as a parameter.

````python
def sum_of_squares(lst):
    """Computes the sum of squares of elements in the list given as parameter."""
    s = 0
    for i in lst:
        s += i**2
    return s

print(sum_of_squares([-2]))
print(sum_of_squares([-2,4,5]))
````

This works perfectly! There is however some extra typing with the brackets around the lists. Let's see if we can do better:

````python
def sum_of_squares(*t):
    """Computes the sum of squares of elements in the list given as parameter."""
    s=0
    for i in t:
        s+=i**2
    return s

print(sum_of_squares(-2))
print(sum_of_squares(-2,4,5))
````

The strange looking argument notation ```(the star)``` is called ```argument packing```.

It packs all the given positional arguments into a ```tuple t```.

## What is an Immutable List?

**Immutable** means **cannot be changed** after it is created.

So an **immutable list** is a list-like collection whose contents you **cannot modify** once it's made.

---

### The two main collection types in Python:

| | List | Tuple |
|---|---|---|
| Syntax | `[1, 2, 3]` | `(1, 2, 3)` |
| Mutable? | ✓ Yes (can change) | ✗ No (cannot change) |
| Called | mutable | **immutable** |

---

### With a list — you can change it:
```python
my_list = [1, 2, 3]
my_list[0] = 99       # ✓ allowed
my_list.append(4)     # ✓ allowed
print(my_list)        # → [99, 2, 3, 4]
```

### With a tuple — you cannot change it:
```python
my_tuple = (1, 2, 3)
my_tuple[0] = 99      # ✗ ERROR! TypeError: 'tuple' object does not support item assignment
my_tuple.append(4)    # ✗ ERROR! tuples have no append method
```

Once a tuple is created — its contents are **locked forever**.

---

### What you can still do with tuples:
You can **read** from them — just not modify:
```python
my_tuple = (1, 2, 3)
print(my_tuple[0])    # ✓ 1
print(len(my_tuple))  # ✓ 3

for x in my_tuple:    # ✓ can iterate
    print(x)
```

---

### Simple analogy:

> A **list** is like a **whiteboard** — you can write, erase, and edit anytime.
> A **tuple** is like a **printed book** — the words are fixed, you can read them but not change them.

---

### Why have immutable tuples?
- **Safety** — guarantees the data won't accidentally change
- **Speed** — Python can optimize them better than lists
- **Hashable** — can be used as dictionary keys (lists cannot)

---

### Summary:
> An **immutable list** = a **tuple**. It behaves like a list (ordered, indexable, iterable) but **cannot be modified** after creation.

## Argument Packing and Unpacking

These are two opposite operations using the `*` symbol — one **collects** arguments, the other **spreads** them out.

---

## 1. Argument Packing — Collecting Many into One

When defining a function, `*args` **packs** all positional arguments into a single tuple:

```python
def my_func(*args):
    print(args)
    print(type(args))

my_func(1, 2, 3, 4)
```
Output:
```
(1, 2, 3, 4)
<class 'tuple'>
```

The `*args` says: *"take all the values passed in and pack them into one tuple called `args`"*.

```
my_func(1, 2, 3, 4)
         ↓
       packed into tuple
         ↓
       args = (1, 2, 3, 4)
```

---

### Why is it useful?
You can accept **any number of arguments** without knowing how many in advance:
```python
def sum_all(*numbers):
    return sum(numbers)

sum_all(1, 2)              # → 3
sum_all(1, 2, 3, 4, 5)     # → 15
sum_all(10, 20, 30, 40)    # → 100
```

---

## 2. Argument Unpacking — Spreading One into Many

When **calling** a function, `*` **unpacks** an iterable into individual arguments:

```python
def add(a, b, c):
    return a + b + c

numbers = (1, 2, 3)
add(*numbers)           # → same as add(1, 2, 3) → 6
```

The `*numbers` says: *"take this tuple and spread its elements out as separate arguments"*.

```
numbers = (1, 2, 3)
            ↓
       unpacked
            ↓
add(1, 2, 3)
```

---

### Without unpacking — you'd have to do it manually:
```python
add(numbers[0], numbers[1], numbers[2])   # tedious
add(*numbers)                              # clean
```

---

## Side by Side Comparison:

| | Packing | Unpacking |
|---|---|---|
| Where | function **definition** | function **call** |
| Direction | many → one tuple | one tuple → many |
| Example | `def f(*args):` | `f(*my_tuple)` |
| Star meaning | "collect into tuple" | "spread out from tuple" |

---

### Complete example showing both:
```python
def show(*args):           # PACKING — collect into args
    print(args)

values = (1, 2, 3)
show(*values)              # UNPACKING — spread values out
# → (1, 2, 3)
```

---

### Simple analogy:

> **Packing** is like putting many items into a **suitcase** — `*args` zips them all together.
> **Unpacking** is like opening a suitcase and **taking each item out** — `*values` lays them out individually.

> Both use the same `*` symbol, but the **direction** is opposite — packing collects, unpacking spreads.

lst=[1,5,8]

## Great Observation — Unpacking Works on Any Iterable!

You're right that this is a list, not a tuple — and that's **completely fine**. Unpacking with `*` works on **any iterable**, not just tuples.

---

### What `*` unpacking really does:
The `*` operator says: *"take this iterable and spread out each element as a separate argument"*. It doesn't care if it's a list, tuple, set, or any other iterable.

```python
lst = [1, 5, 8]
sum_of_squares(*lst)
#               ↓ unpacks the list
sum_of_squares(1, 5, 8)
```

---

### Works on all these iterables:

```python
def add(a, b, c):
    return a + b + c

# Tuple
t = (1, 2, 3)
add(*t)         # → 6

# List
lst = [1, 2, 3]
add(*lst)       # → 6

# Set
s = {1, 2, 3}
add(*s)         # → 6

# String (yes, even strings!)
chars = "ABC"
def show(a, b, c):
    print(a, b, c)
show(*chars)    # → A B C

# range
add(*range(1, 4))   # → 6  (unpacks to 1, 2, 3)
```

---

### So why does the textbook mention tuples?

The textbook was talking about **packing** (`*args` in a function definition), which **always packs into a tuple**:
```python
def f(*args):
    print(type(args))   # → always <class 'tuple'>
```

But **unpacking** (`*` in a function call) is **flexible** — it works on **any iterable**:

| Operation | What it does |
|---|---|
| **Packing** `*args` | always packs **into a tuple** |
| **Unpacking** `*iterable` | works on **any iterable** (list, tuple, set, etc.) |

---

### Verifying your example:
```python
def sum_of_squares(a, b, c):
    return a**2 + b**2 + c**2

lst = [1, 5, 8]
print(sum_of_squares(*lst))   # → 1 + 25 + 64 = 90
```

Behind the scenes:
```
lst = [1, 5, 8]
       ↓ unpacked with *
sum_of_squares(1, 5, 8)
       ↓
returns 1² + 5² + 8² = 90
```

---

### The simple rule:
> - **Packing** always creates a **tuple**.
> - **Unpacking** spreads elements out — from **any iterable** (list, tuple, set, string, range, generator).

## Yes! Both Approaches Work — But They Are Different

Both produce the same output, but they work in **fundamentally different ways**.

---

### Approach 1 — Packing/Unpacking (`*`)
```python
def show(*args):       # accepts ANY NUMBER of arguments
    print(args)

show(1, 2, 3)          # ✓ works
show(1, 2, 3, 4, 5)    # ✓ works
show(1)                # ✓ works
show()                 # ✓ works (empty tuple)
```
Inside the function, `args` becomes a tuple containing **whatever was passed in**.

---

### Approach 2 — Passing a Tuple Directly
```python
def show(t):           # accepts EXACTLY ONE argument (which is a tuple)
    print(t)

show((1, 2, 3))        # ✓ works
show((1, 2, 3, 4, 5))  # ✓ works
show(1, 2, 3)          # ✗ ERROR! too many arguments
```
The function expects **one single argument** — and that argument happens to be a tuple.

---

### The key difference — flexibility:

```python
# Approach 1 — flexible, any count
def show1(*args):
    print(args)

show1(1)
show1(1, 2)
show1(1, 2, 3, 4, 5)   # all work ✓


# Approach 2 — rigid, exactly one tuple
def show2(t):
    print(t)

show2((1,))            # ✓
show2((1, 2))          # ✓
show2(1, 2, 3)         # ✗ ERROR
```

---

### When to use which?

| Approach | Use when |
|---|---|
| `def f(*args)` | You **don't know how many arguments** the user will pass |
| `def f(t)` | You **want exactly one tuple/list** as input |

---

### Real world examples:

**Approach 1 — `print()` uses this:**
```python
print("a")              # 1 arg
print("a", "b")         # 2 args
print("a", "b", "c")    # 3 args
# print accepts any number of arguments via *args
```

**Approach 2 — `len()` uses this:**
```python
len((1, 2, 3))    # ✓ one tuple
len([1, 2, 3])    # ✓ one list
len(1, 2, 3)      # ✗ ERROR — wants ONE argument
```

---

### The summary:
> Both work, but they solve **different problems**:
> - `*args` → *"I want to accept any number of things"*
> - `t` → *"I want exactly one thing (which is a collection)"*
>
> The textbook used `*args` to teach **packing**, but your alternative is perfectly valid Python — just less flexible.

We did not need to specify all the parameters with default values, only those we wanted to change.

````python
def length(*t, degree=2):
    """Compute the length of the vector given as parameter. By default, it computes the Euclidean distance (Degree==2)."""
    s=0
    for x in t:
        s+=abs(x)**degree
    return s**(1/degree)

print(length(-4,3))
print(length(-4,3,degree=3))
````

With the default parameter this is the Euclidean distance, and if p≠2 it is called ```p-norm```.

## Yes — Python has Built-in Functions for This

There are several built-in ways to compute `s^(1/degree)`:

---

### Option 1 — `pow()` (built-in, no import needed)
```python
pow(s, 1/degree)
```
Same as `s**(1/degree)`, just written as a function call:
```python
pow(91, 1/3)    # → 4.4979
pow(25, 1/2)    # → 5.0
```

---

### Option 2 — `math.pow()` (from math module)
```python
import math
math.pow(s, 1/degree)
```
Almost identical to `pow()`, but always returns a **float**:
```python
import math
math.pow(91, 1/3)    # → 4.4979
math.pow(25, 1/2)    # → 5.0
```

---

### Option 3 — `**` operator (cleanest, what you already had)
```python
s ** (1/degree)
```
This is what you already wrote — it is **idiomatic Python** and considered the most natural:
```python
91 ** (1/3)    # → 4.4979
25 ** (1/2)    # → 5.0
```

---

### Special case — only square root → `math.sqrt()`
If you specifically want degree=2:
```python
import math
math.sqrt(25)    # → 5.0  (only works for square roots)
```
But this **doesn't help for other degrees** — that's why the second version of your code was buggy.

---

### Comparison table:

| Method | Import needed | General? | Example |
|---|---|---|---|
| `s ** (1/n)` | no | ✓ yes | `91 ** (1/3)` |
| `pow(s, 1/n)` | no | ✓ yes | `pow(91, 1/3)` |
| `math.pow(s, 1/n)` | yes | ✓ yes | `math.pow(91, 1/3)` |
| `math.sqrt(s)` | yes | ✗ only degree=2 | `math.sqrt(25)` |

---

### So the cleanest version of your function:
```python
def length(*t, degree=2):
    s = sum(abs(x)**degree for x in t)
    return pow(s, 1/degree)        # or: s ** (1/degree)
```

---

### Recommendation:
> Use `s ** (1/degree)` or `pow(s, 1/degree)`. They are general — they work for **any degree**, not just 2.
> Avoid `math.sqrt()` unless you specifically want square roots only.

## What are p-norms?

A **norm** is a mathematical way to measure the **"length" or "size"** of a vector. The `p` is just a number that decides **which formula** to use.

---

### The General p-norm Formula

For a vector `v = [x₁, x₂, ..., xₙ]`:

```
‖v‖ₚ = (|x₁|ᵖ + |x₂|ᵖ + ... + |xₙ|ᵖ)^(1/p)
```

Different values of `p` give you **different ways** to measure length — all valid, just different perspectives.

---

### The famous p-norms:

#### **p = 1 → Manhattan distance (L1 norm)**
```
‖v‖₁ = |x₁| + |x₂| + ... + |xₙ|
```
Just **sum the absolute values**. Like walking through city blocks — you can only move along streets, not diagonally.
```python
# v = [3, 4]
|3| + |4| = 7
```

---

#### **p = 2 → Euclidean distance (L2 norm)** ← default in your function
```
‖v‖₂ = √(x₁² + x₂² + ... + xₙ²)
```
The classic **straight-line distance** (Pythagoras' theorem).
```python
# v = [3, 4]
√(9 + 16) = √25 = 5
```

---

#### **p = 3 → L3 norm**
```
‖v‖₃ = ∛(|x₁|³ + |x₂|³ + ... + |xₙ|³)
```
Less common, but mathematically valid:
```python
# v = [-4, 3]
∛(64 + 27) = ∛91 ≈ 4.4979
```

---

#### **p = ∞ → Chebyshev distance (L∞ norm)**
```
‖v‖∞ = max(|x₁|, |x₂|, ..., |xₙ|)
```
Just the **largest absolute value**. Like a king's move in chess — you can move equal distance in any direction.
```python
# v = [3, 4]
max(3, 4) = 4
```

---

### Visual comparison (for vector `[3, 4]`):

| p | Name | Formula | Result |
|---|---|---|---|
| 1 | Manhattan / Taxicab | sum of \|x\| | 7 |
| **2** | **Euclidean** (most common) | √(sum of x²) | **5** |
| 3 | L3 | ∛(sum of \|x\|³) | ≈4.49 |
| ∞ | Chebyshev | max(\|x\|) | 4 |

---

### Why are they called "p-norms"?

Because they are **all the same general formula** — only the parameter `p` changes:
```
‖v‖ₚ = (Σ |xᵢ|ᵖ)^(1/p)
```
This single formula **becomes a different norm** when you plug in different values of `p`. So `p` is the **parameter** that selects which norm — hence "**p-norm**".

---

### Connection to your function:
```python
def length(*t, degree=2):
    s = sum(abs(x)**degree for x in t)
    return s**(1/degree)
```
Your `degree` parameter **is the `p`** in p-norm. By changing it, you switch between Manhattan, Euclidean, L3, etc.

---

### Real-world analogy:

> Imagine measuring how far apart two places are:
> - **By car** (only on roads) → Manhattan (p=1)
> - **As the crow flies** (straight line) → Euclidean (p=2)
> - **Worst single direction** (largest gap) → Chebyshev (p=∞)
>
> All are valid ways to define "distance" — just from different perspectives.
