# Dictionary Methods — Quick Notes

Dictionary methods come in two flavors: **non-mutating** (they read information without changing the dict) and **mutating** (they modify the dict in place).

---
# Non-Mutating Methods

These methods **do not change** the original dictionary. They only read from it or return new information.

## 1. `d.copy()` — make a shallow copy

Creates a new dict with the same key-value pairs. Changes to the copy don't affect the original.  
> **Note:** 'shallow' means nested objects are shared, not duplicated.

```python
d = {'a': 1, 'b': 2}

d2 = d.copy()
d2['a'] = 99

print(d)   # → {'a': 1, 'b': 2}       ← original unchanged
print(d2)  # → {'a': 99, 'b': 2}      ← copy modified
```

## 2. `d.items()` — get all key-value pairs

Returns a view of all key-value pairs as tuples. Perfect for iterating both keys and values together in a for loop.

```python
d = {'a': 1, 'b': 2}

for key, value in d.items():
    print(key, value)
# → a 1
# → b 2
```

## 3. `d.keys()` — get all keys

Returns a view of all keys. You can also just write `for key in d:` — iterating a dict directly gives keys by default.

```python
d = {'a': 1, 'b': 2}

print(list(d.keys()))   # → ['a', 'b']

for key in d.keys():
    print(key)
# → a
# → b
```

## 4. `d.values()` — get all values

Returns a view of all values. Useful when you only care about the values, e.g. summing them or checking if a value exists.

```python
d = {'a': 1, 'b': 2}

print(list(d.values()))   # → [1, 2]

total = sum(d.values())
print(total)              # → 3
```

## 5. `d.get(k, x=None)` — safely look up a key

Safer than `d[key]` — returns `None` (or your custom default) if the key is missing, instead of raising `KeyError`.

```python
d = {'a': 1, 'b': 2}

print(d.get('a'))         # → 1
print(d.get('z'))         # → None      (key missing, no error!)
print(d.get('z', 0))      # → 0         (default value when missing)
```

---
# Mutating Methods

These methods **change the dictionary in place** — after calling them, the dict is different than before.

## 1. `d.clear()` — remove everything

Empties the dict completely. Same as writing `d = {}`, except it modifies the existing object rather than replacing it.

```python
d = {'a': 1, 'b': 2}
d.clear()
print(d)   # → {}
```

## 2. `d.update(d1)` — merge another dict in

Adds all key-value pairs from another dict. If a key already exists, **its value is overwritten**. If a key is new, it's added.

```python
d = {'a': 1, 'b': 2}
d.update({'b': 99, 'c': 3})
print(d)   # → {'a': 1, 'b': 99, 'c': 3}
```

## 3. `d.setdefault(k, x=None)` — get or insert

If the key exists, returns its current value. If it doesn't, **inserts the default and returns it**. Useful for initializing values only once.

```python
d = {'a': 1}

d.setdefault('a', 99)   # 'a' exists → returns 1, dict unchanged
print(d)                # → {'a': 1}

d.setdefault('b', 5)    # 'b' missing → inserts 'b': 5
print(d)                # → {'a': 1, 'b': 5}
```

## 4. `d.pop(k, x=None)` — remove and return a value

Removes the key and returns its value. If the key is missing, raises `KeyError` — unless you provide a default, in which case that default is returned instead.

```python
d = {'a': 1, 'b': 2}

val = d.pop('a')
print(val)      # → 1
print(d)        # → {'b': 2}   (key 'a' removed)

val = d.pop('z', 'not found')   # missing key + default
print(val)      # → not found  (no error!)
```

## 5. `d.popitem()` — remove and return the last pair

Removes and returns the **last inserted** key-value pair as a tuple. Raises `KeyError` if the dict is empty.  
> Since Python 3.7, dicts remember insertion order — this always removes the newest entry.

```python
d = {'a': 1, 'b': 2, 'c': 3}

pair = d.popitem()
print(pair)     # → ('c', 3)   (the last inserted pair)
print(d)        # → {'a': 1, 'b': 2}
```

# ⚠️ Important Gotchas

## Gotcha 1: `d.get()` vs `d[key]`

```python
d = {'a': 1}

# d['z']         # ✗ KeyError!
print(d.get('z'))      # ✓ returns None (safe)
print(d.get('z', 0))   # ✓ returns 0 (default)
```

## Gotcha 2: `.keys()`, `.values()`, `.items()` return VIEWS, not lists

The view updates automatically if the dict changes. Convert to a list if you need a fixed snapshot.

```python
d = {'a': 1, 'b': 2}
keys = d.keys()
print(keys)   # → dict_keys(['a', 'b'])    ← not a list!

# The view updates if the dict changes:
d['c'] = 3
print(keys)   # → dict_keys(['a', 'b', 'c'])

# Convert to a list if you need a snapshot:
print(list(d.keys()))   # → ['a', 'b', 'c']
```

## Gotcha 3: `.copy()` is SHALLOW

Nested objects (like lists) are shared between the original and the copy. Use `copy.deepcopy()` for fully independent copies.

```python
d = {'x': [1, 2]}
d2 = d.copy()
d2['x'].append(99)   # modifies the SHARED list
print(d)             # → {'x': [1, 2, 99]}    ← original also changed!

# For nested structures, use copy.deepcopy():
import copy
d3 = copy.deepcopy(d)
d3['x'].append(999)
print(d)   # → {'x': [1, 2, 99]}    ← original now safe
```

# Quick Cheat Sheet

## Non-Mutating Methods

| Method | What it does | Example |
|---|---|---|
| `d.copy()` | make a shallow copy of the dict | `d2 = d.copy()` |
| `d.items()` | get all (key, value) pairs | `for k,v in d.items():` |
| `d.keys()` | get all keys | `list(d.keys())` |
| `d.values()` | get all values | `sum(d.values())` |
| `d.get(k, x)` | safe lookup with default | `d.get('a', 0)` |

## Mutating Methods

| Method | What it does | Example |
|---|---|---|
| `d.clear()` | remove all entries | `d.clear()` |
| `d.update(d1)` | merge another dict in | `d.update({'x': 1})` |
| `d.setdefault(k, x)` | get or insert with default | `d.setdefault('a', 0)` |
| `d.pop(k, x)` | remove key and return value | `d.pop('a', None)` |
| `d.popitem()` | remove and return last pair | `k, v = d.popitem()` |

# 💡 The Key Idea

Dictionary methods split into two clear groups:

- **Non-mutating** — read-only. They give you information about the dict without changing it. Safe to use anywhere without side effects.
- **Mutating** — change the dict in place. Use them when you actually want to modify the original.

> ⚠️ As with lists, mutating methods **return `None` or a removed value** — don't try to chain them like `d = d.update(...)` or you'll lose your dict!
