### Dictionaries

A dictionary is a dynamic, unordered container. Instead of using integers to access the elements of the container, the dictionary uses keys to access the stored values. The dictionary can be created by listing the comma separated key-value pairs in braces. Keys and values are separated by a colon. A tuple `(key,value)` is called an `item` of the dictionary.

````python
d={
    'key1':'value1',
    'key2':'value2'
}
````

````python
print(d['key1'])
````

````python
print(d['key2'])
````

````python
d={
    1:'a',
    'z':1
}
````

Keys can have different types even in the same container. So the following code is legal: `d={1:"a", "z":1}`. The only restriction is that the keys must be hashable. That is, there has to be a mapping from keys to integers. Lists are not hashable, but tuples are!

## What "Hashable" Means

**Hashable** means an object can be converted into a **fixed, unique number** (called a **hash**) that Python uses to store and find it quickly.

---

### The Simple Idea

A **hash** is like a **fingerprint** for an object — a number that identifies it:

```python
hash("hello")   # → 5926963372145697509
hash(42)        # → 42
hash((1, 2))    # → 1299869600
```

Every hashable object has one of these fingerprints. Python uses them to organize data efficiently.

---

### Why Does Python Need Hashes?

For **fast lookup** in dictionaries and sets.

When you write `d["key"]`, Python doesn't scan through every entry to find `"key"`. It computes `hash("key")`, uses that number as an **address**, and jumps directly to the right spot.

```
d["hello"]
    ↓
hash("hello") = 5926963...
    ↓
jump directly to that slot → find value
```

This is why dictionary lookups are **almost instant**, even with millions of entries.

---

### The Rule — Hash Must Never Change

For this system to work, the fingerprint must be **stable**. If an object's hash changed after you stored it, Python couldn't find it again — it would be looking in the wrong place.

That's why hashability requires **immutability**:

> **If an object can be modified, its hash could change — so it can't be trusted as a key.**

---

### What's Hashable vs Not

| Hashable ✓ | Not Hashable ✗ |
|---|---|
| `int` | `list` |
| `float` | `dict` |
| `str` | `set` |
| `tuple` (of hashables) | `bytearray` |
| `bool` | `tuple` (containing a list) |
| `None` | |

**Pattern:** immutable types → hashable. Mutable types → not hashable.

---

### Why Lists Fail

```python
d = {[1, 2]: "value"}
# ✗ TypeError: unhashable type: 'list'
```

Because a list can be modified:
```python
key = [1, 2]
d = {}
# Imagine Python allowed this:
d[key] = "value"     # stored under fingerprint of [1, 2]
key.append(3)        # now key is [1, 2, 3] — different fingerprint!
d[key]               # Python looks under new fingerprint → doesn't find it!
```

The value would be **lost** — stored somewhere Python can't find anymore. To prevent this disaster, Python simply refuses lists as keys.

---

### Why Tuples Work — Usually

```python
d = {(1, 2): "value"}   # ✓ works
```

Tuples are immutable — once created, their contents never change → hash never changes → safe as a key.

**Exception:** if a tuple contains something mutable, it becomes unhashable:
```python
d = {(1, [2, 3]): "value"}
# ✗ TypeError: unhashable type: 'list'
```

The tuple itself is immutable, but the list inside can change — which breaks the guarantee. Python catches this and refuses.

---

### Check Hashability Yourself

You can test any object with the `hash()` function:

```python
hash(42)          # → 42               ✓ hashable
hash("hello")     # → some big number  ✓ hashable
hash((1, 2, 3))   # → some number      ✓ hashable

hash([1, 2, 3])   # ✗ TypeError: unhashable type: 'list'
hash({1, 2, 3})   # ✗ TypeError: unhashable type: 'set'
```

If `hash()` succeeds, the object is hashable and can be a dict key or set element.

---

### Where Hashability Matters

Hashable objects can be:
- **Dictionary keys** — `d[key] = value`
- **Set elements** — `s = {1, 2, "hello"}`
- **Frozen keys in caches / memoization**

Non-hashable objects can only be:
- **Values in a dictionary** — `d["key"] = [1, 2, 3]` (lists as values are fine!)
- **Elements of a list** — `L = [[1, 2], [3, 4]]`

---

### Simple Analogy

> Think of hashable objects as **items with a permanent barcode** 🏷️:
>
> - Once you print the barcode on an item, it never changes
> - The library scans the barcode to find the book instantly
> - If the barcode could change while sitting on the shelf, the library would lose it forever
>
> **Immutable objects** (str, int, tuple) — like books with printed barcodes ✓ safe
> **Mutable objects** (list, dict, set) — like items whose barcode could be edited ✗ unsafe
>
> Python only accepts *"permanent barcode"* items as dictionary keys.

---

### The Key Idea

> **Hashable = has a stable, unique fingerprint that Python can use for fast lookup.**
>
> The rule is simple: if an object can be **modified**, its fingerprint could **change**, so it can't be trusted as a key — Python refuses it.
>
> **Immutable → hashable → can be a dict key or set element.**
> **Mutable → not hashable → cannot be a dict key or set element.**
>
> That's why `str`, `int`, and `tuple` work as keys, but `list`, `dict`, and `set` don't. 🎯

There are alternative syntaxes for dictionary creation:

````python
dict([
    ('key1','value1'),
    ('key2','value2'),
    ('key3','value3')
])    #List of items
````

````python
dict(
    key1='value1',
    key2='value2',
    key3='value3',
)
````

````python
dict(
    key1='value1',
    key2='value2',
    key3='value3',
    1='2'
)
````

If a key is not found in a dictionary, the indexing `d[key]` results in an error `(exception KeyError)`. But an assignment with a non-existing key causes the key to be added in the dictionary associated with the corresponding value:

````python
d={}
d[2] = 'value'
print(d)
````
