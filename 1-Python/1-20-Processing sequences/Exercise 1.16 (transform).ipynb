```python
from os.path import split

a='1 5 3'
b='2 6 -1'
L1 = list(map(int,a.split(' ')))
L2 = list(map(int,b.split(' ')))

print(list(map(lambda x,y:x*y,L1,L2)))

```

```python
def transform(s1, s2):
    L1 = list(map(int,s1.split(' ')))
    L2 = list(map(int,s2.split(' ')))
    return list(map(lambda x,y:x*y,L1,L2))
```

```python
a='1 5 3'
b='2 6 -1'
print(transform(a,b))
```
