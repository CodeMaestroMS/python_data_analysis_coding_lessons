### Module lookup

When we try to import a module mod with the import statement, the lookup proceeds in the following order:

- Check if it is a builtin module
- Check if the file mod.py is found in any of the folders in the list sys.path. The first item in this list is the current folder

When Python is started, the `sys.path` list is initialised with the contents of the PYTHONPATH environment variable

````python
import sys
print(sys.path)
````

## How Python Finds a Module When You `import` — Simply

When you write `import mod`, Python has to **locate** where `mod` actually lives. This explains the **search order** it follows. You've actually met `sys.path` before (in the `sys` module lesson) — this is where it gets put to use.

---

### The Search Order — Two Steps

When you write `import mod`, Python looks in this order:

**Step 1 — Is `mod` a built-in module?**

Some modules are **compiled into Python itself** (like `sys`, `math`, `time`). Python checks these first. If `mod` is one of them, done — no file searching needed.

**Step 2 — Is there a file `mod.py` somewhere in `sys.path`?**

If it's not built-in, Python searches for a **file** named `mod.py`, looking through each folder listed in `sys.path`, **in order**, stopping at the first match.

```
import mod
   │
   ├─ Step 1: is "mod" a builtin? ──→ yes ──→ use it, done
   │                                  no
   │                                   ↓
   └─ Step 2: search sys.path folders for "mod.py"
              folder 1 → folder 2 → folder 3 → ...
              use the FIRST one that has mod.py
```

---

### What `sys.path` Is — A List of Folders to Search

`sys.path` is just a **list of folder paths** where Python looks for modules:

```python
import sys
print(sys.path)
# → ['', '/usr/lib/python3.11', '/usr/lib/python3.11/site-packages', ...]
#      ↑ current folder    ↑ standard library    ↑ installed packages
```

Python checks these folders **top to bottom** and uses the **first** place it finds `mod.py`.

---

### "The First Item Is the Current Folder" — Why It Matters

The **first** entry in `sys.path` is usually the **current folder** (where your script is running). This means Python checks *your own folder first*, before the standard library folders.

**This has a consequence (the shadowing trap!):** if you name your own file the same as a standard module, **your file wins** — because it's found first:

```
your_folder/
├── random.py        ← YOUR file, named "random"
└── myprogram.py
```

```python
# in myprogram.py:
import random        # ✗ imports YOUR random.py, NOT Python's real random module!
```

Since the current folder is searched **first**, your `random.py` shadows the real one. This is the same "don't name files after builtins" warning from earlier — now you see **why** it happens: current-folder-first lookup.

---

### Where `sys.path` Comes From — `PYTHONPATH`

The last sentence explains the **origin** of `sys.path`'s contents:

> *"sys.path is initialised with the contents of the PYTHONPATH environment variable."*

**Environment variable** = a setting stored in your operating system (outside Python). `PYTHONPATH` is a special one that lists **extra folders** you want Python to search for modules.

```
When Python STARTS:
   reads PYTHONPATH (an OS setting)  →  uses it to fill in sys.path
```

So you can **add** your own folders to the search path by setting `PYTHONPATH` before running Python — useful if your modules live somewhere non-standard:

```bash
# Linux/Mac — add a custom folder to the module search path:
export PYTHONPATH=/home/me/mymodules
python3 myprogram.py       # now Python also searches /home/me/mymodules
```

(Python also adds its own standard folders automatically — `PYTHONPATH` just contributes **extra** ones on top.)

---

### The Whole Picture

```
1. You start Python
   → PYTHONPATH (OS setting) is read → fills sys.path with folders

2. You write: import mod
   → Step 1: is "mod" a builtin? (sys, math, etc.) → if yes, use it
   → Step 2: search sys.path folders, in order, for "mod.py"
            (current folder first!) → use the first match

3. Not found anywhere? → ModuleNotFoundError
```

---

### A Quick Analogy

> Importing a module is like **looking for a book**:
> - **Step 1 (builtins):** first check the small shelf of "always-here" reference books built into your desk (sys, math...).
> - **Step 2 (sys.path):** if not there, walk through a **list of rooms** (folders), in order, checking each for the book — and you check **your own room first** (current folder), then the library's standard rooms.
> - **PYTHONPATH** is how you tell the system *"also check these extra rooms I added."*
>
> You grab the **first copy** you find — which is why a book in your own room (a file named `random.py`) gets picked up before the library's official copy.

---

### The One-Sentence Summary

> When you `import mod`, Python first checks if `mod` is a **built-in** module (like `sys` or `math`); if not, it searches for a file `mod.py` through the folders listed in **`sys.path`**, in order, using the **first** match — and since the **current folder is searched first**, naming your own file after a standard module (like `random.py`) accidentally shadows the real one. `sys.path` is initially filled from the **`PYTHONPATH`** environment variable (an OS setting listing extra folders to search), plus Python's own standard folders. 🎯
