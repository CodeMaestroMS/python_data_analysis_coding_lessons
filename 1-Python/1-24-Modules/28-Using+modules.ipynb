Let’s say that we need to use the `cosine` function. This function, and many other mathematical functions are located in the `math` module. To tell Python that we want to access the features offered by this module, we can give the statement `import math`. Now the module is loaded into memory. We can now call the function like this:

```python
import math

math.cos(0)
```

Note that we need to include the module name where the cos function is found. This is because other modules may have a function (or other attribute of a module) with the same name. This usage of different namespace for each module prevents name clashes. For example, functions gzip.open and os.open are not to be confused with the builtin open function.

Namespace (නාමඅවකාශය) කියන්නේ සරලවම කිව්වොත් විවිධ දේවල්වලට (විශේෂයෙන්ම functions, variables, classes වැනි දේවලට) නම් තැබීමේදී ඒවා එකිනෙක පැටලෙන්නේ නැතිව වෙන් කරලා හඳුනාගන්න හදපු "නම් එකතුවක්" හෝ "පරිසරයක්" (Container / Boundary) එකකටයි.
ඔයා දීලා තියෙන උදාහරණයෙන්ම මේක ලේසියෙන් තේරුම් ගන්න පුළුවන්:
Python වල file එකක් open කරන්න සාමාන්‍යයෙන් open() කියන බිල්ට්-ඉන් (builtin) function එක තියෙනවා. හැබැයි:

   1. os කියන මොඩියුලයේත් open() කියලා function එකක් තියෙනවා.
   2. gzip කියන මොඩියුලයේත් open() කියලා function එකක් තියෙනවා.

දැන් අපි නිකන්ම open() කියලා කෝඩ් එකේ ලිව්වොත්, පරිගණකයට (හෝ Python වලට) එකපාරටම හිතාගන්න බෑ අපි මේ කියන්නේ සාමාන්‍ය එකද, os එකේද, නැත්නම් gzip එකේද කියලා. අන්න ඒ වගේ වෙලාවට නම් එකිනෙක ගැටීම (Name Clashes) වැළැක්වීමට අපි Namespace පාවිච්චි කරනවා.
අපි මොඩියුලයේ නම ඉස්සරහින් දාලා ලියන්නේ ඒ නිසයි:

* os.open() — මෙතනදී os කියන්නේ namespace එක. ඒ කියන්නේ "os කියන පරිසරය ඇතුළේ තියෙන open function එක ක්‍රියාත්මක කරන්න" කියන එකයි.
* gzip.open() — මෙතනදී gzip කියන්නේ වෙනමම namespace එකක්.

## එදිනෙදා ජීවිතයෙන් සරල උදාහරණයක්:
හිතන්න ඔයාගේ යාළුවෝ දෙන්නෙක් ඉන්නවා "කසුන්" කියලා. එක්කෙනෙක් ඉන්නේ A පන්තියේ, අනෙක් කෙනා ඉන්නේ B පන්තියේ.
ඔයා නිකන්ම "කසුන්" කිව්වොත් කවුද කියලා පැහැදිලි නැහැ. හැබැයි ඔයා "A පන්තියේ කසුන්" හෝ "B පන්තියේ කසුන්" කිව්වොත් කවුද කියලා පැහැදිලියි. මෙතනදී "A පන්තිය" සහ "B පන්තිය" කියන්නේ හරියට Namespace දෙකක් වගෙයි. ඒවයින් කරන්නේ එකම නම තියෙන පුද්ගලයන් දෙන්නෙක්ව පැටලෙන්නේ නැතිව වෙන් කරලා දෙන එකයි.
කෙටියෙන් කිව්වොත්, Programming වලදී Namespace එකක් මඟින් කරන්නේ එකම නම තියෙන විවිධ functions හෝ variables එකිනෙකා සමඟ පටලවා නොගෙන, ස්වාධීනව පාවිච්චි කරන්න ඉඩ සලසා දීමයි.
පයිතන් (Python) වල කෝඩ් ලියද්දී import cos සහ from math import cos කියන ක්‍රම දෙකේදී Namespace වෙනස් වෙන්නේ කොහොමද කියලා උදාහරණයකින් බලන්න කැමතිද?



Namespace (නාමඅවකාශය) කියන්නේ සරලවම කිව්වොත් විවිධ දේවල්වලට (විශේෂයෙන්ම functions, variables, classes වැනි දේවලට) නම් තැබීමේදී ඒවා එකිනෙක පැටලෙන්නේ නැතිව වෙන් කරලා හඳුනාගන්න හදපු "නම් එකතුවක්" හෝ "පරිසරයක්" (Container / Boundary) එකකටයි.
ඔයා දීලා තියෙන උදාහරණයෙන්ම මේක ලේසියෙන් තේරුම් ගන්න පුළුවන්:
Python වල file එකක් open කරන්න සාමාන්‍යයෙන් open() කියන බිල්ට්-ඉන් (builtin) function එක තියෙනවා. හැබැයි:

   1. os කියන මොඩියුලයේත් open() කියලා function එකක් තියෙනවා.
   2. gzip කියන මොඩියුලයේත් open() කියලා function එකක් තියෙනවා.

දැන් අපි නිකන්ම open() කියලා කෝඩ් එකේ ලිව්වොත්, පරිගණකයට (හෝ Python වලට) එකපාරටම හිතාගන්න බෑ අපි මේ කියන්නේ සාමාන්‍ය එකද, os එකේද, නැත්නම් gzip එකේද කියලා. අන්න ඒ වගේ වෙලාවට නම් එකිනෙක ගැටීම (Name Clashes) වැළැක්වීමට අපි Namespace පාවිච්චි කරනවා.
අපි මොඩියුලයේ නම ඉස්සරහින් දාලා ලියන්නේ ඒ නිසයි:

* os.open() — මෙතනදී os කියන්නේ namespace එක. ඒ කියන්නේ "os කියන පරිසරය ඇතුළේ තියෙන open function එක ක්‍රියාත්මක කරන්න" කියන එකයි.
* gzip.open() — මෙතනදී gzip කියන්නේ වෙනමම namespace එකක්.

## එදිනෙදා ජීවිතයෙන් සරල උදාහරණයක්:
හිතන්න ඔයාගේ යාළුවෝ දෙන්නෙක් ඉන්නවා "කසුන්" කියලා. එක්කෙනෙක් ඉන්නේ A පන්තියේ, අනෙක් කෙනා ඉන්නේ B පන්තියේ.
ඔයා නිකන්ම "කසුන්" කිව්වොත් කවුද කියලා පැහැදිලි නැහැ. හැබැයි ඔයා "A පන්තියේ කසුන්" හෝ "B පන්තියේ කසුන්" කිව්වොත් කවුද කියලා පැහැදිලියි. මෙතනදී "A පන්තිය" සහ "B පන්තිය" කියන්නේ හරියට Namespace දෙකක් වගෙයි. ඒවයින් කරන්නේ එකම නම තියෙන පුද්ගලයන් දෙන්නෙක්ව පැටලෙන්නේ නැතිව වෙන් කරලා දෙන එකයි.
කෙටියෙන් කිව්වොත්, Programming වලදී Namespace එකක් මඟින් කරන්නේ එකම නම තියෙන විවිධ functions හෝ variables එකිනෙකා සමඟ පටලවා නොගෙන, ස්වාධීනව පාවිච්චි කරන්න ඉඩ සලසා දීමයි.
පයිතන් (Python) වල කෝඩ් ලියද්දී import cos සහ from math import cos කියන ක්‍රම දෙකේදී Namespace වෙනස් වෙන්නේ කොහොමද කියලා උදාහරණයකින් බලන්න කැමතිද?



ඔයාට ඒක පැටලෙන එක සම්පූර්ණයෙන්ම සාධාරණයි. මොකද මේ දෙකම එකිනෙකට තදින්ම සම්බන්ධයි. මෙහෙම හිතන්න: Module කියන්නේ වස්තුව (Object) නම්, Namespace කියන්නේ ඒක ක්‍රියාත්මක වෙන ක්‍රමය (Concept) ලු.
තවත් ලේසියෙන්, සැබෑ ජීවිතයේ උදාහරණයකින්ම මේක පැහැදිලි කරගමු:
හිතන්න ඔයා ළඟ දිස්ත්‍රික්ක දෙකක සිතියම් පොත් (Maps) දෙකක් තියෙනවා කියලා.

   1. කොළඹ සිතියම් පොත (Colombo Map)
   2. මහනුවර සිතියම් පොත (Kandy Map)

දැන් මේ සිතියම් පොත් දෙකටම භෞතිකව පෙනෙන්න තියෙනවා. අන්න ඒ පොත් දෙක තමයි Modules. (පයිතන් වල os සහ gzip වගේ).

* Module 1: කොළඹ සිතියම් පොත
* Module 2: මහනුවර සිතියම් පොත

දැන් මේ පොත් දෙක ඇතුළේ බලද්දී, පොත් දෙකේම "ගාලු පාර" (Galle Road) කියලා පාරක් තියෙනවා (හිතන්නකෝ මහනුවරත් ගාලු පාර කියලා එකක් තියෙනවා කියලා).
දැන් ඔයා නිකන්ම කාට හරි "ගාලු පාරට යන්න" කිව්වොත් එයා අහනවා "කොහේ තියෙන ගාලු පාරද?" කියලා. අන්න ඒ වෙලාවට ඔයා කියනවා:

* කොළඹ.ගාලු පාර (Colombo.Galle Road)
* මහනුවර.ගාලු පාර (Kandy.Galle Road)

මෙතනදී "කොළඹ" සහ "මහනුවර" කියලා ඔයා පාරවල් දෙක වෙන් කරලා අඳුන්නලා දෙන්න පාවිච්චි කරපු "පරිසරය" (Boundary එක) තමයි Namespace කියන්නේ.
## පයිතන් වලට හැරුණොත්:

* os සහ gzip කියන ෆයිල් දෙක (පොත් දෙක) Modules වේ.
* හැබැයි අපි කෝඩ් එක ලියද්දී os.open() සහ gzip.open() කියලා ලියලා, Python වලට ඒ ඒ සීමාවන් වෙන් කරලා දෙන ක්‍රියාවලියට (System එකට) අපි කියනවා Namespace පාවිච්චි කරනවා කියලා.

කෙටියෙන්ම කිව්වොත්: Module එකක් ඇතුළේ තියෙන දේවල්, තවත් Module එකක දේවල් එක්ක එකතු වෙලා "කැඳ හැලියක්" නොවී, තනි තනියම වෙන් වෙලා පවතින්නේ මේ "Namespace" කියන සංකල්පය නිසයි.
දැන් අදහස පැහැදිලිද, නැත්නම් තව වෙනස් විදිහක උදාහරණයකට යමුද?
මේ අදහස තේරුම් ගන්න තව දුරකථන අංක (Phone numbers) හෝ පවුල්වල නම් (Family Names) වගේ වෙනත් සරල උදාහරණයකින් පැහැදිලි කරන්නද?
