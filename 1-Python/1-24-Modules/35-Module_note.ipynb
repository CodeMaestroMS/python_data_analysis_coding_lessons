# Python Modules

*A simple, complete note — modules, imports, packages, and `__main__`*

---

## 1. What Is a Module?

- **A module = one Python file** (any file ending in `.py`).
- It holds related code — functions, classes, variables — grouped together.
- The point: split a big program into smaller, independent pieces.
- No special syntax needed — **any `.py` file is already a module.** The file name (without `.py`) is the module name.

**Common standard-library modules:** `re`  `math`  `random`  `os`  `sys`  (Python ships with hundreds.)

---

## 2. Using a Module — `import`

To use a module's features, **import** it first — this loads it into memory:

```python
import math
math.cos(0)      # → 1.0
```

**You must write the module name before the function** (`math.cos`, not just `cos`). This is the **namespace** — it prevents name clashes, because two different modules can have functions with the same name.

**Example:** `gzip.open`, `os.open`, and the builtin `open` are three DIFFERENT functions — the namespace keeps them apart.

---

## 3. Importing a Name Directly (skip the namespace)

If a long namespace is tedious, bring a name straight into your file:

```python
from math import cos
cos(1)           # no 'math.' needed now
```

- **Import several names:** `from math import sin, cos, pi`
- **Import everything (avoid this):** `from math import *` — confusing, because you can't tell what names got pulled in. Use it only rarely.

| Import style | How you call it | When |
|---|---|---|
| `import math` | `math.cos(0)` | normal, clearest — keeps the namespace |
| `from math import cos` | `cos(0)` | when you use it a lot |
| `from math import cos as c` | `c(0)` | to rename (avoid clashes / shorten) |
| `from math import *` | `cos(0)` | avoid — unclear what got imported |

---

## 4. How Python Finds a Module

When you write `import mod`, Python searches in this order:

1. **Is it a builtin module?** (compiled into Python, like `sys`) → use it.
2. **Is there a file `mod.py` in any folder in `sys.path`?** → use the FIRST match found.

`sys.path` is a **list of folders** Python searches. The **first folder is the current one** — so your own files are checked before the standard library.

**The trap:** if you name your file `random.py`, then `import random` loads YOUR file (found first), not Python's real `random` module. Never name files after standard modules.

`sys.path` is first filled from the `PYTHONPATH` environment variable (an OS setting listing extra folders), plus Python's own standard folders.

---

## 5. Packages — Modules Organized Into Folders

- **A package = a folder of modules** (a module that can contain other modules/packages).
- This organizes the hundreds of standard modules into a hierarchy.

**Example — the numpy hierarchy:**

```
numpy          ← package
  .linalg      ← subpackage
    .linalg    ← module
      .det     ← function (computes a determinant)
```

### Importing from packages

```python
import numpy                      # the whole package
import numpy.linalg              # just the subpackage
import numpy.linalg.linalg       # just that module

from numpy.linalg import linalg          # skip the long path
from numpy.linalg import linalg as lin   # ...and rename it
from numpy.linalg.linalg import det      # import just the function
```

With the last line you call `det(...)` directly. Had you only done `import numpy`, you'd need the full `numpy.linalg.linalg.det(...)`.

### Folders = packages

A package is a folder containing a special `__init__.py` file (which sets the package up). The folder can hold more modules (`.py` files) or more sub-folders (subpackages):

```
a/                 ← package a
    __init__.py    ← makes 'a' a package
    b.py           ← module a.b
    c/             ← subpackage a.c
        __init__.py
        d.py       ← module a.c.d
        e.py       ← module a.c.e
```

## 6. What's Inside a Module

Every assignment, `def`, and `class` in a module creates an **attribute** of that module. After `import mod`, reach them with the dot:

```python
mod.f()        # a function attribute
mod.a          # a data attribute
mod.MyClass()  # a class attribute
```

**There are no true global variables in Python** — every variable belongs to some module's namespace.

### Inspecting a module

- `dir()` — lists the attributes of the current module (or `dir(mod)` for another).
- `mod.__dict__` — the module's attributes, stored as a dictionary (the namespace IS a dict).

**Common module attributes:**

| Attribute | What it holds |
|---|---|
| `__name__` | the module's name (special value in the main program — see section 7) |
| `__file__` | the module's filename (always present) |
| `__doc__` | the module's docstring (if the file's FIRST non-empty, non-comment line is a string) |
| `__version__` / `__author__` | optional metadata, if the module defines them |

## 7. Top-Level Code and `if __name__ == "__main__"`

**The problem:** code written at the top level of a module (not inside any function) runs automatically — even when someone just `import`s your module. That surprises the importer.

```python
# mod.py
for _ in range(3):
    print('Hello')     # runs on IMPORT too — surprising!
```

**The fix:** put such code inside a function (call it `main` by convention), and guard it:

```python
def main():
    for _ in range(3):
        print('Hello')

if __name__ == '__main__':   # only when run directly,
    main()                   # NOT when imported
```

### How the guard works

- `__name__` equals `"__main__"` when the file is **run directly** (e.g. `python3 mod.py`).
- `__name__` equals the **module's name** when the file was **imported** by someone else.
- So `main()` runs only on direct execution — an importer must call `mod.main()` explicitly if they want it.

**Note:** `main` has no special meaning in Python — it's just a naming convention.

---

## One-Page Recap

- **Module** = one `.py` file. **Package** = a folder of modules (with `__init__.py`).
- `import math` → call as `math.cos()` (keeps namespace, prevents clashes).
- `from math import cos` → call as `cos()`; add `as c` to rename; avoid `import *`.
- **Lookup:** builtin first, then `mod.py` in `sys.path` (current folder first → don't shadow stdlib names).
- **Dotted imports:** `from numpy.linalg.linalg import det` pulls one function out of a deep package.
- **Attributes:** everything in a module is an attribute (`mod.f`, `mod.a`); inspect with `dir()` and `mod.__dict__`.
- `if __name__ == "__main__": main()` → run code only when the file is executed directly, not when imported.
