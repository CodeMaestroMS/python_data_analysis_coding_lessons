In the print function, numerical expressions are first evaluated and then automatically converted to strings. Subsequently the strings are concatenated with spaces:

````python
print(1,'plus',2,'equals', 1+2)
````

## How `print()` Works Internally

When Python executes `print(1, "plus", 2, "equals", 1+2)`, it happens in **three stages**:

### Stage 1 — Arguments are Evaluated
Before `print()` even runs, Python evaluates each argument left to right:

| Argument | Type | Result |
|---|---|---|
| `1` | int literal | `1` |
| `"plus"` | str literal | `"plus"` |
| `2` | int literal | `2` |
| `"equals"` | str literal | `"equals"` |
| `1+2` | expression | `3` |

### Stage 2 — `str()` is Called on Each Argument
`print()` internally calls `str()` on every argument it receives:

```python
str(1)      → "1"
str("plus") → "plus"   # strings stay as-is
str(2)      → "2"
str("equals") → "equals"
str(3)      → "3"
```

### Stage 3 — Strings are Joined with a Space
`print()` uses `sep` (default `" "`) to join all the converted strings:

```python
" ".join(["1", "plus", "2", "equals", "3"])
→ "1 plus 2 equals 3"
```

---

### The Full Picture

```python
print(1, "plus", 2, "equals", 1+2)
#       ↓ evaluate args
#   1, "plus", 2, "equals", 3
#       ↓ str() each one
#   "1", "plus", "2", "equals", "3"
#       ↓ join with sep=" "
#   "1 plus 2 equals 3"          ← printed to stdout
```

---

### Useful to Know
You can customize both the separator and the ending:

```python
print(1, 2, 3, sep="-")   # → 1-2-3
print(1, 2, 3, end="!\n") # → 1 2 3!
```

The key takeaway: **expression evaluation happens first** (by Python itself), then **`str()` conversion and joining** happen inside `print()`.


<span style="color: red;">But it not only evaluate expression (in this case 1+2) right? Its' evaluate each arguments.</span>

You're absolutely right! Every argument is evaluated — not just expressions like `1+2`.

Even `1`, `"plus"`, and `2` are evaluated — they're just **literals**, which evaluate to themselves:

```python
1       → evaluates to the integer 1
"plus"  → evaluates to the string "plus"
1+2     → evaluates to the integer 3
```

The difference is just the **complexity** of the evaluation:

| Argument | Kind | Evaluation |
|---|---|---|
| `1` | int literal | trivially returns `1` |
| `"plus"` | str literal | trivially returns `"plus"` |
| `2` | int literal | trivially returns `2` |
| `"equals"` | str literal | trivially returns `"equals"` |
| `1+2` | expression | actually computes `3` |

So a more accurate way to say it: **Python evaluates all arguments before calling `print()`**, it's just that evaluating a plain literal is an instant no-op, while `1+2` requires an actual addition step.
