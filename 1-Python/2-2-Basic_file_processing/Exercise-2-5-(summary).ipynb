If some line doesn’t represent a number, you can just ignore that line. You can achieve this with the try-except block. An example of recovering from an exceptional situation:


try:
    x = float(line)           # The float constructor raises ValueError exception if conversion is no possible
except ValueError:
    # Statements in here are executed when the above conversion fails
We will cover more about exceptions later in the course.

```python
#!/usr/bin/env python3

import sys
import statistics

def summary(filename):
    """Compute sum, mean, and standard deviation of numbers in a file.

    Each non-blank line of the file must contain one number.Lines that cannot
    be parsed as a float are silently skipped.

    Args:
        filename: Path to a file with one number per line.

    Returns:
        A tuple (total, mean, stdev) of the file's numeric contents.
    """
    numbers = []
    with open(filename) as f:
        for line in f:
            line = line.strip()
            if not line:
                continue
            try:
                numbers.append(float(line))
            except ValueError:
                continue    # not a valid number — skip this line

    total = sum(numbers)
    avg = statistics.mean(numbers)
    stdev = statistics.stdev(numbers)
    return (total, avg, stdev)

def main():
    for filename in sys.argv[1:]:
        total, avg, stdev = summary(filename)
        print(f"File: {filename} Sum: {total:.6f} Average: {avg:.6f} Stddev: {stdev:.6f}")


if __name__ == "__main__":
    main()
```
