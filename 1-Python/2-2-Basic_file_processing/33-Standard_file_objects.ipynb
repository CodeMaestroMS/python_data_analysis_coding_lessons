Standard file objects

Python has automatically three file objects open:

1. ```sys.stdin``` for standard input
2. ```sys.stdout``` for standard output
3. ```sys.stderr``` for standard error

To read a line from a user (keyboard), you can call ```sys.stdin.readline()```.

To write a line to a user (screen), call ```sys.stdout.write(line)```

The standard error is meant for error messages only, even though its output often goes to the same destination as standard output.

The ```print``` function uses the file ```sys.stdout```

```input``` function uses the file ```sys.stdin```. An example of usage:

````python
import sys
import random

i=random.randint(-10,10)

if i>=0:
    sys.stdout.write(f'Got a positive integer or zero: {i}\n')
else:
    sys.stderr.write(f'Got a negative integer: {i}\n')
````

These standard file objects are meant to be a basic input/output mechanism in textual form. The destinations of the file objects can be changed to point somewhere else than the usual keyboard and screen. Very often these are redirected to some files. For example, it is usual to point the stderr to a file where all error messages are logged.
