````python
from collections import defaultdict

data = [
    (['liikuttaa'], 'move'),
    (['piilottaa', 'salata'], 'hide'),
    (['kuusi'], 'six'),
    (['kuusi'], 'fir'),
]

result = defaultdict(list)
for fi_words, en_word in data:
    for fi in fi_words:
        result[fi].append(en_word)

print(dict(result))
# → {'liikuttaa': ['move'], 'piilottaa': ['hide'], 'salata': ['hide'], 'kuusi': ['six', 'fir']}
````

## Yes — Using `defaultdict` from `collections`

This is a great example of where `collections.defaultdict` shines. The transformation is:

- Each Finnish word becomes a **key**
- Each English word becomes a **value in a list**
- If a Finnish word maps to multiple English meanings (like `kuusi`), they should be **collected in one list**

---

### The Cleanest Solution

```python
from collections import defaultdict

data = [
    (['liikuttaa'], 'move'),
    (['piilottaa', 'salata'], 'hide'),
    (['kuusi'], 'six'),
    (['kuusi'], 'fir'),
]

result = defaultdict(list)
for fi_words, en_word in data:
    for fi in fi_words:
        result[fi].append(en_word)

print(dict(result))
# → {'liikuttaa': ['move'], 'piilottaa': ['hide'], 'salata': ['hide'], 'kuusi': ['six', 'fir']}
```

---

### Why `defaultdict` is Perfect Here

Normally, you'd have to check *"has this key been added before?"* on every insertion:

```python
# Without defaultdict — verbose:
result = {}
for fi_words, en_word in data:
    for fi in fi_words:
        if fi not in result:
            result[fi] = []          # initialize if missing
        result[fi].append(en_word)
```

With `defaultdict(list)`, the initialization is **automatic** — any missing key gets a fresh empty list on first access:

```python
# With defaultdict — clean:
result = defaultdict(list)
for fi_words, en_word in data:
    for fi in fi_words:
        result[fi].append(en_word)   # no check needed!
```

---

### How `defaultdict(list)` Works Under the Hood

The argument `list` is a **factory function** — Python calls it whenever you access a missing key:

```python
d = defaultdict(list)
print(d["new_key"])   # → []   (list() was called automatically)
```

So `d[fi].append(en_word)` always works, even the first time you touch a key. The empty list gets created invisibly, then `.append()` adds the value.

You can use other factory functions too:
- `defaultdict(int)` → missing keys default to `0` (great for counters)
- `defaultdict(set)` → missing keys default to empty sets
- `defaultdict(str)` → missing keys default to empty strings

---

### Alternative — Using `setdefault()` (built-in)

If you don't want to import anything, you can achieve the same with the regular `dict.setdefault()` method:

```python
data = [
    (['liikuttaa'], 'move'),
    (['piilottaa', 'salata'], 'hide'),
    (['kuusi'], 'six'),
    (['kuusi'], 'fir'),
]

result = {}
for fi_words, en_word in data:
    for fi in fi_words:
        result.setdefault(fi, []).append(en_word)

print(result)
# → {'liikuttaa': ['move'], 'piilottaa': ['hide'], 'salata': ['hide'], 'kuusi': ['six', 'fir']}
```

`setdefault(fi, [])` says: *"if `fi` exists, return its value; if not, create it with `[]` and return that"*. Then `.append(en_word)` adds to whichever list you got back.

---

### Comparison

| Approach | Import needed | Style |
|---|---|---|
| `defaultdict(list)` | Yes (`collections`) | Most Pythonic for grouping |
| `setdefault([])` | No | Standard, no imports |
| Manual `if-not-in` check | No | Verbose, error-prone |

---

### Simple Analogy

> Imagine you're **organizing translations into folders** 📁:
>
> - You go through each translation card (each pair)
> - Some cards have **multiple Finnish words** on them (like *piilottaa* AND *salata* → both mean "hide")
> - For each Finnish word, you **file the English meaning** into that word's folder
> - If a folder doesn't exist yet, you **create it on the fly**
>
> `defaultdict(list)` is your **auto-creating filing cabinet** — you never have to worry about whether the folder exists. It just does.

---

### The Key Idea

> For grouping values under keys (especially with **multiple keys per value** or **multiple values per key**), `defaultdict(list)` from `collections` is the standard Python idiom:
>
> ```python
> result = defaultdict(list)
> for keys, value in pairs:
>     for k in keys:
>         result[k].append(value)
> ```
>
> If you can't import, use `dict.setdefault(key, []).append(value)` — same behavior, no import needed. Both are cleaner than manual `if key not in dict` checks.
