## List Mutating Methods — Quick Notes

These methods **change the list in place** (they don't return a new list — they modify the original).

---

### 1. `append(x)` — add one item to the end
```python
L = [1, 2, 3]
L.append(4)
print(L)   # → [1, 2, 3, 4]
```
> Adds **one** item to the end. Even if you give a list, it treats it as one item:
```python
L.append([5, 6])
print(L)   # → [1, 2, 3, 4, [5, 6]]   ← nested!
```

---

### 2. `extend(iterable)` — add many items to the end
```python
L = [1, 2, 3]
L.extend([4, 5])
print(L)   # → [1, 2, 3, 4, 5]
```
> Takes each item from the iterable and adds them individually. Compare with append:
```python
L = [1, 2, 3]
L.append([4, 5])   # → [1, 2, 3, [4, 5]]   nested
L.extend([4, 5])   # → [1, 2, 3, 4, 5]     flat
```

---

### 3. `insert(index, x)` — add item at a specific position
```python
L = [1, 2, 3]
L.insert(1, 99)
print(L)   # → [1, 99, 2, 3]
```
> Inserts `x` at position `index`, shifting everything else to the right.

---

### 4. `remove(x)` — delete the first occurrence of a value
```python
L = [1, 2, 3, 2]
L.remove(2)
print(L)   # → [1, 3, 2]   ← only the FIRST 2 is removed
```
> Removes by **value**, not by position. Raises an error if the value isn't found.

---

### 5. `pop(index=-1)` — remove and return an item by position
```python
L = [1, 2, 3]
last = L.pop()      # removes and returns 3
print(L)            # → [1, 2]
print(last)         # → 3

first = L.pop(0)    # removes and returns first item
print(L)            # → [2]
```
> **Removes by position** — default is the last item. Also **returns** the removed value (useful!).

---

### 6. `reverse()` — flip the list in place
```python
L = [1, 2, 3]
L.reverse()
print(L)   # → [3, 2, 1]
```
> Reverses the order of all items. Returns nothing — modifies the list directly.

---

### 7. `sort()` — sort the list in place
```python
L = [3, 1, 2]
L.sort()
print(L)   # → [1, 2, 3]

L.sort(reverse=True)
print(L)   # → [3, 2, 1]
```
> Sorts ascending by default. Use `reverse=True` for descending. Returns nothing — modifies the list directly.

---

### Quick Cheat Sheet

| Method | What it does | Example |
|---|---|---|
| `append(x)` | add ONE item to end | `L.append(4)` |
| `extend(iter)` | add MANY items to end | `L.extend([4,5])` |
| `insert(i, x)` | add item at position i | `L.insert(1, 99)` |
| `remove(x)` | delete first occurrence of value x | `L.remove(2)` |
| `pop(i)` | remove & return item at position i | `L.pop(0)` |
| `reverse()` | flip list order | `L.reverse()` |
| `sort()` | sort ascending | `L.sort()` |

---

### Two important gotchas

**1. These methods return `None`, not the list:**
```python
L = [3, 1, 2]
result = L.sort()
print(result)   # → None    ← NOT [1, 2, 3]!
print(L)        # → [1, 2, 3]  ← the list itself changed
```
So **don't** write `L = L.sort()` — you'll lose your list!

**2. `remove()` vs `pop()`:**
```python
L = [10, 20, 30]

L.remove(20)   # remove by VALUE (20)     → [10, 30]
L.pop(1)       # remove by POSITION (1)   → [10]
```

---

### Simple analogy

> Imagine your list is a **stack of books** 📚
>
> | Method | Action |
> |---|---|
> | `append` | put ONE new book on top |
> | `extend` | dump a pile of books on top (one by one) |
> | `insert` | slide a book into a specific slot |
> | `remove` | take out the first book with a specific title |
> | `pop` | take a book from a specific position (or the top) |
> | `reverse` | flip the whole stack upside down |
> | `sort` | reorder the books alphabetically |

---

### The key idea
> All these methods **modify the original list** and return `None`. They're called **mutating methods** — they change (mutate) the list in place, rather than creating a new one.
