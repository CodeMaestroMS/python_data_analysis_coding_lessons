### String handling

We have already seen how to index, slice, concatenate, and repeat strings. Let's now look into what methods the `str` class offers. In Python strings are immutable. This means that for instance the following assignment is not legal:

````python
s='Text'
s[0]='a' # This is not legal in Python
````

Because of the immutability of the strings, the string methods work by returning a value; they don't have any side-effects. In the rest of this section we briefly describe several of these methods. The methods are here divided into five groups.

## String Methods Return Values, Not Side-Effects — Simply

This is explaining an important consequence of **string immutability** (which you learned earlier — strings can't be changed once created). Because strings can't be modified in place, string methods must **give you back a new string** rather than altering the original. Let me unpack it.

---

### Recall — Strings Are Immutable

From your earlier lessons: a string, once created, **cannot be changed**:

```python
s = "hello"
s[0] = "H"     # ✗ TypeError — can't modify a string in place!
```

There's no way to reach into a string and alter it. So how do string methods like `.upper()` or `.replace()` "change" a string? They **don't change it** — they **create and return a brand-new string**.

---

### "Return a Value" vs "Side-Effect" — The Key Distinction

**Return a value** = the method **hands back a new result**, leaving the original untouched:

```python
s = "hello"
result = s.upper()      # returns a NEW string "HELLO"

print(result)     # → HELLO    (the new string)
print(s)           # → hello    (the ORIGINAL is UNCHANGED!)
```

`s.upper()` didn't modify `s` — it **created a new string** `"HELLO"` and returned it. The original `s` is still `"hello"`.

**Side-effect** = a method **modifies the object itself** (changes it in place), rather than returning a new one. String methods do **NOT** do this.

---

### Contrast With Lists — Which DO Have Side-Effects

Lists are **mutable**, so their methods often modify the list in place (a side-effect) and return `None`:

```python
L = [3, 1, 2]
result = L.sort()      # sorts L IN PLACE (side-effect!)

print(L)          # → [1, 2, 3]    the list ITSELF changed
print(result)      # → None          .sort() returns nothing!
```

`L.sort()` **changed `L` directly** (a side-effect) and gave back `None`. That's the opposite of how strings work.

---

### Side-by-Side — String vs List

```python
# STRING — returns a new value, no side-effect:
s = "hello"
s.upper()         # returns "HELLO"
print(s)           # → hello    (unchanged — you must CAPTURE the return value)

# LIST — side-effect, modifies in place:
L = [3, 1, 2]
L.sort()          # returns None, but CHANGES L
print(L)           # → [1, 2, 3]  (the list itself was modified)
```

| | String methods | List methods (like sort) |
|---|---|---|
| Modify the original? | **No** (can't — immutable) | Yes (mutable) |
| Return | a **new** string | often `None` |
| How to use | **capture** the return value | just call it |

---

### The Practical Consequence — You MUST Capture the Return Value

Because string methods return a new string (and don't change the original), you have to **save** the result or it's lost:

```python
s = "hello"

s.upper()              # ✗ result thrown away! s is still "hello"
print(s)                # → hello

s = s.upper()           # ✓ capture the returned new string back into s
print(s)                 # → HELLO
```

A very common beginner mistake: calling `s.upper()` and expecting `s` to change. It won't — you must **reassign** (`s = s.upper()`) to keep the result.

---

### Examples Across Common String Methods

Every string method follows this pattern — returns a new string, leaves the original alone:

```python
s = "  Hello World  "

s.strip()             # → "Hello World"    (returns new, s unchanged)
s.lower()              # → "  hello world  "
s.replace("o", "0")    # → "  Hell0 W0rld  "
s.upper()               # → "  HELLO WORLD  "

print(s)     # → "  Hello World  "   ← ORIGINAL untouched after ALL of these
```

None of them modified `s` — each just returned a new string based on it.

### The One-Sentence Summary

> Because strings are **immutable** (can't be changed in place), string methods can't modify the original — instead they **return a brand-new string** as their result (a "return value"), leaving the original completely unchanged. This is the opposite of "side-effect" methods like a list's `.sort()`, which modify the object itself. The practical upshot: you must **capture** a string method's return value (e.g. `s = s.upper()`), because just calling `s.upper()` returns a new string that's lost if you don't save it. ### Classification of strings

All the following methods will take no parameters and return a truth value. An empty string will always result in False.


- `s.isalnum()` True if all characters are letters or digits
- `s.isalpha()` True if all characters are letters
- `s.isdigit()` True if all characters are digits
- `s.islower()` True if contains letters, and all are lowercase
- `s.isupper()` True if contains letters, and all are uppercase
- `s.isspace()` True if all characters are whitespace
- `s.istitle()` True if uppercase in the beginning of word, elsewhere lowercase


### String transformations

The following methods do conversions between lower and uppercase characters in the string. All these methods return a new string.


- `s.lower()` Change all letters to lowercase
- `s.upper()` Change all letters to uppercase
- `s.capitalize()` Change the first character to upper case and change the rest to lower case
- `s.title()` Change to titlecase
-`s.swapcase()` Change all uppercase letters to lowercase, and vice versa`
