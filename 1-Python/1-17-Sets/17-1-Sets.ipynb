### Set

Set is a dynamic, unordered container. It works a bit like dictionary, but only the keys are stored. And each key can be stored only once. The set requires that the keys to be stored are hashable. Below are a few ways of creating a set:

````python
s={1,1,1}
print(s)
````

````python
s=set([1,2,2,'a'])
print(s)
````

````python
s=set()  # Empty set
print(s)
````

````python
s.add(7)
print(s)
````

A more useful example:

````python
s='mississippi'
print(f'There are {len(set(s))} distinct characters in {s}.')
````

The set provides the following non-mutating methods:

````python
s=set()
s1=set()
s.copy()
s.issubset(s1)
s.issuperset(s1)
s.union(s1)
s.intersection(s1)
s.difference(s1)
s.symmetric_difference(s1)
````

The last four operation can be tedious to write to create a more complicated expression. The alternative is to use the corresponding operator forms: `|, &, -, and ^`. An example of these:

````python
s={1,2,7}
t={2,8,9}
````

````python
s|t # Union
````

````python
s&t  #Intersection
````

````python
s-t  #Difference
````

````python
s^t #Symmetric difference
````

There are also the following mutating methods:

```
s.add(x)
s.clear()
s.discard()
s.pop()
s.remove(x)
```

And the set operators ```|, &, -, and ^``` have the corresponding mutating, augmented assignment forms: ```|=, &=, -=, and ^=```.

## Understanding Augmented Assignment for Sets

**Augmented assignment** means combining an operation with assignment into a single step. Instead of computing a result and reassigning it manually, you do both at once — and the operation **modifies the set in place**.

---

### The Basic Idea

Two ways to say the same thing:

```python
s = s | t     # regular:  compute s|t, then reassign to s
s |= t        # augmented: modify s in place with the union of t
```

Both give the same result, but the augmented form is:
- **Shorter** to write
- **Faster** (no new set is created — the existing one is modified)
- **In-place** (the set object itself changes, no new object)

---

### The Four Set Augmented Operators

```python
s = {1, 2, 3}
t = {3, 4, 5}
```

**Union assignment `|=`** — add everything from `t` into `s`:
```python
s |= t
print(s)   # → {1, 2, 3, 4, 5}
```
Same as `s.update(t)`.

**Intersection assignment `&=`** — keep only what's in both:
```python
s = {1, 2, 3}
s &= t
print(s)   # → {3}
```
Same as `s.intersection_update(t)`.

**Difference assignment `-=`** — remove elements that are also in `t`:
```python
s = {1, 2, 3}
s -= t
print(s)   # → {1, 2}
```
Same as `s.difference_update(t)`.

**Symmetric difference assignment `^=`** — keep only elements in exactly one:
```python
s = {1, 2, 3}
s ^= t
print(s)   # → {1, 2, 4, 5}
```
Same as `s.symmetric_difference_update(t)`.

---

### The Two Ways Side-by-Side

| Non-mutating (creates new set) | Mutating (modifies s) |
|---|---|
| `s = s \| t` | `s \|= t` |
| `s = s & t` | `s &= t` |
| `s = s - t` | `s -= t` |
| `s = s ^ t` | `s ^= t` |

The augmented form is really just a **shortcut** for the assignment version — but internally it's more efficient because it modifies the existing set.

---

### The Key Difference — Same Object vs New Object

The **regular form** creates a completely new set:
```python
s = {1, 2, 3}
original_id = id(s)

s = s | {4}         # new set is created and assigned back
print(id(s) == original_id)   # → False (different object!)
```

The **augmented form** modifies the existing set:
```python
s = {1, 2, 3}
original_id = id(s)

s |= {4}            # modifies s directly
print(id(s) == original_id)   # → True (same object!)
```

This matters when other variables also point to the same set:
```python
s = {1, 2, 3}
alias = s           # both point to the same set

s |= {4}
print(alias)        # → {1, 2, 3, 4}   alias sees the change!

# vs.

s = {1, 2, 3}
alias = s

s = s | {4}         # creates new set, s points to it
print(alias)        # → {1, 2, 3}      alias unchanged
```

---

### Familiar Pattern from Other Operators

Augmented assignment isn't unique to sets — Python uses the same pattern for arithmetic:

```python
x = x + 5    →    x += 5
x = x * 2    →    x *= 2
x = x - 1    →    x -= 1
```

For sets, the same idea applies to set operations:

```python
s = s | t    →    s |= t
s = s & t    →    s &= t
s = s - t    →    s -= t
s = s ^ t    →    s ^= t
```

Once you know the pattern for one, you know it for all.

---

### When to Use Which

**Use the augmented form (`|=`, `&=`, `-=`, `^=`)** when:
- You want to **update** a set based on another set
- You want to be efficient (no new object created)
- You're building up a set over time

```python
result = set()
for chunk in data:
    result |= chunk        # union in each chunk as you go
```

**Use the regular form (`|`, `&`, `-`, `^`)** when:
- You want to **preserve the original** set
- You need the result but don't want to change the input

```python
common = set_a & set_b     # get the intersection, both originals untouched
```

---

### Simple Analogy

> Imagine your set is a **basket of fruit** 🧺:
>
> **Regular form** (`s = s | t`) — you take the basket, empty it into a new basket along with the new fruit, then keep the new basket. The old basket is discarded.
>
> **Augmented form** (`s |= t`) — you just drop the new fruit **into the existing basket**. Same basket, more fruit.
>
> Both leave you with the combined fruit — but the second is cleaner and doesn't waste a basket.

---

### The Key Takeaway

> Augmented assignment (`|=`, `&=`, `-=`, `^=`) is the **mutating shortcut** for the corresponding set operators. It combines the operation with an in-place update — shorter to write, faster to run, and modifies the original set instead of creating a new one.
>
> **Rule of thumb:**
> - `s |= t` → "make s contain the union"
> - `s &= t` → "keep only what's also in t"
> - `s -= t` → "remove anything that's in t"
> - `s ^= t` → "keep only what's in one"
