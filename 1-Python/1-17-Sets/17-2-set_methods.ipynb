# Set Methods — Quick Notes

Sets are unordered collections of **unique elements**. Their methods come in three groups:
- **Non-mutating** — read-only, return new results
- **Mutating** — change the set in place
- **Operators** — shortcuts for the most common non-mutating methods

---
# Non-Mutating Methods

These methods **do not change** the original sets. They return new sets or booleans without altering the input.

## 1. `s.copy()` — make a copy of the set

Creates a new set with the same elements. Changes to the copy don't affect the original.

````python
s = {1, 2, 3}

s2 = s.copy()
s2.add(99)

print(s)   # → {1, 2, 3}       ← original unchanged
print(s2)  # → {1, 2, 3, 99}   ← copy modified
````

## 2. `s.issubset(s1)` — is s inside s1?

Returns `True` if every element of `s` is also in `s1`. Equivalent to the `<=` operator.

````python
s = {1, 2}
s1 = {1, 2, 3, 4}

print(s.issubset(s1))    # → True    (every element of s is in s1)
print(s1.issubset(s))    # → False   (s1 has elements s doesn't)

# Operator shortcut:
print(s <= s1)           # → True
````

## 3. `s.issuperset(s1)` — is s1 inside s?

Returns `True` if `s` contains every element of `s1`. Equivalent to the `>=` operator. Opposite of `issubset`.

````python
s = {1, 2, 3, 4}
s1 = {1, 2}

print(s.issuperset(s1))   # → True    (s contains all of s1)
print(s1.issuperset(s))   # → False   (s1 doesn't contain all of s)

# Operator shortcut:
print(s >= s1)            # → True
````

## 4. `s.union(s1)` — all elements from both

Returns a new set with **everything** from both sets. Duplicates appear only once (sets have unique elements).

````python
s = {1, 2, 7}
s1 = {2, 8, 9}

print(s.union(s1))   # → {1, 2, 7, 8, 9}

# Operator shortcut:
print(s | s1)        # → {1, 2, 7, 8, 9}
````

## 5. `s.intersection(s1)` — common elements only

Returns a new set with elements that appear in **both** sets. Returns an empty set if they share nothing.

````python
s = {1, 2, 7}
s1 = {2, 8, 9}

print(s.intersection(s1))   # → {2}

# Operator shortcut:
print(s & s1)               # → {2}
````

## 6. `s.difference(s1)` — in s but not in s1

Returns elements in `s` that are **NOT** in `s1`. Not symmetric: `s - s1` and `s1 - s` give different results.

````python
s = {1, 2, 7}
s1 = {2, 8, 9}

print(s.difference(s1))   # → {1, 7}

# Operator shortcut:
print(s - s1)             # → {1, 7}

# Note: reversed direction gives a different result!
print(s1 - s)             # → {8, 9}
````

## 7. `s.symmetric_difference(s1)` — in one, but not both

Returns elements in **either** set, but **not in both**. Excludes the intersection. Useful for finding what's different between two sets.

````python
s = {1, 2, 7}
s1 = {2, 8, 9}

print(s.symmetric_difference(s1))   # → {1, 7, 8, 9}

# Operator shortcut:
print(s ^ s1)                       # → {1, 7, 8, 9}
````

---
# Operator Shortcuts

For the four main set operations, Python provides **operator symbols** that are cleaner to write, especially in complex expressions.

| Operator | Method equivalent | Meaning |
|---|---|---|
| `\|` | `s.union(t)` | everything in either |
| `&` | `s.intersection(t)` | common to both |
| `-` | `s.difference(t)` | in s, not in t |
| `^` | `s.symmetric_difference(t)` | in one, not both |
| `<=` | `s.issubset(t)` | s is inside t |
| `>=` | `s.issuperset(t)` | s contains t |

````python
s = {1, 2, 7}
t = {2, 8, 9}

print('Union:',                s | t)   # → {1, 2, 7, 8, 9}
print('Intersection:',         s & t)   # → {2}
print('Difference:',           s - t)   # → {1, 7}
print('Symmetric difference:', s ^ t)   # → {1, 7, 8, 9}
````

# Mutating Methods

These methods **change the set in place** — the original set is modified.

## 1. `s.add(x)` — add one element

Adds `x` to the set. If it's already there, nothing happens — sets never contain duplicates.

````python
s = {1, 2, 3}

s.add(4)
print(s)   # → {1, 2, 3, 4}

s.add(2)   # already in set — no error, no change
print(s)   # → {1, 2, 3, 4}
````

## 2. `s.clear()` — remove everything

Empties the set completely.  
> **Note:** empty set prints as `set()`, not `{}` — because `{}` means an empty dict.

````python
s = {1, 2, 3}
s.clear()
print(s)   # → set()   (an empty set)
````

## 3. `s.discard(x)` — remove x if present (safe)

Removes `x` if it's in the set. **Does nothing** if `x` is missing — safe to use even when you're not sure the element exists.

````python
s = {1, 2, 3}

s.discard(2)
print(s)         # → {1, 3}

s.discard(99)    # not in set — no error!
print(s)         # → {1, 3}
````

## 4. `s.pop()` — remove and return an arbitrary element

Removes and returns an **arbitrary** element (sets have no order, so you can't predict which one).  
Raises `KeyError` if the set is empty.

````python
s = {1, 2, 3}

x = s.pop()
print(x)   # → 1 (or any other element — order is NOT guaranteed!)
print(s)   # → {2, 3}  (the removed element is gone)
````

## 5. `s.remove(x)` — remove x (strict)

Removes `x` from the set. **Raises `KeyError`** if `x` is not present.  
Use `discard()` instead if you're not sure whether the element exists.

````python
s = {1, 2, 3}

s.remove(2)
print(s)   # → {1, 3}

# s.remove(99)   # ✗ KeyError: 99
````

---
# ⚠️ Important Gotchas

## Gotcha 1: `discard()` vs `remove()`

- Use `discard` when you're **not sure** if the element is there.
- Use `remove` when you **want an error** if it's missing (like a safety check).

````python
s = {1, 2, 3}

s.discard(99)   # ✓ silently does nothing
# s.remove(99)  # ✗ KeyError: 99
````

## Gotcha 2: Empty set is `set()`, not `{}`

````python
empty_set = set()   # ✓ empty set
empty_dict = {}     # ✗ this is an empty DICT, not a set!

print(type(set()))  # → <class 'set'>
print(type({}))     # → <class 'dict'>

# Non-empty sets can still use curly braces:
my_set = {1, 2, 3}  # ✓ this IS a set
````

## Gotcha 3: Sets are unordered — `pop()` is unpredictable

````python
s = {'a', 'b', 'c'}

print(s.pop())   # could return 'a', 'b', or 'c' — you can't tell!

# Don't rely on order when iterating either:
for x in s:
    print(x)   # order NOT guaranteed
````

## Gotcha 4: Non-mutating methods RETURN new sets — they don't modify in place

````python
s = {1, 2, 3}
t = {3, 4, 5}

s.union(t)       # → returns {1,2,3,4,5} — but s is unchanged!
print(s)         # → {1, 2, 3}

# To save the result, reassign:
s = s.union(t)
print(s)         # → {1, 2, 3, 4, 5}

# Or use the mutating version:
# s.update(t)   ← modifies s in place
````

# Quick Cheat Sheet

## Non-Mutating Methods

| Method | What it does | Operator |
|---|---|---|
| `s.copy()` | make a copy of the set | — |
| `s.issubset(t)` | is s inside t? | `s <= t` |
| `s.issuperset(t)` | does s contain t? | `s >= t` |
| `s.union(t)` | all elements from both | `s \| t` |
| `s.intersection(t)` | common elements only | `s & t` |
| `s.difference(t)` | in s, not in t | `s - t` |
| `s.symmetric_difference(t)` | in one, not both | `s ^ t` |

## Mutating Methods

| Method | What it does | Error if missing? |
|---|---|---|
| `s.add(x)` | add x to set | no |
| `s.clear()` | remove all elements | — |
| `s.discard(x)` | remove x if present | no (safe) |
| `s.pop()` | remove & return any element | yes (if empty) |
| `s.remove(x)` | remove x | yes (KeyError) |

---
# 💡 The Key Idea

- **Non-mutating** methods (and their operators `|`, `&`, `-`, `^`) return **new sets** — the originals are untouched. Great for comparisons and combinations without side effects.

- **Mutating** methods (`add`, `discard`, `pop`, `remove`, `clear`) modify the set in place. Use them when you want to build up or shrink a set over time.

- **Operators** (`|`, `&`, `-`, `^`) are usually cleaner than the method form, especially in complex expressions:

```python
(A | B) - C          # ✓ clean and readable
A.union(B).difference(C)   # works, but harder to read
```
