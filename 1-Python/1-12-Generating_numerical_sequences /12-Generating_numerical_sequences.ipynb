````python
L=range(3)

for i in L:
    print(i)

print(L)
````

So L is not a list, but it is a sequence. We can for instance access its last element with L[-1]. If really needed, then it can be converted to a list with the list constructor:

````python
L=range(10)
print(list(L))
````

## Understanding Range vs List

Let's unpack this piece by piece.

---

### "L is not a list, but it is a sequence"

When you write `L = range(3)`, you might expect `L` to be a list like `[0, 1, 2]`. But it's **not** — it's a `range` object:

```python
L = range(3)
print(L)         # → range(0, 3)   ← not a list!
print(type(L))   # → <class 'range'>
```

However, `range` is still a **sequence** — it has all the sequence properties:
- **Ordered** (elements have positions: first, second, third...)
- **Indexable** (you can use `L[0]`, `L[1]`, etc.)
- **Iterable** (works in a `for` loop)
- **Supports `len()`**

So even though it's not a list, it behaves *like* a list in many ways.

**Recall from earlier:** sequences include strings, lists, tuples, **and ranges** — all sharing the same core operations.

---

### "We can for instance access its last element with L[-1]"

Since `range` is a sequence, you can **index** into it — including with negative indexes:

```python
L = range(3)     # represents 0, 1, 2

L[0]    # → 0    (first)
L[1]    # → 1    (second)
L[2]    # → 2    (third / last)
L[-1]   # → 2    (last using negative index)
L[-2]   # → 1    (second-to-last)
```

The `-1` index is a Python shortcut meaning **"the last element"** — it works on any sequence.

**Important — `range` computes on the fly:**
You didn't have to first turn it into a list to access an element. `range(3)[-1]` directly computes the last value **without generating all the numbers first**. That's why range is memory-efficient:

```python
range(1_000_000_000)[-1]   # → 999999999   ← instant, no memory used
```

If it were a real list, this would need 8+ GB of RAM.

---

### "If really needed, then it can be converted to a list with the list constructor"

Sometimes you actually **need** a real list — for example, to modify it, print all values, or use list-only methods like `.append()`.

You can convert with `list()`:

```python
L = range(3)
print(L)         # → range(0, 3)   ← still a range

my_list = list(L)
print(my_list)   # → [0, 1, 2]     ← now it's a real list
```

The word **constructor** means a function that **builds** a new object of a certain type. `list(...)` builds a new list.

Other constructors:
```python
list(range(5))    # → [0, 1, 2, 3, 4]
tuple(range(5))   # → (0, 1, 2, 3, 4)
str(range(5))     # → 'range(0, 5)'  (string version)
set(range(5))     # → {0, 1, 2, 3, 4}
```

---

### "If really needed" — why the emphasis?

The author is hinting: **don't convert unless you have to**.

**Why?** Because `range` is very efficient:
- Uses almost **no memory** (stores only start, stop, step — not every number)
- Generates values **on demand** during iteration
- Supports indexing, slicing, and `len()` without storing anything

Converting to a list **loses these benefits**:
```python
range(1_000_000)         # takes ~50 bytes of memory
list(range(1_000_000))   # takes ~8 MB of memory!
```

For a `for` loop, you don't need the list at all:
```python
# ✓ Efficient
for i in range(1_000_000):
    ...

# ✗ Wasteful — creates a huge list first, then loops
for i in list(range(1_000_000)):
    ...
```

Only convert when you specifically need a **modifiable list** or need to see all values at once.

---

### The three key ideas in this passage

| Idea | Meaning |
|---|---|
| `range` is not a list | but it's a **sequence** — behaves list-like |
| `L[-1]` works on ranges | because sequences support indexing |
| Convert with `list(L)` | but only if you **really** need to |

---

### Simple analogy

> Think of `range(1_000_000)` as a **recipe** 📜:
> - The recipe is small — a few lines of instructions
> - When you need each ingredient, the recipe **tells you what comes next**
> - You don't need the whole finished cake stored somewhere
>
> A **list** would be like **actually baking** all 1 million cakes and stacking them up.
>
> `list(range(1_000_000))` is: *"OK, fine — bake them all now and give me the pile."*
>
> Usually you just want the recipe — cheap, fast, and enough for looping.

---

### The key takeaway
> `range` acts like a **lightweight, on-demand sequence**. It looks and behaves like a list in most reading operations (indexing, slicing, iteration, length), but it's not actually a list. Convert with `list()` **only when necessary** — otherwise, use `range` directly for efficiency.

The range function works in similar fashion as slices. So, for instance the step of the sequence can be given:

````python
print(list(range(0,7,2)))
````
