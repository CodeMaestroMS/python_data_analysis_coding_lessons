A pattern represents a set of strings. This set can even be potentially infinite. They can be used to describe a set of strings that have some commonality; some regular structure. Regular expressions (RE) are a classical computer science topic. They are very common in programming tasks. Scripting languages, like Python, are very fluent in regular expressions. Very complex text processing can be achieved using regular expressions.

- In patterns, normal characters (letters, numbers) just represent themselves, unless preceded by a backslash, which may trigger some special meaning.

- Punctuation characters have special meaning, unless preceded by backslash (\), which deprives their special meaning.
- Use ```\\``` to represent a backslash character without any special meaning



In the following slides we will go through some of the more common RE notations.

```.``` Matches any character<br>
```[...]``` Matches any character contained within the brackets<br>
```[^...]``` Matches any character not appearing after the hat (ˆ)<br>
```^``` Matches the start of the string<br>
```$``` Matches the end of the string<br>
```*``` Matches zero or more previous RE<br>
```+``` Matches one or more previous RE<br>
```{m,n}``` Matches m to n occurences of previous RE<br>
```?``` Matches zero or one occurences of previous RE<br>

## Regex Special Characters — Explained Simply

Think of a regex as a **pattern to hunt for in text**. These special characters are the wildcard rules. Let's take each one with tiny, concrete examples — I'll use `re.search` (looks anywhere) and `re.findall` (finds all matches) throughout.

```python
import re
```

---

### `.` — Any Single Character

The dot is a **wildcard for exactly one character** (any character except newline):

```python
re.findall(r'c.t', 'cat cot cut c9t ct')
# → ['cat', 'cot', 'cut', 'c9t']        ← 'ct' NOT matched — dot needs ONE char
```

Pattern `c.t` reads: *"c, then any one character, then t."*

---

### `[...]` — One Character From This Set

Brackets = a **menu of allowed characters**, matching exactly one:

```python
re.findall(r'c[au]t', 'cat cot cut')
# → ['cat', 'cut']                       ← 'cot' rejected: o not on the menu

re.findall(r'[0-9]', 'room 4B, floor 12')
# → ['4', '1', '2']                      ← ranges work: 0-9, a-z, A-Z
```

You met this in your own code — `[A]pple`!

---

### `[^...]` — One Character NOT in This Set

The hat **inside brackets** flips the menu — *"anything except these"*:

```python
re.findall(r'c[^au]t', 'cat cot cut c9t')
# → ['cot', 'c9t']                       ← a and u now forbidden
```

You've seen this too: `[^\w\s]` in our punctuation-removal regex meant *"anything that's NOT a word character or whitespace."*

---

### `^` — Anchor: Start of String

The hat **outside brackets** means something totally different — *"the match must be at the very beginning"*:

```python
re.search(r'^Apple', 'Applepie')     # ✓ matches
re.search(r'^pie', 'Applepie')       # → None — pie isn't at the start
```

Same character, two meanings depending on position — inside `[...]` it negates, outside it anchors. A classic regex gotcha!

---

### `$` — Anchor: End of String

The mirror of `^` — *"the match must be at the very end"*:

```python
re.search(r'pie$', 'Applepie')       # ✓ matches
re.search(r'Apple$', 'Applepie')     # → None — Apple isn't at the end
```

Combined, `^...$` means *"the whole string must be exactly this"*:

```python
re.search(r'^[0-9]+$', '12345')      # ✓ string is ONLY digits
re.search(r'^[0-9]+$', '123a5')      # → None — the 'a' ruins it
```

---

### `*` — Zero or More of the Previous Thing

A **repetition counter** applying to whatever is just before it:

```python
re.findall(r'ab*c', 'ac abc abbc abbbc')
# → ['ac', 'abc', 'abbc', 'abbbc']       ← b repeated 0, 1, 2, 3 times — ALL ok
```

`ab*c` = *"a, then any number of b's (even none), then c."* Note zero is allowed — that's why plain `ac` matched.

---

### `+` — One or More of the Previous Thing

Same as `*` but **at least one is required**:

```python
re.findall(r'ab+c', 'ac abc abbc abbbc')
# → ['abc', 'abbc', 'abbbc']             ← 'ac' rejected: zero b's not enough
```

Memory hook: `+` means *"add at least one."*

---

### `{m,n}` — Between m and n Repetitions

Precise counting:

```python
re.findall(r'ab{2,3}c', 'abc abbc abbbc abbbbc')
# → ['abbc', 'abbbc']                    ← exactly 2–3 b's; 1 too few, 4 too many

re.search(r'^[0-9]{4}$', '2026')         # ✓ exactly 4 digits — a year!
re.search(r'^[0-9]{4}$', '123')          # → None
```

Variants: `{3}` = exactly 3, `{2,}` = 2 or more.

---

### `?` — Zero or One (Optional)

*"The previous thing may appear once — or not at all"*:

```python
re.findall(r'colou?r', 'color colour')
# → ['color', 'colour']                  ← u optional: both spellings match!

re.findall(r'-?[0-9]+', 'temp: -5 or 12')
# → ['-5', '12']                          ← optional minus sign
```

---

### Cheat Sheet

| Symbol | Meaning | Example | Matches |
|---|---|---|---|
| `.` | any one char | `c.t` | cat, c9t |
| `[au]` | one from the set | `c[au]t` | cat, cut |
| `[^au]` | one NOT in the set | `c[^au]t` | cot |
| `^` | start anchor | `^App` | **App**lepie |
| `$` | end anchor | `pie$` | Apple**pie** |
| `*` | 0 or more | `ab*c` | ac, abc, abbc |
| `+` | 1 or more | `ab+c` | abc, abbc |
| `{2,3}` | 2 to 3 times | `ab{2,3}c` | abbc, abbbc |
| `?` | 0 or 1 | `colou?r` | color, colour |

---

### The Repetition Family — One Picture

The last four are all the same idea, just different allowed counts:

```
?        →  {0,1}       optional
*        →  {0,∞}       any amount
+        →  {1,∞}       at least one
{m,n}    →  exact range
```

So really there's **one repetition mechanism** with four spellings — `?`, `*`, `+` are just shorthands for common `{m,n}` cases. 🎯

We have already seen that a | character denotes alternatives. For example, the pattern ```r"Get (on|off|ready)"``` matches the following strings: "Get on", "Get off", "Get ready".

We can use parentheses to create groupings inside a pattern: ```r"(ab)+"``` will match the strings "ab", "abab", "ababab", and so on

These groups are also given a reference number starting from ```1```.

We can refer to groups using backreferences: ```\number```.

For example, we can find separated patterns that get repeated: ```r"([a-z]{3,}) \1 \1"```
- This will recognise, for example, the following strings: "aca aca aca", "turn turn turn".
- But not the strings "aca aba aca" or "ac ac ac".


## Groups and Backreferences — Explained Simply

Two new ideas here, building on the repetition symbols you just learned. Let's take them one at a time.

---

### Idea 1 — Parentheses Group Things Together

Remember: `+` applies to **the thing just before it**. But what if you want to repeat a *multi-character chunk*?

```python
re.search(r'ab+', 'abbb')      # b+ → "a, then one or more b's" → matches abbb
re.search(r'(ab)+', 'ababab')  # (ab)+ → "one or more ab-chunks" → matches ababab
```

Without parentheses, `+` grabs only the single `b`. With parentheses, `+` treats `ab` as **one unit** and repeats the whole unit:

```
ab+     =  a  b  b  b  ...      (repeat just the b)
(ab)+   =  ab  ab  ab  ...      (repeat the pair)
```

Exactly like parentheses in math: `2x²` vs `(2x)²` — the grouping decides what the operation applies to. Same precedence lesson as `solve_quadratic`, appearing yet again!

---

### Idea 2 — Groups Get Numbered, and `\1` Means "That Same Text Again"

Every parenthesized group gets an automatic **reference number**: first `(` is group 1, second is group 2, and so on.

A **backreference** `\1` in the pattern means:

> *"Match here **exactly the same text** that group 1 captured."*

Not the same *pattern* — the same *actual text*. This is the key point.

---

### Walking Through the Example

```python
r"([a-z]{3,}) \1 \1"
```

Reading it piece by piece (all familiar symbols now!):

| Piece | Meaning |
|---|---|
| `([a-z]{3,})` | group 1: **3 or more** lowercase letters — capture whatever it was |
| ` ` | a space |
| `\1` | the **exact same text** group 1 captured |
| ` ` | a space |
| `\1` | that same text once more |

So the pattern means: *"a word of ≥3 letters, repeated three times, separated by spaces."*

---

### Testing the Four Strings from the Text

**`"aca aca aca"` ✓ matches**
```
([a-z]{3,}) captures "aca" → \1 must be "aca" → is ✓ → \1 must be "aca" → is ✓
```

**`"turn turn turn"` ✓ matches**
```
group 1 captures "turn" → both \1's find "turn" again ✓
```
Note: the group captured *different text* than before ("turn" vs "aca") — that's fine! `\1` only demands consistency **within one match**.

**`"aca aba aca"` ✗ fails**
```
group 1 captures "aca" → \1 expects "aca" → finds "aba" ✗ STOP
```
This is the crucial demo: `aba` **does** fit the pattern `[a-z]{3,}` — but `\1` doesn't ask for the pattern, it asks for the *captured text* `"aca"`. Echo, not re-match!

**`"ac ac ac"` ✗ fails**
```
[a-z]{3,} needs AT LEAST 3 letters → "ac" has only 2 ✗ group 1 never captures
```
Fails at the first hurdle — nothing to do with the backreference.

---

### See It in Python

```python
import re

m = re.search(r"([a-z]{3,}) \1 \1", "turn turn turn")
print(m.group())    # → 'turn turn turn'    the whole match
print(m.group(1))   # → 'turn'              what group 1 captured!

print(re.search(r"([a-z]{3,}) \1 \1", "aca aba aca"))   # → None
```

Notice `m.group(1)` — the same `.group()` method from your match-object question, now with the group number to extract just the captured piece.

---

### The One-Sentence Mental Model

> **Parentheses = "remember what matched here." Backreference `\1` = "now that exact text must appear again."**
>
> - `(ab)+` — parentheses as *grouping*: repeat the chunk
> - `(...)` ... `\1` — parentheses as *capture memory*: echo the chunk
>
> Think of group 1 as a little clipboard 📋: the first parentheses **copy** whatever they matched, and every `\1` **pastes** it as a requirement. Same-pattern-different-text won't do — it must be the identical captured text. 🎯

In the following, note that a hat (^) as the first character inside brackets will create a complement set of characters:

`\d` same as `[0-9]`, matches a digit<br>
`\D` same as `[ˆ0-9]`, matches anything but a digit<br>
`\s` matches a whitespace character (space, newline, tab, ... )<br>
`\S` matches a nonwhitespace character<br>
`\w` same as `[a-zA-Z0-9_]`, matches one alphanumeric character<br>
`\W` matches one non-alphanumeric character<br>

## The Shorthand Character Classes — Explained Simply

These are just **abbreviations** for bracket sets you already understand! Notice the beautiful pattern: each **lowercase** letter has an **UPPERCASE** twin that means the exact opposite (via the `[^...]` negation you learned).

```python
import re
```

---

### `\d` — a Digit

Shorthand for `[0-9]` — matches **one** digit character:

```python
re.findall(r'\d', 'Room 4B, floor 12')
# → ['4', '1', '2']                      one character each

re.findall(r'\d+', 'Room 4B, floor 12')
# → ['4', '12']                          + groups digit RUNS — whole numbers!
```

Combining with what you know: `\d{4}` = exactly four digits:

```python
re.search(r'^\d{4}$', '2026')    # ✓ a year
```

Much nicer than writing `[0-9][0-9][0-9][0-9]`!

---

### `\D` — anything BUT a Digit

Shorthand for `[^0-9]` — the complement:

```python
re.findall(r'\D+', 'Room 4B, floor 12')
# → ['Room ', 'B, floor ']               all the non-digit stretches
```

---

### `\s` — a Whitespace Character

Matches space, tab (`\t`), newline (`\n`)... — one "blank" character:

```python
re.split(r'\s+', 'hello   world\tfoo\nbar')
# → ['hello', 'world', 'foo', 'bar']     split on ANY blank run
```

(`re.split` is like `str.split` but pattern-powered — this one-liner handles messy mixed spacing that plain `.split(' ')` would fumble.)

---

### `\S` — a NON-Whitespace Character

The opposite — any visible character:

```python
re.findall(r'\S+', '  hello   world  ')
# → ['hello', 'world']                   grab the "words", ignore all spacing
```

---

### `\w` — a "Word" Character

Shorthand for `[a-zA-Z0-9_]` — letters, digits, underscore (one character):

```python
re.findall(r'\w+', "Hello, world! It's 2026.")
# → ['Hello', 'world', 'It', 's', '2026']
```

You've already used this! Our punctuation-removal pattern `[^\w\s]` = *"anything that's neither a word character nor whitespace"* = the punctuation.

**Why the underscore?** Because `\w` is designed to match **identifier** characters — things like variable names (`my_var2`). That's also why hyphens and apostrophes are *not* included (notice `It's` split into `It` and `s`).

---

### `\W` — a NON-Word Character

Punctuation, spaces, symbols — anything not in `[a-zA-Z0-9_]`:

```python
re.findall(r'\W', "Hi, there!")
# → [',', ' ', '!']
```

---

### The Pattern Behind the Pattern

| Lowercase | Matches | UPPERCASE | Matches |
|---|---|---|---|
| `\d` | digit `[0-9]` | `\D` | non-digit `[^0-9]` |
| `\s` | whitespace | `\S` | non-whitespace |
| `\w` | word char `[a-zA-Z0-9_]` | `\W` | non-word char |

**Memory hooks:** `\d` = **d**igit, `\s` = **s**pace, `\w` = **w**ord. **Capitalize to complement** — the uppercase twin is the lowercase set with a `[^...]` hat on it. One rule, three pairs.

---

### Putting It All Together — A Real Pattern

A Finnish phone number like `040-123 4567`:

```python
pattern = r'^\d{3}-\d{3}\s\d{4}$'
#           │ │    │ │    │  │   │
#           │ 3    │ 3    │  4   end
#         start  digits  digits digits
#              literal-  whitespace

re.search(pattern, '040-123 4567')    # ✓ match
re.search(pattern, '040-123-4567')    # → None (- where \s expected)
```

Every piece is something you now know: anchors `^ $`, counted repetition `{3}`, shorthands `\d \s`, and a literal `-`. Regex patterns are just these small LEGO bricks snapped together. 🎯

Using the above notation we can now shorten our previous variable name example to ```r’[a-zA-Z_]\w*\Z’```

````python
#[a-zA-Z_] = First character should be [a-z] or [A-Z] or _
#\w = Same as [a-zA-Z_]
#\w* = 0 or more of [a-zA-Z_]
#\Z = End of the string
# r'[a-zA-Z_]\w*\Z' == r'\w\w*\Z'
````

The patterns ```\A, \b, \B,``` and ```\Z``` will all match an empty string, but in specific places.

The patterns,
```\A``` and ```\Z``` will recognise the beginning and end of the string, respectively.

Note that the patterns ```^``` and ```$``` can in some cases match also after a newline and before a newline, correspondingly.

So, ```\A``` is distinct from ```^```, and ```\Z``` is distinct from ```$```.

The pattern ```\b``` matches at the start or end of a word. The pattern ```\B``` does the reverse.



## Zero-Width Anchors — Matching *Places*, Not Characters

First, the strange phrase *"match an empty string in specific places"*: these patterns don't consume any characters. They match **positions** — invisible spots *between* characters, like the start of the string or the boundary between a word and a space. Think of them as **checkpoints**, not characters.

```python
import re
```

---

### `\A` and `\Z` — Absolute Start and End

These match only the true beginning / true end of the **whole string**:

```python
re.search(r'\AHello', 'Hello world')    # ✓ Hello is at the absolute start
re.search(r'world\Z', 'Hello world')    # ✓ world is at the absolute end
```

*"But wait — isn't that what `^` and `$` do?"* Almost — and here's the subtle difference the text warns about.

---

### `^` vs `\A` — The Newline Gotcha

With the `re.MULTILINE` flag, `^` changes meaning: it matches at the start of **every line**, not just the string:

```python
text = "first line\nsecond line"

re.findall(r'^\w+', text, re.MULTILINE)
# → ['first', 'second']        ← ^ matched at BOTH line starts!

re.findall(r'\A\w+', text, re.MULTILINE)
# → ['first']                  ← \A only at the absolute string start
```

Similarly `$` can match before each newline in multiline mode, while `\Z` only matches the absolute end. (Even without the flag, `$` has a tiny quirk: it also matches just *before* a final trailing newline.)

**Rule of thumb:**

| | Normal use | Multiline text |
|---|---|---|
| `^` `$` | fine, same as `\A` `\Z` mostly | per-**line** anchors |
| `\A` `\Z` | strict | per-**string** anchors, always |

For everyday patterns `^...$` is what everyone writes; reach for `\A...\Z` when you must be strict about multiline input (e.g., validating user input that might contain sneaky newlines).

---

### `\b` — Word Boundary

`\b` matches the invisible position **where a word starts or ends** — technically, wherever a `\w` character meets a `\W` character (or the string edge). Its superpower: matching **whole words only**.

The problem it solves:

```python
re.findall(r'cat', 'cat catalog concatenate')
# → ['cat', 'cat', 'cat']        ← found cat INSIDE other words! 😱

re.findall(r'\bcat\b', 'cat catalog concatenate')
# → ['cat']                       ← only the standalone word ✓
```

Visualizing the checkpoints in `"cat catalog"`:

```
|cat| |catalog|
↑   ↑ ↑       ↑
\b  \b \b     \b       ← boundaries: word↔space and string edges
```

`\bcat\b` demands: boundary, then c-a-t, then boundary — so `catalog` fails (after `cat` comes `a`, a word character, no boundary there).

This is *the* tool for find-and-replace of whole words:

```python
re.sub(r'\bis\b', 'was', 'this is his island')
# → 'this was his island'        ← "this", "his", "island" untouched!
```

Without `\b`, you'd get the classic blunder `thwas was hwas wasland`. 😄

---

### `\B` — NOT a Word Boundary

The complement (capitalize-to-complement, same convention as `\D \S \W`!): matches only at positions **inside** a word or inside non-word runs:

```python
re.findall(r'\Bcat\B', 'cat catalog concatenate')
# → ['cat']            ← only the cat in con·cat·enate — buried on both sides!

re.findall(r'\Bcat', 'cat catalog concatenate')
# → ['cat']            ← cat NOT at a word start → again only concatenate's
```

Use case: finding a fragment *only* when embedded, e.g., a syllable inside words but not as a standalone word.

---

### Cheat Sheet

| Pattern | Matches at | Consumes chars? |
|---|---|---|
| `\A` | absolute start of string | no — position only |
| `\Z` | absolute end of string | no |
| `^` | start of string (or line, in multiline) | no |
| `$` | end of string (or line) | no |
| `\b` | word ↔ non-word transition | no |
| `\B` | anywhere that's NOT such a transition | no |

---

### The Mental Model

> Regular patterns like `\d` or `cat` are **character eaters** — each match consumes text.
> Anchors are **location inspectors** 📍 — they stand between characters and answer one question: *"is this the right kind of spot?"* — string edge (`\A \Z`), line edge (`^ $`), word edge (`\b`), or explicitly not-an-edge (`\B`).
>
> Combine eaters with inspectors and you get precise patterns: `\b\d{4}\b` = *"a four-digit number standing alone as its own word."* 🎯
