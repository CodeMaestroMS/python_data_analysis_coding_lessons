Attribute lookup

Suppose ```x``` is an instance of class ```X```, and we want to read an attribute ```x.a```. The lookup has three phases:

1. First it is checked whether the attribute ```a``` is an attribute of the instance ```x```
2. If not, then it is checked whether ```a``` is a class attribute of ```x```’s class ```X```
3. If not, then the base classes of ```X``` are checked

If instead we want to bind the attribute ```a```, things are much simpler.

```x.a``` = value will set the instance attribute.

And ```X.a``` = value will set the class attribute.

Note that if ```a``` base of ```X```, the class ```X```, and the instance ```x``` each have an attribute called ```a```,
then ```x.a``` hides ```X.a```, and ```X.a``` hides the attribute of the base class.

````python
class Prepend(object):

    def __init__(self,para1):
        self.start = para1

    def write(self, prepend):
        print(prepend+self.start)
````
