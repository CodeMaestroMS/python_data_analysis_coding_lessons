### ```__del__```

The method ```__del__``` gets called when the corresponding instance gets deleted

The opposite bookend to ```__init__```:

````python
class Resource:
    def __init__(self, name):
        self.name = name
        print(f"{self.name} created")
    def __del__(self):
        print(f"{self.name} destroyed")

r = Resource("file handle")
del r
````

Useful for cleanup — e.g., releasing a resource when nobody needs the object anymore.
