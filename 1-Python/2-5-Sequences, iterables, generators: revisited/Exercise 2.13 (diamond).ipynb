```python
def diamond(n):
    """
    Create a diamond-shaped pattern of 1s in a 2D numpy array of size (2n-1) x (2n-1).
    The diamond is formed by combining an identity matrix and its mirrored version.
    """
    I=np.eye(n, dtype='int')
    I_mirrored=np.fliplr(I)
    top_half=np.concatenate((I,I_mirrored[1:,:]))
    left_half=np.fliplr(top_half)
    full=np.concatenate( (left_half, top_half[:,1:]), axis=1 )
    print(full)
```

```python
def diamond(n):
    """Return a (2n-1)x(2n-1) integer array whose 1s form a diamond outline.

    Built from an n x n identity matrix (one quarter of the diamond):
    concatenate it with its vertical mirror to make the top-and-bottom half,
    then concatenate that with its horizontal mirror to complete the shape.
    Overlapping middle row/column are removed by slicing so they aren't doubled.
    """
    quarter = np.eye(n, dtype=int)                            # a diagonal — one arm of the diamond
    half = np.concatenate((quarter, np.fliplr(quarter)[1:, :]))   # stack vertically (drop shared row)
    return np.concatenate((np.fliplr(half)[:, :-1], half), axis=1)  # mirror left (drop shared col)
```
