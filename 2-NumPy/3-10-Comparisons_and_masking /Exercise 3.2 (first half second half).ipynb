```python
def first_half_second_half(a):
    r,c = a.shape
    a1=a[:,:int(c/2)]
    a2=a[:,int(c/2):]
    c =a1.sum(axis=1) > a2.sum(axis=1)
    return a[c]
```

```python
first_half_second_half(a)
```
