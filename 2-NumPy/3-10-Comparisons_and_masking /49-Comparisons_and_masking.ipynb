### Comparisons and masking

Just like NumPy allows element-wise arithmetic operations between arrays, for example addition of two arrays, it is also possible to compare two arrays element-wise. For example

````python
import numpy as np
````

````python
a=np.array([1,3,4])
b=np.array([2,2,7])
````

````python
c=a < b
print(c)
````

Now we can query whether all comparisons resulted `True`, or whether some comparison resulted `True`:

````python
print(c.all()) # were all True
print(c.any()) # was some comparison True
````

We can also count the number of comparisons that were True. This solution relies on the interpretation that True corresponds to 1 and False corresponds to 0:

````python
print(np.sum(c))
````

Because the broadcasting rules apply also to comparison, we can write

````python
print(a > 0)
````

To try these operations on real data, we download some weather statistics from Helsinki (Kumpula) weather station. More precisely, we will get the daily average temperatures from year 2017. We use the Pandas library, which we will cover later during this course, to load the data.

````python
import pandas as pd
import numpy as np

a=pd.read_csv("https://raw.githubusercontent.com/csmastersUH/data_analysis_with_python_2020/master/kumpula-weather-2017.csv")['Air temperature (degC)'].values
````

````python
a
````

````python
np.sum(a < 0)
````

````python
print("Number of days with the temperature below zero", np.sum(a < 0))
````

In core Python we can combine truth values using the `and, or, and not` keywords. For boolean array however we have to use the elementwise operators `&, |, and ~`, respectively. An example of these:

````python
print(np.sum((0 < a) & (a > 10))) # Tempeture is greater than 0 and less that 10
````

Note that we had to use parentheses around the comparisons, because the precedence of the `&` is higher than that of `<`, and so it would have been performed before the comparisons.

Another use of boolean arrays is that they can be used to select a subset of elements. For example

````python
c = a > 0
print(c)
````

````python
c.ndim
````

````python
print(c[:10]) # print only the first te elements
````

````python
print(a[c]) # select only the positive temperatures
````

This operation is called `masking`. It can also be used to assign a new value. For example the following zeroes out the negative temperatures:

````python
a[~c] = 0
print(a)
````

## Boolean Masking — Simply

This is a powerful NumPy technique called **boolean indexing** (or "masking"). It lets you **filter an array** — keeping only the elements that satisfy a condition — without any loop. Let's break down the two operations.

---

### The Setup

```python
a = ... ['Air temperature (degC)'].values     # a 1D array of temperatures
```

So `a` is an array of numbers like `[-2.1, 3.4, 0.0, 5.2, -1.8, ...]` — daily air temperatures, some positive, some negative.

---

### Part 1 — `c = a > 0` Creates a Boolean Array (the "mask")

```python
c = a > 0
```

This is a **vectorized comparison** — it checks `> 0` for **every element at once**, producing an array of `True`/`False` values:

```
a:  [-2.1,  3.4,  0.0,  5.2, -1.8, ...]
         │     │     │     │     │
       -2.1>0 3.4>0 0.0>0 5.2>0 -1.8>0
         │     │     │     │     │
c:  [False, True, False, True, False, ...]
```

**`c` has the same length as `a`**, but every element is a boolean: `True` where the temperature is positive, `False` where it isn't. This boolean array is called a **mask**.

```python
print(c[:10])     # shows the first 10 True/False values
# → [False  True False  True ... ]
```

---

### Part 2 — `a[c]` Uses the Mask to SELECT Elements

```python
print(a[c])
```

This is the clever part: indexing an array **with a boolean array** keeps **only the elements where the mask is `True`**, and drops the rest:

```
a:  [-2.1,  3.4,  0.0,  5.2, -1.8, ...]
c:  [False, True, False, True, False, ...]
        ✗     ✓     ✗     ✓     ✗
             keep        keep

a[c]:  [3.4, 5.2, ...]     ← only the elements where c was True!
```

So `a[c]` returns a **new array containing only the positive temperatures** — every element where the condition `a > 0` held true. Elements where `c` was `False` (negatives and zero) are excluded.

---

### The Two Steps Together

```python
c = a > 0        # STEP 1: build a True/False mask ("which ones are positive?")
a[c]              # STEP 2: use the mask to keep only the True positions
```

You can even combine them into one line:

```python
a[a > 0]          # same result — filter directly, no separate 'c' variable
# → array of only the positive temperatures
```

---

### Why This Is Powerful — No Loop Needed

Without boolean masking, you'd write a loop:

```python
# The slow, manual way:
positives = []
for temp in a:
    if temp > 0:
        positives.append(temp)
```

Boolean masking does the **exact same thing** in one vectorized expression:

```python
positives = a[a > 0]      # one line, no loop, much faster
```

Both give the positive temperatures — but the masking version is shorter and runs in fast compiled code (the vectorization benefit you learned earlier).

---

### Other Things You Can Do With Masks

```python
a[a > 0]          # positive temperatures
a[a < 0]          # negative temperatures (freezing days)
a[a == 0]          # exactly zero
a[(a > 0) & (a < 10)]   # BETWEEN 0 and 10 (& = AND, note the parentheses!)
```

You can also **count** or **compute** on the filtered results:

```python
np.sum(a > 0)      # → how MANY positive days (True counts as 1)
np.mean(a[a > 0])   # → average of just the positive temperatures
len(a[a > 0])        # → number of positive temperatures
```

---

### Step-by-Step on a Tiny Example

```python
a = np.array([-2, 5, 0, 8, -1])

c = a > 0
print(c)          # → [False  True False  True False]

print(a[c])        # → [5 8]     only elements where c is True
```

```
a:      [-2,    5,    0,    8,   -1]
c:      [False, True, False, True, False]
a[c]:         [5,          8      ]    ← kept only the True positions
```

---

### The One-Sentence Summary

> `c = a > 0` creates a **boolean mask** — an array of `True`/`False`, one per element, marking which temperatures are positive; then `a[c]` uses that mask to **select only the elements where `c` is `True`**, returning a new array of just the positive temperatures. This is **boolean indexing** — a fast, loop-free way to filter an array by a condition, equivalent to writing `a[a > 0]` in one step. 🎯

"Indexing an Array" — Simply

Indexing means accessing element(s) from an array by their position or by some selector — reaching into the array to pull out specific values. It's the general term for "getting stuff out of an array using the square brackets [ ]."

The Basic Idea — Square Brackets Reach Into the Array
python
a = np.array([10, 20, 30, 40, 50])

Whenever you use [ ] after an array to grab something from it, that's indexing:

python
a[0]        # → 10    grab the element at position 0
a[2]         # → 30    grab the element at position 2
