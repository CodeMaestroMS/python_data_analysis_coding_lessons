```python
from functools import reduce
from numpy.linalg import inv

def matrix_power(m,n):
    if n == 0:
        return np.eye(m.shape[0])
    else:
        if n < 0:
            m=inv(m)
            n=n*-1
        return reduce(lambda x,y: x@y, [m] * n)
```

```python
matrix_power(a,-2)
```
