````python
import numpy as np
````

There are several ways of creating NumPy arrays.

1. One way is to give a (nested) list as a parameter to the ```array``` constructor:

````python
np.array([1,2,3])
````

````python
print(np.array([1,2,3]).ndim)
````

## No — This Is a Plain (Non-Nested) List

```python
np.array([1, 2, 3])
```

Look at `[1, 2, 3]` carefully — it's just **one level** of brackets, containing **plain numbers** directly:

```
[1, 2, 3]
 │  │  │
 1  2  3      ← each element is just a NUMBER, not another list
```

No list appears **inside** another list here — so this is **not** nested. It's a single, flat list.

---

### Comparing Both Cases Side by Side

```python
np.array([1, 2, 3])              # flat list → 1D array
np.array([[1, 2, 3], [4, 5, 6]])   # NESTED list → 2D array
```

| Code | Structure | Levels of brackets | Result |
|---|---|---|---|
| `[1, 2, 3]` | flat | one `[ ]` | 1D array — a single row |
| `[[1,2,3], [4,5,6]]` | nested | `[ [ ] , [ ] ]` — a list containing lists | 2D array — rows and columns |

---

### Verifying With NumPy

```python
import numpy as np

a = np.array([1, 2, 3])
print(a.ndim)     # → 1     (1-dimensional — confirms: NOT nested)

b = np.array([[1, 2, 3], [4, 5, 6]])
print(b.ndim)      # → 2     (2-dimensional — confirms: nested list → 2D array)
```

`.ndim` (number of dimensions) directly reflects how many levels of nesting the original list had.

---

### The Simple Test

> **Ask: does any element of the list contain ANOTHER list?**
>
> `[1, 2, 3]` → elements are `1`, `2`, `3` — plain numbers → **NOT nested** → 1D array
> `[[1,2,3], [4,5,6]]` → elements are `[1,2,3]` and `[4,5,6]` — themselves lists → **nested** → 2D array

---

### The One-Sentence Summary

> No — `[1, 2, 3]` is a **flat**, non-nested list; its elements are plain numbers, not other lists. Nesting only happens when a list's elements are **themselves lists**, like `[[1,2,3], [4,5,6]]` — which is exactly why the first example is labeled a "one dimensional array" and would need actual nesting to become 2D. 🎯
