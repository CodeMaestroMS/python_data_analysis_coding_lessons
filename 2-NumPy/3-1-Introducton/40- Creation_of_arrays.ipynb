### Creation of arrays

We start by importing the NumPy library, and we use the standard abbreviation ```np``` for it.

````python
import numpy as np
````

There are several ways of creating NumPy arrays.

1. One way is to give a (nested) list as a parameter to the ```array``` constructor:

````python
np.array([1,2,3])
````

````python
print(np.array([1,2,3]).ndim)
````

## No — This Is a Plain (Non-Nested) List

```python
np.array([1, 2, 3])
```

Look at `[1, 2, 3]` carefully — it's just **one level** of brackets, containing **plain numbers** directly:

```
[1, 2, 3]
 │  │  │
 1  2  3      ← each element is just a NUMBER, not another list
```

No list appears **inside** another list here — so this is **not** nested. It's a single, flat list.

---

### Comparing Both Cases Side by Side

```python
np.array([1, 2, 3])              # flat list → 1D array
np.array([[1, 2, 3], [4, 5, 6]])   # NESTED list → 2D array
```

| Code | Structure | Levels of brackets | Result |
|---|---|---|---|
| `[1, 2, 3]` | flat | one `[ ]` | 1D array — a single row |
| `[[1,2,3], [4,5,6]]` | nested | `[ [ ] , [ ] ]` — a list containing lists | 2D array — rows and columns |

---

### Verifying With NumPy

```python
import numpy as np

a = np.array([1, 2, 3])
print(a.ndim)     # → 1     (1-dimensional — confirms: NOT nested)

b = np.array([[1, 2, 3], [4, 5, 6]])
print(b.ndim)      # → 2     (2-dimensional — confirms: nested list → 2D array)
```

`.ndim` (number of dimensions) directly reflects how many levels of nesting the original list had.

---

### The Simple Test

> **Ask: does any element of the list contain ANOTHER list?**
>
> `[1, 2, 3]` → elements are `1`, `2`, `3` — plain numbers → **NOT nested** → 1D array
> `[[1,2,3], [4,5,6]]` → elements are `[1,2,3]` and `[4,5,6]` — themselves lists → **nested** → 2D array

---

### The One-Sentence Summary

> No — `[1, 2, 3]` is a **flat**, non-nested list; its elements are plain numbers, not other lists. Nesting only happens when a list's elements are **themselves lists**, like `[[1,2,3], [4,5,6]]` — which is exactly why the first example is labeled a "one dimensional array" and would need actual nesting to become 2D. 🎯

Note that leaving out the brackets from the above expression, i.e. calling ```np.array(1,2,3)``` will result in an error.

````python
np.array(1,2,3)
````

Two dimensional array can be given by listing the rows of the array:

````python
np.array([
    [1,2,3],
    [4,5,6]
])
````

Similarly, three dimensional array can be described as a list of lists of lists:

````python
t_d = np.array([
 [[1,2],[3,4]],
 [[5,6],[7,8]]
])
````

````python
t_d.ndim
````

There are some helper functions to create common types of arrays:

````python
np.zeros((3,4))
````

To specify that elements are ints instead of floats, use the parameter ```dtype```:

````python
np.zeros((3,4), dtype=int)
````

Similarly ones initializes all elements to one, full initializes all elements to a specified value, and empty leaves the elements uninitialized:

````python
np.ones((2,3), dtype=int)
````

````python
np.full((2,3), 7)
````

````python
np.empty((2,4), dtype=int)
````

The eye function creates the identity matrix, that is, a matrix with elements on the diagonal are set to one, and non-diagonal elements are set to zero:

````python
np.eye(5,dtype=int)
````

The ```arange``` function works like the range function, but produces an array instead of a list.

````python
np.arange(0,10,2)
````

For non-integer ranges it is better to use ```linspace```:

````python
np.linspace(0,np.pi,5)
````

With linspace one does not have to compute the length of the step, but instead one specifies the wanted number of elements. By default, the endpoint is included in the result, unlike with arange.
