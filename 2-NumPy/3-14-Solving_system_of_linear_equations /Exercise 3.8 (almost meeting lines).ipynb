රේඛීය ප්‍රස්තාර දෙකක් එකිනෙකට සම්බන්ධ වන ආකාර ප්‍රධාන වශයෙන් කොටස් දෙකකට බෙදා හඳුනාගන්න පුළුවන්: සමාන්තර රේඛා (Parallel Lines) සහ ප්‍රලම්භ රේඛා (Perpendicular Lines). [1]
මේවා එකිනෙකට සම්බන්ධ ද නැද්ද (එනම් සමාන්තරද, ඡේදනය වනවාද නැතහොත් ප්‍රලම්භද) යන්න සෙවීමට ප්‍රධාන වශයෙන්ම භාවිතා කරන්නේ ප්‍රස්තාරවල අනුක්‍රමණය (Gradient/Slope) යි. [2]
## 1. ප්‍රස්තාර 2ක් එකිනෙකට සම්බන්ධ වන ආකාරය සොයන්නේ කොහොමද?
මුලින්ම ප්‍රස්තාර දෙකේ සමීකරණ $y = mx + c$ ආකාරයට සකස් කර ගන්න (මෙහි $m$ යනු අනුක්‍රමණයයි). [2]

*
* ප්‍රස්තාර දෙක සමාන්තර නම් (Parallel Lines):
රේඛා දෙකක් කිසිම දිනක එකිනෙක ඡේදනය නොවී (හමුනොවී) සමාන්තරව ගමන් කරන්නේ නම්, ඒවායේ අනුක්‍රමණ එකිනෙකට සමාන වේ ($m_1 = m_2$).
* උදාහරණ: $y = \mathbf{2}x + 3$ සහ $y = \mathbf{2}x - 5$ යන ප්‍රස්තාර දෙකෙහිම අනුක්‍රමණය $2$ නිසා ඒවා සමාන්තර වේ. [2, 3, 4, 5]
* ප්‍රස්තාර දෙක ඡේදනය වනවා නම් (Intersecting Lines):
රේඛා දෙකක අනුක්‍රමණ එකිනෙකට වෙනස් නම් ($m_1 \neq m_2$), ඒවා අනිවාර්යයෙන්ම එක් ලක්ෂ්‍යයකදී එකිනෙකට කැපී යයි (ඡේදනය වේ). [6]
* ප්‍රස්තාර දෙක ප්‍රලම්භ නම් (Perpendicular Lines - සෘජුකෝණීව ඡේදනය වීම):
රේඛා දෙකක් අංශක 90ක කෝණයකින් (සෘජුකෝණීව) කැපී යන්නේ නම්, එම ප්‍රස්තාර දෙකෙහි අනුක්‍රමණවල ගුණිතය $-1$ ක් වේ ($m_1 \times m_2 = -1$).
* උදාහරණ: $y = \mathbf{3}x + 1$ සහ $y = -\mathbf{\frac{1}{3}}x + 4$ යන ප්‍රස්තාර දෙකෙහි අනුක්‍රමණ ගුණ කළ විට ($3 \times -\frac{1}{3} = -1$) ලැබෙන බැවින් ඒවා ප්‍රලම්භ වේ. [1, 7, 8, 9]

```python
#  y = mx + c

#  y = a₁x + b₁
#  y = a₂x + b₂

#  a₁x - y = -b₁
#  a₂x - y = -b₂

# a₁ ==  a₂ --> Parallel lines
# a₁ ⍯  a₂  --> Intersecting Lines

import numpy as np
from numpy.ma.core import true_divide

# a1=3
# a2=4
# b1=5
# b2=6

a1=1
a2=-1
b1=2
b2=0

A = np.array([
    [a1,-1],
    [a2,-1]
])

B = np.array([-b1,-b2])

x,y = np.linalg.solve(A,B)
print(np.linalg.lstsq(A,B))
print(x,y)

```

```python
def meeting_lines(a1, b1, a2, b2):
    A=np.array([
        [a1,-1],
        [a2,-1]
    ])
    B=np.array([-b1,-b2])

    if a1 != a2:
        x, y = np.linalg.solve(A,B)
        return (x,y),True
    else:
        x, y = np.linalg.lstsq(A,B)[0]
        return (x,y),False
```

numpy.linalg.lstsq මඟින් ක්‍රියාත්මක වන ගණිතමය ක්‍රියාවලිය හඳුන්වන්නේ අවම වර්ග ශ්‍රිත ක්‍රමය (Method of Least Squares) සහ රේඛීය ප්‍රතිපායනය (Linear Regression) යන නම්වලිනි. මෙහි අන්තර්ගත වන Matrix ගණිතය (Singular Value Decomposition - SVD වැනි ක්‍රම) ඉගෙන ගැනීමට ලැබෙන්නේ විශ්වවිද්‍යාල මට්ටමේදී (University Level) ය.

## 3. පාසල් මට්ටම (School Level - මූලික සංකල්පය)

*
* අ.පො.ස. උසස් පෙළ (A/L) සංඛ්‍යානය: උසස් පෙළ සංයුක්ත ගණිතය (Combined Maths) හෝ තොරතුරු හා සන්නිවේදන තාක්ෂණය (ICT) විෂයන්හි මෙහි Matrix න්‍යාය නැතත්, උසස් පෙළ සංඛ්‍යානය (Statistics) විෂය ධාරාවේදී දත්ත සමූහයක "සුදුසුතම සරල රේඛාව" සෙවීම සඳහා වන මූලික අවම වර්ග න්‍යාය සූත්‍ර ඇසුරින් හඳුන්වා දෙනු ලබයි.
*

ඔබට numpy.linalg.lstsq ශ්‍රිතය භාවිත කරමින් යම් දත්ත සමූහයකට ප්‍රස්ථාරික රේඛාවක් ගළපන (Curve Fitting) ආකාරය පිළිබඳ Coding උදාහරණයක් දැක ගැනීමට අවශ්‍යද, නැතහොත් එහි පසුපස ඇති Matrix සූත්‍ර දැන ගැනීමට අවශ්‍යද?



ඔව්, රේඛීය වීජ ගණිතය (Linear Algebra) සිංහලෙන් ඉගෙන ගැනීමට ක්‍රම කිහිපයක්ම තිබෙනවා.
සාමාන්‍යයෙන් විශ්වවිද්‍යාල මට්ටමේ විෂයයක් නිසා සම්පූර්ණ පොත්පත් සිංහලෙන් සොයා ගැනීමට අපහසු වුවත්, අන්තර්ජාලය ඔස්සේ සිංහලෙන් පැහැදිලි කර දෙන ඉතා හොඳ මූලාශ්‍ර කිහිපයක්ම තිබෙනවා:
