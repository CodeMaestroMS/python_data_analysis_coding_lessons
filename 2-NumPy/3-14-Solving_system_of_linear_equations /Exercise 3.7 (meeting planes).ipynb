```
z = a₁y + b₁x + c₁
z = a₂y + b₂x + c₂
z = a₃y + b₃x + c₃

-c₁ = a₁y + b₁x - z
-c₂ = a₂y + b₂x - z
-c₃ = a₃y + b₃x - z

```

````python
import numpy as np

def meeting_planes(a1, b1, c1,  a2, b2, c2,  a3, b3, c3):
   A= np.array([
        [a1,b1,-1],
        [a2,b2,-1],
        [a3,b3,-1]
    ])
   B= np.array([-c1, -c2, -c3])
   return np.linalg.solve(A,B)

````

````python
a1=1
b1=4
c1=5
a2=3
b2=2
c2=1
a3=2
b3=4
c3=1

x, y, z = meeting_planes(a1, b1, c1, a2, b2, c2, a3, b3, c3)
print(f"Planes meet at x={x}, y={y} and z={z}")
````
