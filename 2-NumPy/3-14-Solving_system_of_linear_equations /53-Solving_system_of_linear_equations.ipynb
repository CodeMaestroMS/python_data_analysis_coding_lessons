## Solving Systems of Linear Equations — Simply

This is about using `np.linalg.solve` to find the **unknown values** (`x`) in a set of equations, by writing the whole system as a single matrix equation `A · x = b`. Let me build it up with a concrete example.

---

### First — What's a "System of Linear Equations"?

It's a set of equations that share the same unknowns, which you solve **together**. A small example with two unknowns `x₁` and `x₂`:

$$2x_1 + 3x_2 = 8$$
$$1x_1 + 2x_2 = 5$$

You want to find the `x₁` and `x₂` that make **both** equations true at once.

---

### The Three Pieces — A, x, b

The passage splits every such system into three parts:

**`A` — the coefficient matrix** (the numbers multiplying the unknowns):
$$A = \begin{pmatrix} 2 & 3 \\ 1 & 2 \end{pmatrix}$$
(The `2, 3` from the first equation; `1, 2` from the second.)

**`x` — the unknowns** (what you're solving for):
$$x = \begin{pmatrix} x_1 \\ x_2 \end{pmatrix}$$

**`b` — the constants** (the right-hand side):
$$b = \begin{pmatrix} 8 \\ 5 \end{pmatrix}$$

---

### Why `A · x = b` Represents the Whole System

Matrix multiplication `A · x` **reconstructs** the left sides of your equations:

$$\begin{pmatrix} 2 & 3 \\ 1 & 2 \end{pmatrix} \begin{pmatrix} x_1 \\ x_2 \end{pmatrix} = \begin{pmatrix} 2x_1 + 3x_2 \\ 1x_1 + 2x_2 \end{pmatrix}$$

Setting that equal to `b`:

$$\begin{pmatrix} 2x_1 + 3x_2 \\ x_1 + 2x_2 \end{pmatrix} = \begin{pmatrix} 8 \\ 5 \end{pmatrix}$$

Reading row by row, this says exactly:
- `2x₁ + 3x₂ = 8` ✓
- `x₁ + 2x₂ = 5` ✓

**So `A · x = b` is just a compact way to write the entire system** — matrix multiplication unpacks it back into the individual equations.

---

### What the Shapes Mean

The passage gives shapes — here's what they encode:

| Symbol | Shape | Meaning |
|---|---|---|
| `A` | `(n, m)` | `n` equations, `m` unknowns (coefficients) |
| `x` | `(m, 1)` | `m` unknowns to find (a column) |
| `b` | `(n, 1)` | `n` constants, one per equation (a column) |

For our example: 2 equations, 2 unknowns → `A` is `(2,2)`, `x` is `(2,1)`, `b` is `(2,1)`.

---

### Solving It With `np.linalg.solve`

Instead of solving by hand (substitution/elimination), you hand `A` and `b` to NumPy:

```python
import numpy as np

A = np.array([[2, 3],
              [1, 2]])
b = np.array([8, 5])

x = np.linalg.solve(A, b)
print(x)      # → [1. 2.]   meaning x₁ = 1, x₂ = 2
```

**You give it `A` and `b`; it returns `x`** — the values of the unknowns.

---

### Verifying the Answer

Plug `x₁ = 1, x₂ = 2` back in:
- `2(1) + 3(2) = 2 + 6 = 8` ✓
- `1(1) + 2(2) = 1 + 4 = 5` ✓

Both equations satisfied — NumPy found the correct solution.

---

### The Intuition — Why It's Like Division

For a single equation, `a · x = b` is solved by `x = b / a`. The matrix version is the same idea:

$$A \cdot x = b \quad\longrightarrow\quad x = A^{-1} \cdot b$$

`np.linalg.solve` effectively does this "matrix division" for you (more efficiently and accurately than literally computing the inverse). You're asking: *"what `x`, when multiplied by `A`, gives `b`?"* — the matrix analog of simple division.

---

### Why Use This Instead of Solving By Hand?

For 2 equations, hand-solving is easy. But real systems can have **hundreds or thousands** of equations and unknowns — far too many to solve manually. `np.linalg.solve` handles any size in one call:

```python
A = np.array([[3, 2, -1],
              [2, -2, 4],
              [-1, 0.5, -1]])
b = np.array([1, -2, 0])

x = np.linalg.solve(A, b)      # solves 3 equations, 3 unknowns instantly
```

---

### The One-Sentence Summary

> A system of linear equations (several equations sharing the same unknowns) can be written compactly as the matrix equation **`A · x = b`**, where `A` holds the coefficients (shape `(n,m)` = n equations × m unknowns), `x` is the column of unknowns to find (shape `(m,1)`), and `b` is the column of constants (shape `(n,1)`); matrix multiplication `A · x` unpacks back into the individual equations. **`np.linalg.solve(A, b)` returns `x`** — the values of the unknowns — doing the matrix equivalent of "dividing `b` by `A`," which is far more practical than solving by hand for large systems. 🎯
