```python
import numpy as np

a1=3
a2=4
b1=5
b2=6

# y = a₁x + b₁
# y = a₂x + b₂

# y = 3x+5
# y = 4x+6

# 3x-y = -5
# 4x-y = -6

A=np.array([
    [a1,-1],
    [a2,-1]
])

B = np.array([-5,-6])
x=np.linalg.solve(A,B)
print(x)
```

```python
def meeting_lines(a1, b1, a2, b2):
    A=np.array([
        [a1,-1],
        [a2,-1]
    ])
    B=np.array([-b1,-b2])

    x=np.linalg.solve(A,B)

    print(x)
```

```python
meeting_lines(a1, b1, a2, b2)
```
