## `scipy.linalg.norm` — What It Is (and Why the Exercise Bans It)

`scipy.linalg.norm` is a **ready-made function** that computes the norm (length/magnitude) of a vector or matrix in **one call** — which is exactly why your exercise forbids it: the whole point of the exercise is for **you** to build that calculation yourself from `np.sum` and `np.sqrt`, rather than call a function that does it all for you.

---

### What It Does

```python
from scipy.linalg import norm

v = [3, 4]
norm(v)      # → 5.0     computes √(3² + 4²) in one call
```

It's essentially the **complete `vector_lengths` calculation, pre-packaged**. Instead of writing `np.sqrt(np.sum(v**2))` yourself, `norm(v)` does all of it internally.

---

### Why the Exercise Bans It

The exercise says: *"You must use at least `np.sum` and `np.sqrt`. You can't use `scipy.linalg.norm`."*

The reasoning is **pedagogical** — if you were allowed to use `norm`, the entire exercise would collapse to:

```python
def vector_lengths(a):
    return scipy.linalg.norm(a, axis=1)     # ← would work, but defeats the purpose!
```

You'd learn nothing about **how** a norm is actually computed. By forcing you to use `np.sum` and `np.sqrt`, the exercise makes you **build the Euclidean norm from its definition** — understanding that "length = √(sum of squares)" — rather than treating it as a black box.

```python
# What the exercise WANTS you to write (understanding the formula):

# What it FORBIDS (black-box shortcut):

```

### `scipy` vs `numpy` — Two Different Libraries

Worth clarifying, since the names look similar:

- **NumPy** (`numpy`) — the foundational array library (what you've been learning)
- **SciPy** (`scipy`) — built **on top of** NumPy, adds advanced scientific computing (linear algebra, optimization, statistics, signal processing)

`scipy.linalg` is SciPy's **linear algebra** module, and `norm` is one of its functions. NumPy *also* has its own version, `numpy.linalg.norm` — very similar. The exercise specifically bans the SciPy one (and the spirit clearly bans `numpy.linalg.norm` too, since both are the black-box shortcut you're meant to avoid).

---

### What "Norm" Means Mathematically

A **norm** is a general concept for "the size/length of a vector." The **Euclidean norm** (the one your exercise uses) is the most common — the everyday straight-line distance:

$$\|v\|_2 = \sqrt{v_1^2 + v_2^2 + \cdots + v_n^2}$$

But there are **other** norms too, and `scipy.linalg.norm` can compute several via its `ord` parameter:

```python
from scipy.linalg import norm

v = [3, -4]
norm(v)           # → 5.0    Euclidean (L2) norm — √(9+16), the default
norm(v, 1)         # → 7.0    L1 norm (Manhattan) — |3| + |-4|
norm(v, np.inf)     # → 4.0    max norm — largest absolute value
```

Your exercise wants specifically the **Euclidean (L2)** norm — the default — which is why √(sum of squares) is the formula you build.
