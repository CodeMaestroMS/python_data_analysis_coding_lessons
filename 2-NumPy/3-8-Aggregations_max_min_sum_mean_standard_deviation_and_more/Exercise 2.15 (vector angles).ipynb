## The Angle Between Two Vectors — Simply

This is Exercise 2.15 (`vector_angles`). It gives you a formula to compute the **angle between two vectors** using their inner product and their lengths. Let's decode every symbol.

---

### The Two Formulas

**Formula 1 — the cosine of the angle:**

$$\cos_{xy}(\alpha) = \frac{\langle x, y \rangle}{\|x\| \, \|y\|}$$

**Formula 2 — the length (norm), defined via inner product:**

$$\|x\| = \sqrt{\langle x, x \rangle}$$

Let me explain each piece.

---

### Symbol 1 — `⟨x, y⟩` (Angle Brackets) = Inner Product = Dot Product

The angle brackets $\langle x, y \rangle$ mean the **inner product** (also called the **dot product**) of two vectors. It's computed by **multiplying matching components and adding them up**:

$$\langle x, y \rangle = x_1 y_1 + x_2 y_2 + \cdots + x_m y_m$$

```python
x = [1, 2, 3]
y = [4, 5, 6]

# inner product = 1*4 + 2*5 + 3*6 = 4 + 10 + 18 = 32
np.dot(x, y)          # → 32
np.sum(np.array(x) * np.array(y))    # → 32   (same thing, built from pieces)
```

**One number** comes out — a single value summarizing how much the two vectors "align."

---

### Symbol 2 — `‖x‖` (Double Bars) = the Length (Norm) of a Vector

Those double bars $\|x\|$ mean the **length/magnitude** of vector `x` — exactly the Euclidean norm from your `vector_lengths` exercise!

**Formula 2** tells you how to compute it using the inner product:

$$\|x\| = \sqrt{\langle x, x \rangle}$$

This says: *"the length of x = the square root of x's inner product with itself."* Let's verify it's the same √(sum of squares) you already know:

$$\langle x, x \rangle = x_1 x_1 + x_2 x_2 + \cdots = x_1^2 + x_2^2 + \cdots$$

$$\|x\| = \sqrt{x_1^2 + x_2^2 + \cdots}$$

**It IS the Euclidean norm** — just written using inner-product notation. `⟨x, x⟩` (a vector dotted with itself) = sum of squares, and the square root of that = the length.

```python
x = [3, 4]
np.sqrt(np.dot(x, x))     # → 5.0    same as √(9+16) = √25
```

---

### Symbol 3 — `:=` Means "Is Defined As"

The `:=` (colon-equals) means **"is defined to be"** — it's introducing a definition, not stating an equation to solve. So:

- $\cos_{xy}(\alpha) := \ldots$ means *"we DEFINE the cosine of the angle to be..."*
- $\|x\| := \sqrt{\langle x, x \rangle}$ means *"we DEFINE the length to be..."*

---

### Putting Formula 1 Together — The Whole Idea

$$\cos(\alpha) = \frac{\langle x, y \rangle}{\|x\| \, \|y\|} = \frac{\text{inner product of x and y}}{\text{length of x} \times \text{length of y}}$$

In words: *"the cosine of the angle between two vectors = their dot product, divided by the product of their lengths."*

This gives you $\cos(\alpha)$ — and to get the actual **angle** $\alpha$, you take the inverse cosine (arccos):

$$\alpha = \arccos\left(\frac{\langle x, y \rangle}{\|x\| \, \|y\|}\right)$$

---

### Why This Formula Works — Intuition

The dot product measures how much two vectors "point the same way":

```
Same direction    → cos(α) = 1   → α = 0°     (angle between them is zero)
Perpendicular     → cos(α) = 0   → α = 90°
Opposite direction → cos(α) = -1  → α = 180°
```

Dividing by the lengths (`‖x‖‖y‖`) **removes the effect of the vectors' sizes**, leaving only the **direction** relationship — so you get a pure angle, regardless of how long the vectors are.

---

### Coding It (For the Exercise)

The exercise wants `vector_angles(X, Y)` — using `np.arccos`, and the requirements likely mirror `vector_lengths` (build the norm from `np.sqrt`/`np.sum`, no `scipy`). The core computation:

```python
import numpy as np

def vector_angle(x, y):
    """Angle (in radians) between two vectors x and y."""
    inner = np.sum(x * y)                    # ⟨x, y⟩ — the dot product
    len_x = np.sqrt(np.sum(x**2))             # ‖x‖ — length of x
    len_y = np.sqrt(np.sum(y**2))              # ‖y‖ — length of y
    cos_alpha = inner / (len_x * len_y)         # the formula
    return np.arccos(cos_alpha)                  # α = arccos(...)
```

**Watch out for the numerical caveat the exercise mentions:** floating-point rounding can push `cos_alpha` slightly above 1 or below -1 (e.g., `1.0000000002`), which makes `arccos` return `nan`. The exercise hints to guard against this with `np.clip(cos_alpha, -1, 1)` to force it back into the valid `[-1, 1]` range before calling `arccos`.

---

### Symbol Reference Table

| Symbol | Name | Meaning | Computed as |
|---|---|---|---|
| `⟨x, y⟩` | inner/dot product | how much x and y align | `np.sum(x*y)` |
| `‖x‖` | norm / length | magnitude of x | `np.sqrt(np.sum(x**2))` |
| `:=` | "defined as" | introduces a definition | — |
| `α` | alpha | the angle between the vectors | `np.arccos(...)` |

---

### The One-Sentence Summary

> This defines the **angle between two vectors**: $\cos(\alpha) = \frac{\langle x,y\rangle}{\|x\|\|y\|}$ — the **inner product** (dot product, `np.sum(x*y)`) divided by the product of the two vectors' **lengths** (`‖x‖ = √⟨x,x⟩`, the same Euclidean norm from your last exercise). Dividing by the lengths strips away the vectors' sizes, leaving only their directional relationship, so `arccos` of that ratio gives the actual angle — just remember to `np.clip` the cosine into `[-1, 1]` first to avoid floating-point errors making `arccos` return `nan`. 🎯

## Degrees vs Radians — Two Ways to Measure Angles

This is about the **units** used to measure an angle. Just like distance can be measured in meters *or* feet, angles can be measured in **degrees** *or* **radians** — same angle, different number.

---

### The Two Units

**Degrees** — the everyday unit you already know:
```
Full circle    = 360°
Half circle    = 180°
Right angle    = 90°
```

**Radians** — the "math/physics" unit, based on π:
```
Full circle    = 2π   ≈ 6.283 radians
Half circle    = π    ≈ 3.14159 radians
Right angle    = π/2  ≈ 1.5708 radians
```

**Same angles, different numbers.** A right angle is `90` in degrees but `1.5708` in radians — it's the *same corner*, just measured on a different scale.

---

### The Conversion

| Degrees | Radians |
|---|---|
| 0° | 0 |
| 90° | π/2 ≈ 1.571 |
| 180° | π ≈ 3.142 |
| 360° | 2π ≈ 6.283 |

$$\text{radians} = \text{degrees} \times \frac{\pi}{180} \qquad\qquad \text{degrees} = \text{radians} \times \frac{180}{\pi}$$

---

### Why This Matters for Your Exercise

**NumPy's `np.arccos` returns the angle in RADIANS by default.** So when you compute:

```python
angle = np.arccos(cos_alpha)      # → this is in RADIANS
```

...you get something like `1.5708`, not `90`. But the exercise wants the answer **in degrees** (`90`), which is more human-readable. So you need to **convert radians → degrees**.

---

### How to Convert in NumPy

Two easy ways:

**Option 1 — `np.degrees()` (cleanest):**
```python
angle_radians = np.arccos(cos_alpha)
angle_degrees = np.degrees(angle_radians)      # radians → degrees
```

**Option 2 — Do it in one line:**
```python
angle_degrees = np.degrees(np.arccos(cos_alpha))
```

**Option 3 — Manual formula (if you prefer):**
```python
angle_degrees = np.arccos(cos_alpha) * 180 / np.pi
```

`np.degrees()` is the built-in for exactly this conversion — it's the recommended, clearest choice.

---

### Seeing the Difference

```python
import numpy as np

x = np.array([4, 0])
y = np.array([0, 3])     # perpendicular vectors

cos_alpha = np.sum(x*y) / (np.sqrt(np.sum(x**2)) * np.sqrt(np.sum(y**2)))
# → 0.0  (perpendicular)

radians = np.arccos(cos_alpha)
print(radians)                 # → 1.5707963...   (π/2, in RADIANS)

degrees = np.degrees(radians)
print(degrees)                  # → 90.0           (much more readable!)
```

Both `1.5708` and `90.0` describe the **same right angle** — but `90.0` is the degrees form the exercise asks for.

---

### Applying It to `vector_angles`

```python
import numpy as np

def vector_angles(X, Y):
    """Angle between each pair of row-vectors in X and Y, in DEGREES."""
    inner = np.sum(X * Y, axis=1)
    len_X = np.sqrt(np.sum(X**2, axis=1))
    len_Y = np.sqrt(np.sum(Y**2, axis=1))
    cos_alpha = inner / (len_X * len_Y)
    cos_alpha = np.clip(cos_alpha, -1, 1)          # guard against float errors → nan
    return np.degrees(np.arccos(cos_alpha))          # ← radians → DEGREES
```

The `np.degrees(...)` wrapping the `np.arccos(...)` is the piece that satisfies *"in degrees, not in radians."*

---

### Why Two Units Even Exist

Quick context:
- **Degrees** — older, human-friendly (360 divides nicely into halves, quarters, etc.), used in everyday life, navigation, geometry class.
- **Radians** — the "natural" mathematical unit (based on the circle's own geometry), used in calculus, physics, and — importantly — as the **default** in nearly all programming math functions (`np.sin`, `np.cos`, `np.arccos` all work in radians).

Because code defaults to radians, you often have to **explicitly convert to degrees** when you want a human-readable answer — which is exactly what your exercise is asking you to do.

---

### The One-Sentence Summary

> "In degrees, not in radians" means the angle should be expressed on the **0–360° scale** you know from everyday life (a right angle = 90°), rather than the **π-based radian scale** (a right angle = π/2 ≈ 1.571) that NumPy's `np.arccos` returns by default. Since `arccos` gives radians, you convert to degrees by wrapping it in **`np.degrees(...)`** — same angle, just the more human-readable unit the exercise wants. 🎯

## Yes — `np.dot()` Is the Main One (Plus a Few Alternatives)

The inner/dot product $\langle x, y \rangle$ has a dedicated NumPy function: **`np.dot()`**. But there are a few ways to compute it, and which fits best depends on whether you're doing single vectors or many at once.

---

### 1. `np.dot()` — The Standard Function

```python
import numpy as np

x = np.array([1, 2, 3])
y = np.array([4, 5, 6])

np.dot(x, y)      # → 32     (1·4 + 2·5 + 3·6 = 4+10+18)
```

This is the direct built-in for the inner product of two 1D vectors.

---

### 2. The `@` Operator — Same Thing, Cleaner Syntax

Python has a dedicated matrix/dot-product operator, `@`:

```python
x @ y      # → 32     same as np.dot(x, y)
```

`@` is the operator form (it calls `__matmul__` under the hood — same "operator is a method" pattern you've seen). For two 1D vectors, `x @ y` and `np.dot(x, y)` are identical.

---

### 3. `np.sum(x * y)` — Building It From Pieces

This is what your **exercise likely wants**, since it restricts you to `np.sum` and `np.sqrt` (and bans shortcut functions):

```python
np.sum(x * y)      # → 32
```

Breaking it down:
```python
x * y              # → [4, 10, 18]    element-wise multiply (vectorized)
np.sum([4,10,18])   # → 32             add them up
```

This **is** the dot product, built from its definition: multiply matching components, then sum. If the exercise wants you to avoid black-box functions (like it did with `norm`), this is the form to use.

---

### 4. `np.inner()` — Literally Named "Inner Product"

```python
np.inner(x, y)      # → 32
```

For 1D vectors, `np.inner` gives the same result as `np.dot` — it's named directly after the mathematical term.

---

### The Important Case — Many Vectors at Once (Row-wise)

If `X` and `Y` are 2D (each row a vector, as in `vector_angles`), you want the dot product **per row** — and here `np.dot` does NOT do what you want:

```python
X = np.array([[1, 2, 3],
              [4, 5, 6]])
Y = np.array([[1, 0, 0],
              [0, 1, 0]])

# ✗ np.dot(X, Y) does MATRIX multiplication — not row-wise dot products!

# ✓ This gives one dot product PER ROW:
np.sum(X * Y, axis=1)     # → [1, 5]
#                              │  │
#                              │  └── row 1: 4·0 + 5·1 + 6·0 = 5
#                              └───── row 0: 1·1 + 2·0 + 3·0 = 1
```

**`np.sum(X * Y, axis=1)`** is the row-wise inner product — the `axis=1` collapses each row into one dot-product value, exactly matching the `(n, m) → (n,)` pattern you need for `vector_angles`.

---

### Comparison Table

| Method | For 1D vectors | For row-wise (2D) | Notes |
|---|---|---|---|
| `np.dot(x, y)` | ✓ `32` | ✗ does matrix mult | classic function |
| `x @ y` | ✓ `32` | ✗ does matrix mult | operator form |
| `np.inner(x, y)` | ✓ `32` | — | named "inner" |
| `np.sum(x*y)` | ✓ `32` | — | built from definition |
| `np.sum(X*Y, axis=1)` | — | ✓ one per row | **row-wise dot product** |

---

### Which to Use in `vector_angles`

Since the exercise style restricts you to `np.sum`/`np.sqrt` (mirroring the `vector_lengths` rules), and you need **per-row** dot products:

```python
def vector_angles(X, Y):
    inner = np.sum(X * Y, axis=1)              # ⟨x, y⟩ for each row
    len_X = np.sqrt(np.sum(X**2, axis=1))       # ‖x‖ per row
    len_Y = np.sqrt(np.sum(Y**2, axis=1))        # ‖y‖ per row
    cos_alpha = np.clip(inner / (len_X * len_Y), -1, 1)
    return np.degrees(np.arccos(cos_alpha))
```

`np.sum(X * Y, axis=1)` is the inner product `⟨x, y⟩`, computed for every row at once — the vectorized, per-row version.

---

### The One-Sentence Summary

> Yes — `np.dot(x, y)` (or the `@` operator, or `np.inner`) computes the inner product $\langle x, y \rangle$ for two 1D vectors. But for **row-wise** dot products across a 2D array (each row a vector, as in `vector_angles`), use **`np.sum(X * Y, axis=1)`** — which multiplies matching elements and sums each row into one value — since `np.dot` on 2D arrays does matrix multiplication instead. The `np.sum(X*Y)` form also has the advantage of building the dot product from its definition, matching the exercise's "use np.sum" style. 🎯

## `np.clip()` — Simply

`np.clip()` **forces values to stay within a given range** — anything below the minimum gets raised up to the minimum, anything above the maximum gets pulled down to the maximum. It "clips off" values that stray outside the allowed bounds.

---

### The Basic Behavior

```python
np.clip(value, minimum, maximum)
```

```python
import numpy as np

np.clip(5, 0, 10)      # → 5     (already inside [0,10] → unchanged)
np.clip(-3, 0, 10)      # → 0     (below 0 → raised UP to 0)
np.clip(15, 0, 10)       # → 10    (above 10 → pulled DOWN to 10)
```

Think of it as **"squeeze everything into the range [min, max]"**:

```
        clipped to min          untouched          clipped to max
   ←────────●────────[  min ────────────── max  ]────────●────────→
        -3 becomes 0                            15 becomes 10
```

Works element-wise on whole arrays too:

```python
np.clip([-5, 2, 8, 20], 0, 10)
# → [0, 2, 8, 10]     -5→0, 20→10, the rest unchanged
```

---

### Why It's Needed in `vector_angles`

```python
cos_alpha = np.clip(inner / (len_X * len_Y), -1, 1)
```

The problem it solves: **`cos_alpha` should mathematically always be between −1 and +1** (cosine can never exceed that range). But **floating-point rounding errors** can push it *slightly* outside — like `1.0000000002` or `-1.0000000001`:

```python
# In theory, two identical-direction vectors give cos = exactly 1.0
# But floating-point math might compute:
cos_alpha = 1.0000000002      # ← microscopically over 1, due to rounding!
```

Then when you feed that into `np.arccos`:

```python
np.arccos(1.0000000002)     # → nan!   (arccos is undefined outside [-1, 1])
```

**`arccos` only accepts inputs in `[-1, 1]`** — anything outside returns `nan` (not-a-number), silently breaking your result.

---

### What `clip` Fixes

`np.clip(cos_alpha, -1, 1)` **forces the value back into the valid range** before `arccos` ever sees it:

```python
np.clip(1.0000000002, -1, 1)     # → 1.0    (the tiny overshoot is snapped back to exactly 1)
np.clip(-1.0000000001, -1, 1)     # → -1.0   (snapped back to exactly -1)
```

So the harmless floating-point overshoot gets corrected to the nearest valid boundary, and `arccos` works correctly:

```python
np.arccos(np.clip(1.0000000002, -1, 1))    # → np.arccos(1.0) → 0.0   ✓ works!
```

---

### Demonstrating the Bug It Prevents

```python
import numpy as np

x = np.array([1, 2, 3])
y = np.array([1, 2, 3])     # SAME vector — angle should be exactly 0°

# Without clip:
cos_alpha = np.sum(x*y) / (np.sqrt(np.sum(x**2)) * np.sqrt(np.sum(y**2)))
print(cos_alpha)            # → maybe 1.0000000000000002  (tiny overshoot!)
print(np.arccos(cos_alpha))  # → nan  ✗ BROKEN!

# With clip:
cos_alpha = np.clip(cos_alpha, -1, 1)
print(np.arccos(cos_alpha))   # → 0.0  ✓ correct — the angle is 0
```

Two identical vectors have a **0° angle** — but without `clip`, the tiny rounding error makes `arccos` return `nan` instead of `0`. `clip` is the safeguard that prevents this.

---

### Why This Matters — The Exercise Even Warns About It

Recall the exercise's own hint (which we noted earlier):

> *"you might have to guard against numerical errors... use `np.clip` to force the value into `[-1, 1]`"*

This is **exactly** what `clip` is doing here — it's the recommended fix for the floating-point issue, applied right before `arccos`. It's a defensive measure: mathematically unnecessary (the value *should* already be in range), but practically essential because floating-point arithmetic isn't perfectly exact (remember the `2.675` rounding and `1/3+1/3+1/3 ≠ 1.0` examples from earlier).

---

### The One-Sentence Summary

> `np.clip(value, -1, 1)` forces a value to stay within `[-1, 1]` — raising anything below −1 up to −1, and pulling anything above 1 down to 1. In `vector_angles`, it guards against floating-point rounding errors that can push `cos_alpha` slightly outside `[-1, 1]` (like `1.0000000002`), which would otherwise make `np.arccos` return `nan`; clipping snaps that harmless overshoot back to exactly ±1 so `arccos` computes the correct angle. 🎯
