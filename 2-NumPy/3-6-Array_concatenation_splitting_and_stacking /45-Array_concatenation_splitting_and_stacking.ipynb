### Array concatenation, splitting and stacking

There are two ways of combining several arrays into one bigger array:
1. concatenate and
2. stack.

`Concatenate` takes n-dimensional arrays and returns an n-dimensional array, whereas `stack` takes n-dimensional arrays and returns n+1-dimensional array. Few examples of these:

```python
import numpy as np
```

```python
a=np.arange(2)
b=np.arange(2,5)
```

```python
print(f'a has shape {a.shape}: {a}')
print(f'b has shape {b.shape}: {b}')
```

```python
np.concatenate((a,b))  # concatenating 1-D arrays
```

```python
c=np.arange(1,5).reshape(2,2)
print(f'c has shape {c.shape}: {c}', sep='\n')
d= np.concatenate((c,c))  # concatenating 2-D arrays
print(f'd has shape {d.shape}: {d}', sep='\n')
```

By default concatenate joins the arrays along axis 0. To join the arrays horizontally, add parameter `axis=1`:

<img src="images/image1.png" wideth=1000>

```python
e=np.concatenate((c,c), axis=1)
print(f'e has shape {e.shape}: {e}', sep='\n')
```

If you want to catenate arrays with different dimensions, for example to add a new column to a 2d array, you must first reshape the arrays to have same number of dimensions:

```python
print('New row')
print(np.concatenate((c, a.reshape(1,2))))
```

```python
print('New column')
print(np.concatenate((c, a.reshape()), axis=1))
```

Use `stack` to create higher dimensional arrays from lower dimensional arrays:

```python
print(np.stack((b,b)))
```

```python
print(np.stack((b,b), axis=1))
```

Inverse operation of concatenate is `split`. Its argument specifies either

1. the number of equal parts the array is divided into, or
2. it specifies explicitly the break points.

```python
d=np.arange(12).reshape(6,2)
print('d:\n', d)
d1,d2=np.split(d,2)
print('d1:\n', d1)
print('d2:\n', d2)
```

```python
d=np.arange(12).reshape(2,6)
print('d:\n', d)
parts= np.split(d, (2,3,5), axis=1)
for i, p in enumerate(parts):
    print(f'part {i}\n {p}')
```
